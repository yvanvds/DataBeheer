## Doelen

- Je kan in een schema aanwijzen via welke sleutels twee tabellen samenhangen (SQL3.1)
- Je kan twee tabellen combineren met `INNER JOIN` (SQL3.2)
- Je kan een kolom aanduiden met de naam of de alias van haar tabel, bv. `c.first_name` (SQL3.3)
- Je kan drie of meer tabellen combineren met meerdere JOINs (SQL3.4)
- Je kan gecombineerde gegevens filteren met `WHERE` en sorteren met `ORDER BY` (SQL3.5)
- Je kan met `LEFT JOIN` alle rijen van een tabel tonen, ook zonder match (SQL3.6)
- Je kan de rijen vinden die geen match hebben, bv. klanten zonder bestelling (SQL3.7)
- Je kan een onderzoek in stappen oplossen, met het resultaat van de ene query in de volgende (SQL3.8)
- Je kan rijen samenvatten met `COUNT`, `SUM`, `AVG`, `MIN` en `MAX` (SQL4.1)
- Je kan het verschil tussen `COUNT(*)` en `COUNT(kolom)` uitleggen (SQL4.2)
- Je kan met `WHERE` kiezen welke rijen meetellen in een samenvatting (SQL4.3)
- Je kan een berekening samenvatten, bv. een omzet (SQL4.4)
- Je kan per groep samenvatten met `GROUP BY` (SQL4.5)
- Je kan tabellen eerst combineren en daarna groeperen (SQL4.6)
