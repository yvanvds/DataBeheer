- Je kan kolommen opvragen uit een tabel met `SELECT … FROM` (SQL1.1)

- Je kan voorwaarden combineren met `AND` en `OR` (SQL1.3)
- Je kan een resultaat sorteren met `ORDER BY` (SQL1.4)
- Je kan met `LIMIT` alleen de eerste rijen tonen, bv. een top 5 (SQL1.5)
- Je kan elke waarde maar één keer tonen met `DISTINCT` (SQL1.6)
- Je kan rekenen in een query en een kolom een naam geven met `AS` (SQL1.7)
- Je kan tekstkolommen aan elkaar plakken (SQL1.8)
- Je kan in het schema van een onbekende databank de juiste tabel en kolommen vinden (SQL1.9)

- Je kan tekst zoeken op een patroon met `LIKE` (SQL2.2)
- Je kan filteren op een lijst van waarden met `IN` (SQL2.3)
- Je kan filteren op een bereik met `BETWEEN` (SQL2.4)
- Je kan filteren op een datum (SQL2.5)
- Je kan `WHERE`, `ORDER BY` en `LIMIT` in de juiste volgorde in één query zetten (SQL2.6)

## Vraag 9

Welke query toont de naam en de prijs van alle oortjes (`category_id` 7) en speakers (`category_id` 8) die minder dan 100 euro kosten, de goedkoopste eerst?

````sql
`SELECT name, price FROM products WHERE category_id = 7 OR category_id = 8 AND price < 100 ORDER BY price;`
`SELECT name, price FROM products ORDER BY price WHERE (category_id = 7 OR category_id = 8) AND price < 100;`
`SELECT name, price FROM products WHERE (category_id = 7 OR category_id = 8) AND price < 100 ORDER BY price;`
`SELECT name, price FROM products WHERE category_id = 7 AND category_id = 8 AND price < 100 ORDER BY price;`
````
