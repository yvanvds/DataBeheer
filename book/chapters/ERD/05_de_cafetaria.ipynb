# 5. Van gegevens naar structuur

In deze les werken we met een eenvoudige maar herkenbare situatie:
een **schoolcafetaria waar leerlingen broodjes kunnen bestellen**.

We gebruiken dit voorbeeld om opnieuw na te denken over **hoe je gegevens ordent** in een database.

## Fase 1 — Welke gegevens willen we bijhouden

We starten **niet** met tabellen of SQL, maar met de vraag:

> *Welke informatie hebben we nodig om broodjesbestellingen correct te kunnen opvolgen?*

Voor deze oefening beperken we ons bewust tot de volgende gegevens.

### Leerlingen

Van elke leerling willen we weten:

* naam
* klas

### Producten (broodjes)

Van elk product willen we weten:

* naam
* prijs

### Bestellingen

Van elke bestelling willen we weten:

* wie de bestelling plaatst
* wanneer er besteld wordt
* wat de status is van de bestelling
  (bijvoorbeeld: *open*, *betaald*, *geannuleerd*)

### Bestelde producten

Bij een bestelling willen we ook weten:

* welke producten besteld zijn
* hoeveel stuks van elk product

⚠️ Belangrijk
Dit is **inhoudelijke informatie**:
dit zijn de dingen die we *willen weten*, niet hoe we ze technisch opslaan.


## Fase 2 — Wat hoort samen? (structuur zoeken)

Nu stellen we ons een andere vraag:

> *Welke gegevens horen logisch samen, en welke niet?*

### Stap 1: groeperen

* Gegevens over **leerlingen** horen samen
* Gegevens over **producten** horen samen
* Een **bestelling** is iets anders dan een leerling of een product
* Eén bestelling kan **meerdere producten** bevatten

Dat laatste is belangrijk:

* een bestelling is niet één product
* en een product hoort niet bij één vaste bestelling

### Stap 2: logisch gevolg

Als we alles in één tabel zouden zetten:

* dan zouden leerlinggegevens telkens opnieuw voorkomen
* productnamen en prijzen zouden herhaald worden
* een wijziging (bv. prijs aanpassen) zou op meerdere plaatsen nodig zijn

Daarom splitsen we de gegevens logisch op.

### Resultaat van deze redenering

We komen uit op vier duidelijk verschillende groepen gegevens:

1. **Leerlingen**

   * naam
   * klas

2. **Producten**

   * naam
   * prijs

3. **Bestellingen**

   * wanneer besteld
   * status
   * door welke leerling

4. **Bestellijnen**

   * bij welke bestelling
   * welk product
   * aantal

Dit is **normalisatie**, ook al gebruiken we het woord niet expliciet:

* we zorgen ervoor dat elke soort informatie maar op één plaats voorkomt
* en dat herhaling vermeden wordt

## Fase 3 — Technische vastlegging (kort)

Tot nu toe hebben we enkel **inhoudelijk** nagedacht.
Om hiermee te kunnen werken in een database, moeten we dit technisch vastleggen.

### Een noodzakelijke afspraak

In een database:

* krijgt **elke tabel** een eigen identificatieveld (id)
* dat id dient om rijen **betrouwbaar te kunnen aanduiden en koppelen**

Dit is **technisch**, geen inhoudelijke informatie.
We staan hier niet verder bij stil: dit kennen jullie al uit eerdere lessen.

Concreet betekent dit:

* elke leerling krijgt een eigen id
* elk product krijgt een eigen id
* elke bestelling krijgt een eigen id
* bestellijnen verwijzen naar:

  * de juiste bestelling
  * het juiste product

Vanaf nu veronderstellen we dus:

* dat elke tabel een id heeft
* dat we die id’s gebruiken om tabellen met elkaar te verbinden

## Opdracht 3 — Toon met SELECT waarom dit later problemen geeft

Je gebruikt nu `SELECT`-queries om te bewijzen dat de database “foute” situaties toelaat.

### 3.1 Overzicht: bestellingen met leerlingnaam

```sql
SELECT
  datum,
  status,
  naam,
  klas
FROM bestellingen 
LEFT JOIN leerlingen 
  ON leerlingen.leerling_id = bestellingen.leerling_id
ORDER BY bestelling_id;
```

**Wat moet je zien?**

* Bestelling met `leerling_id = 999` krijgt geen naam/klas (NULL).
  Dat betekent: *er bestaat een bestelling “voor iemand die niet bestaat”*.

### 3.2 Zoek bestellijnen met een bestelling die niet bestaat

```sql
SELECT 
	bestellingen.bestelling_id,
	bestellingen.datum,
	bestelling_lijnen.aantal,
	bestelling_lijnen.product_id
FROM bestelling_lijnen
LEFT JOIN bestellingen
  ON bestellingen.bestelling_id = bestelling_lijnen.bestelling_id
;
```

### 3.3 Zoek bestellijnen met een product dat niet bestaat

```sql
SELECT bl.*
FROM bestelling_lijnen bl
LEFT JOIN producten p
  ON p.product_id = bl.product_id
WHERE p.product_id IS NULL;
```

### 3.4 Zoek “slechte” waarden die we later willen verbieden

Negatieve prijs:

```sql
SELECT *
FROM producten
WHERE prijs < 0;
```

Aantallen die niet logisch zijn:

```sql
SELECT *
FROM bestelling_lijnen
WHERE aantal <= 0;
```

Status die we niet willen:

```sql
SELECT *
FROM bestellingen
WHERE status NOT IN ('open', 'betaald', 'geannuleerd');
```

### 3. (Extra) Bonnetje per bestelling: waar gaat het mis?

```sql
SELECT
  bestellingen.bestelling_id,
  leerlingen.naam AS leerling,
  producten.naam AS product,
  bestelling_lijnen.aantal,
  producten.prijs,
  (bestelling_lijnen.aantal * producten.prijs) AS lijn_totaal
FROM bestelling_lijnen 
LEFT JOIN bestellingen ON bestellingen.bestelling_id = bestelling_lijnen.bestelling_id
LEFT JOIN leerlingen   ON leerlingen.leerling_id = bestellingen.leerling_id
LEFT JOIN producten    ON producten.product_id = bestelling_lijnen.product_id
ORDER BY bestellingen.bestelling_id, bestelling_lijnen.bestellijn_id;
```

**Wat moet je zien?**

* Waar een bestelling of product ontbreekt, krijg je NULL’s.
* Je kan geen betrouwbaar “bonnetje” maken als verwijzingen niet kloppen.
* Een negatieve prijs of aantal 0 geeft een totaal dat inhoudelijk fout is.

* Je kan een database maken die “werkt” (SQL faalt niet), maar toch **foute situaties** toelaat.
* Zonder constraints en foreign keys kan je:

In de volgende fase gaan we precies deze problemen oplossen met **constraints** en daarna met **ON DELETE**-acties bij foreign keys.



# Fase 5 — Constraints: regels afdwingen in de database

In fase 4 zagen we dat de database technisch “werkt”, maar inhoudelijk foute situaties toelaat (negatieve prijzen, aantal 0, vreemde status, …).
In fase 5 lossen we dat op door **constraints** toe te voegen: regels die de database automatisch afdwingt.


## 5.1 Welke mogelijkheden hebben we om restricties te forceren?

Een database kan op verschillende manieren regels afdwingen. Dit zijn de basisopties die je nu moet kennen.

### 1) `NOT NULL`

**Wat doet het?**
Verplicht dat een kolom altijd een waarde bevat.

**Voorbeeld**

```sql
CREATE TABLE demo (
  naam TEXT NOT NULL
);

INSERT INTO demo (naam) VALUES (NULL);  -- faalt
```


### 2) `UNIQUE`

**Wat doet het?**
Zorgt dat een waarde niet twee keer kan voorkomen in een kolom.

**Voorbeeld**

```sql
CREATE TABLE demo (
  email TEXT UNIQUE
);

INSERT INTO demo (email) VALUES ('a@school.be');
INSERT INTO demo (email) VALUES ('a@school.be');  -- faalt
```


### 3) `CHECK(...)`

**Wat doet het?**
Controleert een voorwaarde. Als de voorwaarde niet klopt, faalt de `INSERT` of `UPDATE`.

**Voorbeeld**

```sql
CREATE TABLE demo (
  leeftijd INTEGER CHECK(leeftijd >= 0)
);

INSERT INTO demo (leeftijd) VALUES (-5); -- faalt
```

Je kan dit ook met tekst doen. Zo geef je aan dat een tekst veld bijvoorbeeld enkel `ja` of `nee` mag zijn.

```sql
CREATE TABLE demo (
  antwoord TEXT CHECK(antwoord IN ('ja', 'nee')
);

INSERT INTO demo (antwoord) VALUES ('misschien'); -- faalt
```



### 4) `DEFAULT`

**Wat doet het?**
Geeft een standaardwaarde als je niets invult.

**Voorbeeld**

```sql
CREATE TABLE demo (
  status TEXT DEFAULT 'open'
);

INSERT INTO demo DEFAULT VALUES;  -- status wordt automatisch 'open'
```


### 5) `PRIMARY KEY` (herhaling, maar als constraint)

**Wat doet het?**
Maakt een rij uniek identificeerbaar en voorkomt dubbele sleutels.

**Voorbeeld**

```sql
CREATE TABLE demo (
  id INTEGER PRIMARY KEY
);
```

> Opmerking: foreign keys behandelen we pas in fase 6. In fase 5 beperken we ons tot constraints die binnen één tabel kunnen werken.

# Fase 6 — Foreign keys en ON DELETE-acties

In fase 5 hebben we al veel fouten kunnen uitsluiten met constraints binnen één tabel.
Toch blijft er een fundamenteel probleem bestaan:

> De database controleert nog niet of verwijzingen **naar andere tabellen** kloppen.

In deze fase lossen we dat op met **foreign keys** en bepalen we wat er moet gebeuren **als gegevens verwijderd worden**.

---

## 6.1 Wat is een foreign key?

Een **foreign key** is een regel die zegt:

> *De waarde in deze kolom moet bestaan als primaire sleutel in een andere tabel.*

Zo kan de database zelf controleren:

* of een bestelling verwijst naar een **bestaande leerling**
* of een bestellijn verwijst naar een **bestaande bestelling**
* of een bestellijn verwijst naar een **bestaand product**

---

## 6.2 Foreign keys inschakelen in SQLite

In SQLite zijn foreign keys **niet automatisch actief**.
Je moet dit expliciet inschakelen.

```sql
PRAGMA foreign_keys = ON;
```

> Doe dit **altijd** vóór je foreign keys test.
> Zonder deze regel worden foreign keys gewoon genegeerd.

---

## 6.3 ON DELETE-acties: wat moet er gebeuren bij verwijderen?

Wanneer een rij in een tabel wordt verwijderd, kan dat gevolgen hebben voor andere tabellen die ernaar verwijzen.

Daarom moet je bij een foreign key **bewust kiezen** wat er gebeurt bij `DELETE`.

### 1) `ON DELETE RESTRICT`

* Verwijderen wordt geweigerd als er nog verwijzingen bestaan.

**Betekenis**

> “Je mag dit niet verwijderen zolang het nog gebruikt wordt.”

**Voorbeeld**

```sql
FOREIGN KEY (leerling_id)
REFERENCES leerlingen(leerling_id)
ON DELETE RESTRICT
```

---

### 2) `ON DELETE CASCADE`

* Verwijderen werkt door naar afhankelijke rijen.

**Betekenis**

> “Als dit verdwijnt, mag alles wat erbij hoort ook weg.”

**Voorbeeld**

```sql
FOREIGN KEY (bestelling_id)
REFERENCES bestellingen(bestelling_id)
ON DELETE CASCADE
```

---

### 3) `ON DELETE SET NULL`

* De verwijzing wordt leeggemaakt (`NULL`).

**Betekenis**

> “De koppeling verdwijnt, maar de rij blijft bestaan.”

Deze optie gebruiken we hier **niet**, omdat een bestellijn zonder bestelling of product geen zin heeft.

---

## 6.4 Welke keuzes maken we in de cafetaria-database?

We maken nu **bewuste inhoudelijke keuzes**.

### Bestellingen → leerlingen

* Een bestelling hoort bij een leerling.
* Een leerling met bestellingen mag je niet zomaar verwijderen.

Welke optie kies je hier?
➡️ **ON DELETE RESTRICT**
➡️ **ON DELETE CASCADE**

---

### Bestellijnen → bestellingen

* Bestellijnen bestaan enkel als onderdeel van een bestelling.
* Als een bestelling wordt verwijderd, mogen de lijnen mee verdwijnen.

Welke optie kies je hier?
➡️ **ON DELETE RESTRICT**
➡️ **ON DELETE CASCADE**

---

### Bestellijnen → producten

* Een product dat ooit besteld werd, mag je niet verwijderen.
* Anders zouden oude bestellingen onvolledig worden.

Welke optie kies je hier?
➡️ **ON DELETE RESTRICT**
➡️ **ON DELETE CASCADE**

---

## 6.5 Tabellen opnieuw aanmaken met foreign keys

Net zoals bij constraints:

* we droppen eerst de tabellen
* daarna maken we ze opnieuw aan, nu mét foreign keys

### Stap 1 — Tabellen verwijderen

```sql
DROP TABLE bestelling_lijnen;
DROP TABLE bestellingen;
```
De tabel `leerlingen` en `producten` kunnen we behouden. Die wijzigt niet.
---

### Stap 2 — Tabellen opnieuw maken

```sql
CREATE TABLE bestellingen (
  bestelling_id INTEGER PRIMARY KEY,
  datum TEXT NOT NULL,
  status TEXT NOT NULL CHECK (status IN ('open', 'betaald', 'geannuleerd')),
  leerling_id INTEGER NOT NULL,
  FOREIGN KEY (leerling_id)
    REFERENCES leerlingen(leerling_id)
    ON DELETE RESTRICT
);
```

```sql
CREATE TABLE bestelling_lijnen (
  bestellijn_id INTEGER PRIMARY KEY,
  bestelling_id INTEGER NOT NULL,
  product_id INTEGER NOT NULL,
  aantal INTEGER NOT NULL CHECK (aantal > 0),
  FOREIGN KEY (bestelling_id)
    REFERENCES bestellingen(bestelling_id)
    ON DELETE CASCADE,
  FOREIGN KEY (product_id)
    REFERENCES producten(product_id)
    ON DELETE RESTRICT
);
```

## Wat je moet onthouden uit fase 6

* Foreign keys zorgen ervoor dat tabellen **niet los van elkaar kunnen liegen**
* `PRAGMA foreign_keys = ON` is noodzakelijk in SQLite
* `ON DELETE`-acties zijn **inhoudelijke keuzes**, geen technische details
* Constraints + foreign keys samen maken van een “vrije” database
  een **betrouwbaar systeem**

Hiermee is de structuur van de database inhoudelijk correct en technisch beschermd.

## 7.2 Controle: overzicht met SELECT (bonnetje)

Gebruik nu één overzichtsquery om te controleren of alles logisch samenkomt.

```sql
SELECT
  bestellingen.bestelling_id,
  leerlingen.naam AS leerling,
  producten.naam AS product,
  bestelling_lijnen.aantal,
  producten.prijs,
  (bestelling_lijnen.aantal * producten.prijs) AS lijn_totaal
FROM bestelling_lijnen 
LEFT JOIN bestellingen ON bestellingen.bestelling_id = bestelling_lijnen.bestelling_id
LEFT JOIN leerlingen   ON leerlingen.leerling_id = bestellingen.leerling_id
LEFT JOIN producten    ON producten.product_id = bestelling_lijnen.product_id
ORDER BY bestellingen.bestelling_id, bestelling_lijnen.bestellijn_id;
```

**Wat moet je zien?**

* geen `NULL`-waarden meer
* correcte producten bij elke bestelling
* een logisch totaal per lijn

Dit toont dat:

* de structuur klopt
* foreign keys correct werken
* constraints geen geldige data blokkeren

## 7.4 DELETE — wat mag en wat mag niet?

### 1) Bestelling verwijderen (cascade)

```sql
DELETE FROM bestellingen
WHERE bestelling_id = 1;
```

Controleer:

```sql
SELECT *
FROM bestelling_lijnen;
```

➡️ De bijhorende bestellijnen zijn automatisch verdwenen
(**ON DELETE CASCADE**).

---

### 2) Product verwijderen dat nog gebruikt wordt

```sql
DELETE FROM producten
WHERE product_id = 2;
```

➡️ Deze `DELETE` faalt
(**ON DELETE RESTRICT**).

---

### 3) Leerling zonder bestellingen verwijderen

```sql
DELETE FROM leerlingen
WHERE leerling_id = 3;
```

➡️ Dit lukt, want deze leerling heeft geen bestellingen.

---

### 4) Leerling mét bestellingen verwijderen

```sql
DELETE FROM leerlingen
WHERE leerling_id = 2;
```

➡️ Deze `DELETE` faalt
(**ON DELETE RESTRICT**).

## 7.6 Eindconclusie

In deze les heb je gezien dat:
