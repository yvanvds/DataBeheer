## 1 Wat is een ERD?

Een **ERD** (Entity Relationship Diagram) is een schema dat in één oogopslag toont:

* welke **entiteiten** er zijn: de "dingen" waarover je gegevens bijhoudt
* welke **attributen** elke entiteit heeft: de gegevens zelf
* welke **relaties** er tussen de entiteiten bestaan
* de **cardinaliteit** van elke relatie: hoeveel-op-hoeveel

Vergelijk het met een bouwplan. Een architect tekent eerst een plan vóór er één steen gemetseld wordt, zodat iedereen — de klant, de aannemer, de elektricien — over hetzelfde plan kan discussiëren. Een ERD doet precies hetzelfde voor een databank: je bespreekt en verbetert het **ontwerp** vóór je één regel `CREATE TABLE` schrijft.

Waarom niet meteen in SQL?

* Een tekening overzie je in één oogopslag, twintig `CREATE TABLE`-instructies niet.
* Een fout in een tekening verbeter je in enkele seconden; een verkeerde tabelstructuur achteraf rechtzetten is veel meer werk (denk aan het droppen en heropbouwen van tabellen in hoofdstuk 2).
* Je kan een ERD bespreken met mensen die geen SQL kennen: je opdrachtgever, de uitbater van de cafetaria, het bestuur van de voetbalclub.

### Entiteiten en attributen

Een **entiteit** is iets waarvan je gegevens wil bijhouden: een klant, een product, een bestelling, een speler. In een ERD teken je elke entiteit als een rechthoek met:

* bovenaan de **naam** van de entiteit
* daaronder de **attributen**: de gegevens die je over de entiteit bijhoudt
* een aanduiding van de **identificator**: het attribuut dat elke rij uniek maakt — in de databank wordt dat de primaire sleutel (PK)

Herken je dit? Het zijn de tabellen, kolommen en sleutels uit hoofdstuk 1. Het ERD is de tekening, de tabellen zijn de uitvoering.

## 2 Relaties en cardinaliteit (crow's foot)

Tussen entiteiten teken je lijnen: de **relaties**. Maar een lijn alleen zegt niet genoeg. Bij "een klant plaatst een bestelling" wil je ook weten:

* Hoeveel bestellingen kan één klant plaatsen?
* Bij hoeveel klanten kan één bestelling horen?

Dat noemen we de **cardinaliteit** van de relatie. Wij noteren ze in **crow's foot-notatie**: de vorm van elk lijnuiteinde toont hoeveel rijen er aan die kant kunnen meedoen. Het "veel"-uiteinde lijkt op een kraaienpoot (Engels: *crow's foot*) — vandaar de naam.

![Crow's foot-notatie: de vier lijnuiteinden en een voorbeeld met klassen en leerlingen](../../figures/erd_crowsfoot_notatie.svg)

Een relatie lees je altijd in **twee richtingen**. In het voorbeeld hierboven:

* één klas telt **één of meer** leerlingen
* elke leerling zit in **precies één** klas

### De drie soorten relaties

| Soort | Voorbeeld | In het ERD |
|-------|-----------|------------|
| één-op-één (1–1) | elke leerling huurt hoogstens één kastje | streepje aan beide uiteinden |
| één-op-veel (1–n) | één klant plaatst veel bestellingen | kraaienpoot aan de veel-kant |
| veel-op-veel (n–m) | één bestelling bevat veel producten én één product zit in veel bestellingen | teken je nooit rechtstreeks: koppeltabel! |

### Veel-op-veel? Koppeltabel!

In hoofdstuk 2 zag je al dat een veel-op-veel relatie niet rechtstreeks in tabellen past: je lost ze op met een **koppeltabel** (zoals `order_items`). In een ERD doe je exact hetzelfde: je tekent de koppeltabel als een extra entiteit in het midden, met aan beide kanten een één-op-veel relatie. De twee kraaienpoten wijzen dan allebei naar de koppeltabel.

## 3 De webshop als ERD

Je kent **webshop.db** ondertussen door en door. Zo ziet haar ontwerp eruit als ERD in crow's foot-notatie:

![ERD van de webshop-database in crow's foot-notatie](../../figures/webshop_erd.svg)

Lees het diagram hardop, telkens in twee richtingen:

* één klant plaatst **nul of meer** bestellingen; elke bestelling hoort bij **precies één** klant
* één bestelling bevat **één of meer** orderlijnen; elke orderlijn hoort bij **precies één** bestelling
* één product komt voor in **nul of meer** orderlijnen; elke orderlijn gaat over **precies één** product

Zie je het? `order_lines` is de koppeltabel die de veel-op-veel relatie tussen bestellingen en producten oplost — met twee kraaienpoten die ernaar wijzen.

:::{note}
Waarom staat er aan de kant van `orders` een cirkeltje (nul of meer) en aan de kant van `order_lines` een streepje (één of meer)? Denk na: kan een klant bestaan zonder bestellingen? Kan een bestelling bestaan zonder orderlijnen?
:::

## 4 Tekenen met dbdiagram.io

Een ERD schetsen op papier blijft handig voor een eerste ruwe versie. Maar voor een verzorgd, aanpasbaar diagram gebruiken we **[dbdiagram.io](https://dbdiagram.io)**:

* gratis en volledig in de browser — niets te installeren
* je **typt** je schema in plaats van vakjes te slepen: snel, ook op een trage laptop
* het diagram wordt automatisch getekend terwijl je typt
* je kan je schema achteraf exporteren als SQL-code (daarover straks meer)

Ga naar [dbdiagram.io](https://dbdiagram.io) en klik op **Create your diagram**. Je krijgt een voorbeeldschema te zien: verwijder die code en typ onderstaand schema over.

💡 Je kan meteen tekenen zonder account. Wil je je diagram bewaren of exporteren, maak dan een gratis account aan.

```text
Table klanten {
  klant_id integer [pk]
  naam varchar
  email varchar
}

Table bestellingen {
  bestelling_id integer [pk]
  datum date
  klant_id integer
}

Ref: bestellingen.klant_id > klanten.klant_id
```

Wat betekent dit?

* elk `Table`-blok wordt een entiteit met haar attributen
* `[pk]` markeert de primaire sleutel
* `integer`, `varchar` (tekst), `date` en `decimal` zijn de datatypes
* een `Ref:`-regel tekent een relatie; het teken in het midden bepaalt de cardinaliteit:

| Teken | Betekenis | Voorbeeld |
|-------|-----------|-----------|
| `>` | veel-op-één | `bestellingen.klant_id > klanten.klant_id` |
| `<` | één-op-veel | `klanten.klant_id < bestellingen.klant_id` |
| `-` | één-op-één | `leerlingen.leerling_id - kastjes.leerling_id` |

:::{note}
dbdiagram.io zet geen kraaienpoten op de lijnen, maar een `1` en een `*` (veel). De betekenis is exact dezelfde, alleen de tekenstijl verschilt. Op papier en op een toets gebruik je crow's foot; in dbdiagram lees je `1` en `*`.
:::

dbdiagram kent ook het teken `<>` voor veel-op-veel, maar dat gebruiken we **niet**: wij tekenen de koppeltabel altijd zelf. Zo zie je meteen welke tabel je straks moet aanmaken — en welke extra attributen (zoals `aantal`) bij de koppeltabel horen.

Wil je liever échte kraaienpoten tekenen door vormen te slepen? Dat kan gratis in [draw.io](https://app.diagrams.net) (zoek de vormen onder *Entity Relation*). Voor deze cursus raden we dbdiagram.io aan, omdat het sneller werkt en je schema meteen kan omzetten naar SQL.

## 5 Van ERD naar tabellen

Een ERD is een ontwerp, nog geen databank. De laatste stap is de **vertaling** van het diagram naar tabellen. Daarvoor bestaan vaste regels:

| In het ERD | In de databank |
|------------|----------------|
| entiteit | tabel |
| attribuut | kolom |
| identificator | primaire sleutel (PK) |
| één-op-veel relatie | vreemde sleutel (FK) aan de **veel**-kant |
| veel-op-veel relatie | koppeltabel met twee FK's |
| één-op-één relatie | FK met `UNIQUE` aan één van beide kanten |

De belangrijkste regel onthoud je zo: **de vreemde sleutel staat altijd aan de kant van de kraaienpoot.** Elke bestelling hoort bij precies één klant, dus in `bestellingen` volstaat één kolom `klant_id`. Omgekeerd zou je in `klanten` een hele lijst bestelnummers in één veld moeten proppen — en dat is precies wat we in hoofdstuk 2 leerden vermijden.

### Voorbeeld

Het fragment `klanten` (precies één) — `bestellingen` (nul of veel) uit het schema hierboven vertaal je zo:

```sql
CREATE TABLE klanten (
    klant_id INTEGER PRIMARY KEY,
    naam TEXT NOT NULL,
    email TEXT UNIQUE
);

CREATE TABLE bestellingen (
    bestelling_id INTEGER PRIMARY KEY,
    datum TEXT NOT NULL,
    klant_id INTEGER NOT NULL,
    FOREIGN KEY (klant_id) REFERENCES klanten(klant_id)
);
```

### dbdiagram doet het (bijna) voor je

Klik in dbdiagram.io op **Export** en kies bijvoorbeeld *Export to PostgreSQL*. Je krijgt kant-en-klare `CREATE TABLE`-code: de vertaalregels hierboven, automatisch toegepast. De code is niet altijd letterlijk bruikbaar in SQLite — datatypes heten er soms anders — maar de structuur (tabellen, PK's, FK's) herken je meteen.

Een goed ERD vertaalt zich dus bijna automatisch naar een databank. En andersom geldt ook: kan je je ERD niet vlot vertalen, dan is het ontwerp nog niet af.

### Oefening 2 — De schoolbibliotheek

De bibliotheek van de school wil een databank. Dit is wat ze willen bijhouden:

* van elk **boek**: de titel en de auteur
* van elke **leerling**: de naam en de klas
* welke leerling welk boek **ontleende**, wanneer het werd meegenomen en wanneer het werd teruggebracht

Beantwoord eerst op papier:

1. Welke entiteiten herken je?
2. Welke relatie is veel-op-veel? Hoe los je dat op?
3. Bij welke entiteit horen de datums van ontlenen en terugbrengen?

Teken daarna het ERD in dbdiagram.io.

:::{admonition} Hint
:class: tip dropdown
Een leerling kan veel boeken ontlenen, en hetzelfde boek wordt door veel leerlingen ontleend. Je hebt dus een koppeltabel nodig — noem ze bijvoorbeeld `ontleningen`. De datums horen bij de ontlening zelf: niet bij het boek en niet bij de leerling.
:::
