### Tabellen maken (voorbeeld)

Probeer het voorbeeld uit met **Run alles** en bekijk het resultaat via **Schema**.


```python
CREATE TABLE spelers (
    speler_id INTEGER PRIMARY KEY,
    naam TEXT,
    geboortedatum TEXT
);
```

Belangrijkste types:

* `INTEGER`
* `TEXT`

Belangrijkste sleutel:

* `PRIMARY KEY` — een `INTEGER PRIMARY KEY` krijgt vanzelf een oplopend nummer als je er bij `INSERT` geen meegeeft.

Een vreemde sleutel schrijf je zoals in hoofdstuk 3: `FOREIGN KEY (ploeg_id) REFERENCES ploegen(ploeg_id)`. Of de databank die verwijzing ook echt controleert, onderzoek je in hoofdstuk 5.

### Gegevens toevoegen


```python
INSERT INTO spelers (naam, geboortedatum)
VALUES ('Jan Peeters', '2007-03-12');
```

### Gegevens aanpassen


```python
UPDATE spelers
SET naam = 'J. Peeters'
WHERE speler_id = 1;
```

### Tabel verwijderen

📌 **Let op**: dit verwijdert de tabel én alles wat erin zit.
