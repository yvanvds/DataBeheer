## Stap 1 — Welke gegevens wil je onthouden?

### Opdracht

Stel je voor dat je een databank ontwerpt voor een voetbalclub.

Beantwoord samen:

> **Welke gegevens zou een voetbalclub willen bijhouden?**

Denk aan:

* spelers
* ploegen
* trainers
* wedstrijden
* prestaties

📌 **Nog niet denken aan databanken of relaties.**
Schrijf gewoon *wat* je zou willen onthouden.


## Stap 2 — Zet alles in tabellen (zonder verbeteren)

### Opdracht (in duo)

Maak voor elk “soort ding” een **tabel** met kolommen.

Bijvoorbeeld (dit is slechts een voorbeeld):

**Spelers**

* naam
* geboortedatum
* rugnummer

**Ploegen**

* naam
* leeftijdscategorie

Gebruik:

* papier
* of een Excel-blad
* of een tekstbestand


## Stap 3 — Concrete scenario’s om structuurproblemen bloot te leggen

### A. Wijzigingen

* **De trainer van ploeg U17 verandert midden het seizoen.**

  * In welke tabel(len) moet je dat aanpassen?
  * Hoeveel rijen moet je wijzigen?

* **De naam van een ploeg verandert.**

  * Wat moet je aanpassen?
  * Zie je ergens nog de oude naam staan?

* **Een speler verandert van rugnummer.**

  * Waar staat dat nummer?
  * Kan dat nummer op meerdere plaatsen voorkomen?

* **De trainer verandert zijn telefoonnummer.**

  * Hoeveel keer komt dat nummer voor?
  * Wat als één rij niet aangepast wordt?

### B. Nieuwe gegevens toevoegen

* **Een nieuwe speler schrijft zich in, maar heeft nog geen wedstrijd gespeeld.**

  * Kan je die speler al toevoegen?
  * Waar zet je die informatie?

* **Een nieuwe ploeg wordt opgericht voor volgend seizoen.**

  * Kan je die ploeg al toevoegen?
  * Moet je dan ook al spelers invullen?

* **Er is een wedstrijd gepland, maar de opstelling is nog niet gekend.**

  * Kan je de wedstrijd al opslaan?
  * Wat laat je leeg?


### C. Meerdere keren hetzelfde

* **Een speler speelt drie wedstrijden in één maand.**

  * Hoe noteer je dat?
  * Komt zijn naam drie keer voor?

* **Een speler scoort in twee verschillende wedstrijden.**

  * Waar zet je die doelpunten?
  * Kan je nog zien in welke wedstrijd ze vielen?

* **Twee spelers hebben exact dezelfde naam.**

  * Kan je ze uit elkaar houden?
  * Waar loopt dit fout?


### D. Tijd en verandering

* **Een speler speelt dit seizoen bij ploeg A, volgend seizoen bij ploeg B.**

  * Hoe noteer je dat?
  * Blijft zijn oude informatie bewaard?

* **Een trainer traint eerst ploeg U15 en later U17.**

  * Verlies je de oude situatie?
  * Kan je zien wie hij vorig seizoen trainde?

### E. Wedstrijden & prestaties

* **Een speler zit op de bank en speelt 0 minuten.**

  * Komt hij voor in je tabellen?
  * Hoe noteer je dat?

* **Een speler krijgt een gele kaart maar scoort niet.**

  * Waar noteer je dat?
  * Is dat een probleem als doelpunten en kaarten samen staan?

* **Een speler scoort 2 doelpunten in wedstrijd X en 0 in wedstrijd Y.**

  * Hoe onderscheid je die twee situaties?


### F. Controle en overzicht

* **Je wil weten hoeveel wedstrijden elke speler speelde.**

  * Kan je dat tellen zonder fouten?
  * Tel je misschien dubbel?

* **Je wil weten hoeveel doelpunten ploeg X maakte dit seizoen.**

  * Staat die info duidelijk bij elkaar?

* **Je wil alle wedstrijden van één speler tonen.**

  * Moet je zoeken in meerdere tabellen of rijen?


## Stap 4 — Verbeter je structuur

### Opdracht (in duo)

Werk nu de echte structuur van je database uit. Pas op voor dubbele gegevens. Denk naar over koppelingen via ID's. Duid Primary keys en foreign keys aan.


## Stap 5 — Denken vanuit vragen

Een databank is pas goed als je er vragen aan kan stellen.

### Opdracht (in duo)

Beantwoord deze vragen:

* Welke spelers zitten in ploeg X?
* Hoeveel wedstrijden speelde speler Y?
* Hoeveel doelpunten maakte elke speler?
* Welke wedstrijden speelde ploeg Z?

En telkens:

> *“Kan je dit makkelijk beantwoorden met jullie tabellen?”*

Deel twee:

Bedenk 5 andere vragen die je database zou moeten kunnen beantwoorden. Kan je die informatie uit je database halen?

Waar het moeilijk wordt:

* ontbreekt een tabel
* staat info op de verkeerde plaats
* is er te veel herhaling

## Stap 6 — Teken het ERD

### Opdracht (in duo)

Jullie structuur staat op punt. Tijd om ze vast te leggen zoals een professionele ontwerper dat doet: als een **ERD**.

Teken het ERD van jullie voetbaldatabase in [dbdiagram.io](https://dbdiagram.io), zoals je leerde in hoofdstuk 3:

* maak voor elke tabel een `Table`-blok met haar kolommen
* markeer elke primaire sleutel met `[pk]`
* teken elke relatie met een `Ref:`-regel en denk na over de richting (`>` betekent veel-op-één)
* zit er ergens een veel-op-veel relatie in (bijvoorbeeld tussen spelers en wedstrijden)? Los ze op met een koppeltabel

Controleer daarna jullie diagram:

* Lees elke relatie hardop in twee richtingen ("één ploeg telt veel spelers, elke speler speelt bij ..."). Klopt de cardinaliteit?
* Staat elke vreemde sleutel aan de veel-kant?
* Kan het ERD alle vragen uit stap 5 beantwoorden?

📌 Laat je ERD goedkeuren door je leerkracht vóór je verder gaat naar stap 7.

## Stap 7 — Van ERD naar databank

Nu vertalen we het ERD naar een echte databank, met de vertaalregels uit hoofdstuk 3: elke entiteit wordt een tabel, elke identificator een `PRIMARY KEY` en elke relatie een vreemde sleutel aan de veel-kant.

We gebruiken:

* **DB Browser for SQLite**
* een lege database


### Tabellen maken (voorbeeld)

```sql
CREATE TABLE spelers (
    speler_id INTEGER PRIMARY KEY,
    naam TEXT,
    geboortedatum TEXT
);
```

Belangrijkste types:

* `INTEGER`
* `TEXT`

Belangrijkste sleutel:

* `PRIMARY KEY`

### Gegevens toevoegen

```sql
INSERT INTO spelers (naam, geboortedatum)
VALUES ('Jan Peeters', '2007-03-12');
```

### Gegevens aanpassen

```sql
UPDATE spelers
SET naam = 'J. Peeters'
WHERE speler_id = 1;
```

### Tabel verwijderen

```sql
DROP TABLE spelers;
```

📌 **Let op**: dit verwijdert alles.


## Samenvatting

In dit hoofdstuk heb je geleerd dat:

* een databank start bij **nadenken over gegevens**
* structuur ontstaat door **vragen te stellen**
* herhaling leidt tot problemen
* relaties niet altijd simpel zijn
* tabellen logisch moeten zijn, niet toevallig
* een ERD je ontwerp vastlegt vóór je begint te bouwen
* SQL slechts de **technische vertaling** is van een goed ontwerp
