# 4. Liegen met grafieken

In de vorige les maakte je je eerste grafieken met echte Eurostat-data. Zo'n grafiek voelt objectief aan: het zijn tenslotte gewoon de cijfers. Toch kan je met exact dezelfde cijfers heel verschillende indrukken wekken. Soms gebeurt dat onbedoeld, door een onhandige keuze. Maar het gebeurt ook bewust: om een product te verkopen, een beleid te verdedigen of desinformatie te verspreiden.

Denk terug aan de **5 V's** uit het eerste hoofdstuk. Bij **veracity** leerde je dat je kritisch moet kijken naar de bron en de meetmethode van data. Deze les voegt daar iets belangrijks aan toe: zelfs wanneer de cijfers volledig kloppen, kan de **grafiek** nog liegen. Wie data wil begrijpen, moet dus niet alleen tabellen leren lezen, maar ook visualisaties leren wantrouwen.

:::{admonition} Leerdoelen
:class: tip

Na deze les kan je:
- zes veelgebruikte misleidingstechnieken herkennen en benoemen
- uitleggen waarom een grafiek een verkeerde indruk wekt, ook al kloppen de onderliggende cijfers
- van dezelfde dataset zowel een eerlijke als een bewust misleidende visualisatie maken in Power BI
- grafieken uit media en sociale media kritisch beoordelen voor je ze gelooft of deelt
:::

:::{admonition} Fictieve cijfers
:class: note
Alle grafieken in deze les zijn gemaakt met **fictieve cijfers**. Ze tonen de technieken, niet de echte werkloosheids- of verkoopcijfers. Echte voorbeelden ga je verderop zelf zoeken.
:::

## 1. Een grafiek is altijd een keuze

Wie een grafiek maakt, neemt beslissingen: welk grafiektype, welke periode, waar de as begint, welke kleuren, welke titel. Elk van die keuzes stuurt hoe de lezer de data interpreteert. Meestal kiest de maker gewoon wat het duidelijkst is. Maar wie een bepaald verhaal wil verkopen, kan met diezelfde keuzes ook naar een **verkeerde conclusie** sturen — zonder ook maar één cijfer te veranderen.

Waarom zou iemand dat doen?

* **Marketing**: een grafiek die een klein verschil groot doet lijken, verkoopt beter.
* **Politiek**: een lichte verbetering presenteren als een triomf, of een stijging bij de tegenstander uitvergroten — zeker in campagnetijd.
* **Clicks**: "Werkloosheid explodeert!" trekt meer lezers dan "Werkloosheid stijgt licht".
* **Desinformatie**: wie mensen bewust wil misleiden, gebruikt liever een suggestieve grafiek dan een aantoonbare leugen. Correcte cijfers in een misleidende grafiek zijn veel moeilijker te weerleggen.

Hieronder leer je de zes meest voorkomende trucs kennen. Eens je ze kent, zie je ze overal.

## 2. De trukendoos

### 2.1 De afgekapte y-as

De klassieker. Laat de y-as niet bij nul beginnen, maar net onder de laagste waarde. Elk klein verschil wordt dan visueel enorm.

![Dezelfde werkloosheidscijfers, links met afgekapte y-as en rechts met een as die bij nul begint](../../figures/misleiding_afgekapte_yas.svg)

Links lijkt de werkloosheid te crashen: de laatste balk is nog maar een derde van de eerste. Kijk echter naar de as: die loopt van 7,8% tot 8,6%. In werkelijkheid daalde de werkloosheid van 8,4% naar 8,0% — een reëel maar bescheiden verschil, zoals de rechtergrafiek eerlijk laat zien.

**Hoe herken je het?** Kijk bij elke grafiek eerst waar de y-as begint. Begint ze niet bij nul, vraag je dan af of daar een goede reden voor is — of dat iemand een klein verschil wil opblazen.

:::{admonition} Is een afgekapte as altijd fout?
:class: note
Nee. Soms is nul niet relevant en is de kleine variatie zelf het nieuws — denk aan een wisselkoers die rond een vaste waarde schommelt. Maar dan hoort de maker dat duidelijk aan te geven, en mag de grafiek niet suggereren dat iets "instort" of "explodeert". Bij een **staafdiagram** is afkappen bijna altijd misleidend: de lengte van de balk *is* daar de boodschap.
:::

### 2.2 Cherry-picking: kies de periode die je verhaal steunt

Toon niet de hele geschiedenis, maar precies dat stukje dat jouw verhaal ondersteunt.

![Dezelfde tijdreeks, links enkel de laatste drie jaar en rechts de volledige periode met context](../../figures/misleiding_cherry_picking.svg)

Links: "jeugdwerkloosheid stijgt weer fors" — en dat klopt, voor de laatste twee jaar. Rechts zie je de context: over tien jaar bekeken is de jeugdwerkloosheid sterk gedaald, en de "forse stijging" is een kleine rimpel. Merk ook op dat de linkergrafiek de rimpel extra uitvergroot met een afgekapte y-as: trucs worden graag gecombineerd.

**Hoe herken je het?** Vraag je bij elke tijdreeks af: waarom begint deze grafiek precies in dát jaar? Wat zou ik zien als de periode langer was?

### 2.3 De dubbele y-as

Zet twee reeksen met totaal verschillende schalen in één grafiek, elk met een eigen y-as. Door beide assen slim te kiezen kan je de lijnen laten samenvallen — en zo een sterk verband suggereren dat er niet of nauwelijks is.

![Links een dubbele-asgrafiek die marketingbudget en omzet laat samenvallen, rechts dezelfde reeksen elk in een eigen grafiek](../../figures/misleiding_dubbele_as.svg)

Links lijken marketingbudget en omzet perfect samen te stijgen: "elke euro marketing betaalt zichzelf terug!" Kijk naar de assen: de linkeras loopt van 0 tot 50 (duizend euro), de rechteras van 0,95 tot 1,25 (miljoen euro). Rechts zie je dezelfde cijfers eerlijk, elk met een eigen as die bij nul begint: het budget is verviervoudigd, de omzet steeg zo'n 17%. Misschien wérkt de marketing — maar de linkergrafiek suggereert veel meer dan de data toelaat.

**Hoe herken je het?** Twee y-assen in één grafiek is altijd een alarmbel. Controleer beide schalen afzonderlijk en stel je voor hoe de grafiek eruit zou zien met elke reeks apart.

### 2.4 Het verkeerde grafiektype

Elk grafiektype draagt een eigen boodschap. Een cirkeldiagram zegt: "deze delen vormen samen één geheel". Gebruik je het voor cijfers die géén geheel vormen, dan misleid je — zonder één fout cijfer.

![Links een cirkeldiagram van percentages die samen 269 procent vormen, rechts dezelfde cijfers als staafdiagram op een schaal tot 100 procent](../../figures/misleiding_grafiektype.svg)

In de enquête links mochten jongeren **meerdere** platformen aanduiden: de percentages tellen op tot 269%. In een cirkeldiagram geperst lijkt WhatsApp plots "maar een derde" — terwijl 82% van de jongeren het gebruikt. Het staafdiagram rechts toont elk percentage op een schaal van 0 tot 100%, zoals het hoort.

**Hoe herken je het?** Tel bij een cirkeldiagram de percentages op. Komt de som niet uit op 100%, dan is het diagram fout — of misleidend.

### 2.5 Absolute versus relatieve cijfers

Hetzelfde feit kan je vertellen in absolute aantallen of in percentages. Beide kunnen kloppen — en toch een tegengesteld gevoel geven. Een fictief voorbeeld:

| | 2004 | 2024 |
|---|---:|---:|
| Bevolking op beroepsactieve leeftijd | 6,50 miljoen | 7,10 miljoen |
| Werkende personen | 4,20 miljoen | 4,55 miljoen |
| Werkzaamheidsgraad | 64,6% | 64,1% |

"**Nog nooit waren zoveel Belgen aan het werk!**" — dat klopt: 4,55 miljoen is een record. Maar de bevolking groeide sneller dan het aantal jobs, dus de **werkzaamheidsgraad** (het aandeel werkenden in de bevolking op beroepsactieve leeftijd — daarover gaat de volgende opdracht) daalde licht. Beide cijfers zijn juist; welk cijfer je kiest, bepaalt het verhaal.

Het omgekeerde bestaat ook: percentages op kleine aantallen klinken spectaculair. "50% meer inbraken in gemeente X!" kan betekenen: van twee naar drie inbraken.

**Hoe herken je het?** Vraag altijd: is dit een absoluut aantal of een percentage — en verandert het verhaal als ik het andere bekijk?

### 2.6 Kleuren en 3D-effecten

De laatste twee trucs werken niet op je verstand, maar op je gevoel:

* **Kleuren dragen een boodschap.** Rood betekent gevaar, groen betekent veilig. Wie een gewone stijging in vuurrood zet, wekt onrust zonder één woord te liegen. Ook op kaarten gebeurt dit: door de grenzen tussen kleurzones te verschuiven, kleurt een land "dieprood" voor een verschil dat nauwelijks bestaat.
* **3D-effecten vervormen.** In een 3D-cirkeldiagram lijkt het segment vooraan groter dan het is: het staat dichter bij de kijker en je ziet ook de zijkant van de schijf. Handig als je jouw marktaandeel vooraan wilt zetten. 3D voegt nooit informatie toe: als je het ziet, is het versiering — of misleiding.

**Hoe herken je het?** Vraag je af: zegt de kleur of het effect iets wat de cijfers zelf niet zeggen?

## 3. Opdracht 1: word grafiekdetective

Nu je de trukendoos kent, ga je ze herkennen in het echt. Zoek een **echt voorbeeld** van een misleidende grafiek in de (Belgische) media of op sociale media.

Waar zoek je?

* nieuwssites en kranten, zeker bij opiniestukken
* posts van politici en partijen op sociale media, vooral in campagnetijd
* reclame en "studies" van bedrijven
* grafieken die viraal gaan op sociale media

Wat lever je in via MS Teams?

1. een screenshot van de grafiek;
2. de bron: waar en wanneer je ze vond;
3. welke techniek(en) uit deze les je herkent;
4. hoe de eerlijke versie eruit had gezien (beschrijven volstaat, namaken hoeft niet).

We bespreken een selectie van de inzendingen klassikaal.

:::{admonition} Niet elke slechte grafiek is kwaad bedoeld
:class: note
Soms is een grafiek gewoon onhandig gemaakt, zonder de bedoeling om te misleiden. Geef in je inzending aan wat jij denkt — bewuste misleiding of slordigheid — en waarom.
:::

## 4. Opdracht 2: één dataset, twee verhalen

Je gaat nu zelf één keer liegen — met toestemming. Wie zelf een misleidende grafiek heeft gebouwd, doorziet er daarna elke andere.

Gebruik de **werkloosheidsdataset uit de vorige les**: die heb je al in Power BI zitten. Kies één invalshoek, bijvoorbeeld de evolutie van de werkloosheid bij één opleidingsniveau, of het verschil tussen twee groepen.

Maak in je Power BI-bestand twee nieuwe pagina's:

* **Pagina "Eerlijk"**: een grafiek die de data correct en duidelijk weergeeft.
* **Pagina "Misleidend"**: een grafiek met **exact dezelfde gegevens** die een ander (fout) verhaal vertelt. Gebruik minstens **twee** technieken uit deze les en geef de grafiek een sensationele titel.

Daarna presenteer je aan de klas:

1. Je toont eerst alleen je **misleidende** grafiek. De klas benoemt: welke indruk wekt de grafiek, en welke technieken zitten erin?
2. Daarna toon je de **eerlijke** versie en leg je uit wat je precies veranderd hebt.

Checklist:

- [ ] Beide pagina's gebruiken exact dezelfde gegevens.
- [ ] De misleidende versie bevat minstens twee technieken uit deze les.
- [ ] Je kan bij elke gebruikte techniek uitleggen wat ze met de lezer doet.
- [ ] De eerlijke versie heeft een correcte titel, een y-as die bij nul begint (of een goed verantwoorde uitzondering) en de volledige relevante periode.

:::{admonition} Hints: zo "lieg" je in Power BI
:class: dropdown

* **Y-as afkappen**: selecteer de visual → penseel-icoontje → *Y-as* → *Bereik*: vul zelf een minimum en een maximum in.
* **Cherry-picking**: sleep het veld `jaar` naar het deelvenster *Filters* en beperk de periode.
* **Kleuren**: penseel-icoontje → *Kleuren* (of *Lijnen*): zet je reeks in dreigend rood of geruststellend groen.
* **Titel**: penseel-icoontje → *Algemeen* → *Titel*: de tekst boven een grafiek stuurt de lezer minstens zo hard als de grafiek zelf.
:::

## 5. Wat betekent dit voor jouw eigen rapporten?

In de onderzoeksopdrachten en de overhoring hoort bij elke visualisatie een **interpretatie**, en die telt mee in de rubric. Vanaf deze les hoort daar ook dit bij: een grafiek "lezen" is óók kunnen zien wat er **níet** klopt — in grafieken van anderen én in die van jezelf.

Gebruik daarom deze checklist telkens je een grafiek maakt of beoordeelt:

- [ ] Waar begint de y-as? Bij nul — en zo niet, is daar een goede reden voor?
- [ ] Welke periode toont de grafiek, en waarom precies die?
- [ ] Staat er maar één y-as? Twee assen: extra kritisch kijken.
- [ ] Past het grafiektype bij de data? Bij een cirkeldiagram: telt alles samen op tot 100%?
- [ ] Gaat het om absolute aantallen of om percentages — en verandert het verhaal als je het andere bekijkt?
- [ ] Zeggen de kleuren, effecten of de titel iets wat de cijfers zelf niet zeggen?

Een grafiek die deze test doorstaat, mag in je rapport. In de volgende opdracht — de werkzaamheidsgraad in de EU — werk je een stuk zelfstandiger. Hou de checklist bij de hand.
