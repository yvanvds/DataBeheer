# 1. Big data: van transactiedata naar maatschappelijke data

Tot nu toe werkte je vooral met **klassieke bedrijfsdata**. Denk bijvoorbeeld aan verkoopscijfers, bestellingen, klanten, producten of voorraad. Zulke data ontstaat meestal binnen één organisatie en wordt verzameld om dagelijkse processen te ondersteunen. Een webshop bewaart bijvoorbeeld welke klant welk product bestelde, wanneer dat gebeurde en hoeveel er betaald werd. Zulke databanken zijn belangrijk voor de werking van een bedrijf: zonder die data kan je geen bestellingen verwerken, facturen opmaken of voorraden beheren.

Bij **big data** gaat het om een ander soort gegevensverzameling. Het gaat niet alleen om “veel data”, maar ook om data die uit verschillende bronnen kan komen, snel verandert, en vaak gebruikt wordt om bredere patronen te onderzoeken. Denk bijvoorbeeld aan werkloosheidscijfers, bevolkingsdata, gezondheidsstatistieken, beurskoersen, verkeersinformatie of weergegevens. Zulke data wordt niet verzameld om één concrete bestelling of één transactie op te slaan, maar om **trends, verbanden en evoluties** te kunnen analyseren.

## 1. Wat maakt big data anders?

Bij klassieke bedrijfsdata werk je vaak met een goed afgebakende databank. De structuur ligt vooraf vast. Je weet welke tabellen er zijn, welke velden ze bevatten en hoe die tabellen aan elkaar gekoppeld zijn. Een webshop heeft bijvoorbeeld tabellen zoals `customers`, `orders`, `products` en `order_lines`. Die data is meestal **genormaliseerd**, zodat er zo weinig mogelijk redundantie is en de databank goed onderhoudbaar blijft.

Big data werkt anders. Daar vertrek je vaak niet vanuit een databank die speciaal voor jouw toepassing gebouwd werd. Je vertrekt eerder vanuit een grote bestaande gegevensbron en stelt jezelf de vraag: **welke informatie kan ik hieruit halen?** Dat vraagt een andere manier van denken. Je bent minder bezig met transacties invoeren en meer met onderzoeken, filteren, vergelijken en interpreteren.

Big data wordt vaak uitgelegd aan de hand van de **5 V’s**:

### Volume

Er is veel data beschikbaar. Soms gaat het om duizenden rijen, soms om miljoenen of miljarden. Belangrijker dan het exacte aantal is dat de hoeveelheid te groot wordt om nog manueel te analyseren.

### Velocity

Data kan snel veranderen of voortdurend aangevuld worden. Denk aan realtime beurskoersen, weerdata of verkeerstellingen. Bij zulke data is het belangrijk dat systemen snel kunnen verwerken en bijwerken.

### Variety

Data komt in veel vormen voor. Sommige data staat netjes in tabellen, maar andere data zit in tekstbestanden, sensorgegevens, afbeeldingen of online databronnen. Ook binnen tabellen kan de structuur sterk verschillen.

### Veracity

Niet alle data is even betrouwbaar. Je moet dus kritisch kijken naar de bron, de meetmethode en mogelijke fouten of vertekeningen. Bij maatschappelijke data is dat extra belangrijk, omdat cijfers vaak gebruikt worden in debatten of in de media.

### Value

Data heeft pas waarde als je er zinvolle informatie uit haalt. Een grote dataset op zich is nog niet interessant. Ze wordt pas nuttig wanneer je er een onderzoeksvraag mee kan beantwoorden of wanneer je patronen en verbanden zichtbaar kan maken.

We verwachten dat je deze kenmerken niet alleen kan opsommen, maar ook kan inschatten in functie van een onderzoek. De 5 V’s hebben namelijk invloed op hoe bruikbaar een dataset is, hoe betrouwbaar je conclusies zijn en welke analyses mogelijk worden. 

## 2. Klassieke bedrijfsdata tegenover big data

Het verschil tussen beide soorten data kan je als volgt samenvatten.

Bij klassieke bedrijfsdata stel je vragen zoals:

* Welke klant plaatste de bestelling?
* Hoeveel producten werden verkocht?
* Wat is de omzet per maand?
* Welke producten zitten bijna zonder voorraad?

Bij big data stel je eerder vragen zoals:

* Hoe evolueert de werkloosheid doorheen de tijd?
* Is er een verband tussen opleidingsniveau en werkloosheid?
* Verschilt werkloosheid tussen groepen in de bevolking?
* Welke maatschappelijke patronen zie je in de cijfers?

In beide gevallen werk je met data, maar het **doel** verschilt. Bij bedrijfsdata gaat het meestal om het ondersteunen van een operationeel proces. Bij big data gaat het vaker om **analyse, beleid, onderzoek en besluitvorming**.

Dat verschil zie je ook in de manier waarop data georganiseerd wordt. Een klassieke relationele databank is ontworpen om gegevens correct op te slaan en te wijzigen. Een analytische dataset of een datawarehouse is eerder bedoeld om gegevens te bevragen, samen te voegen en te visualiseren. 

## 3. Waar vind je dit soort data?

Er bestaan heel wat websites waarop overheden, internationale organisaties en onderzoeksinstellingen datasets publiceren. Enkele bronnen waarmee je een analytische dataset kan opbouwen, zijn Statbel, Provincies in Cijfers, data.europa.eu, data.worldbank.org, stat.nbb.be, data.imf.org en data.unicef 

Hieronder bespreken we enkele belangrijke voorbeelden.

### Statbel

Statbel is het Belgische statistiekbureau. Je vindt er officiële cijfers over bevolking, arbeid, economie, prijzen, inkomens en nog veel meer. Als je onderzoeksvraag enkel over België gaat, is dit vaak een logische bron.

### Provincies in Cijfers

Deze website bundelt cijfermateriaal op lokaal en regionaal niveau. Ze is interessant wanneer je verschillen tussen provincies, steden of gemeenten wilt onderzoeken.

### data.europa.eu

Dit is een Europees dataportaal waarop datasets uit verschillende landen en instellingen verzameld worden. De hoeveelheid data is groot, maar daardoor is de kwaliteit en bruikbaarheid ook niet altijd even gelijkmatig.

### World Bank, IMF, UNICEF

Deze internationale organisaties publiceren data over thema’s zoals economie, onderwijs, armoede, gezondheid en demografie. Dat is vooral nuttig wanneer je landen wereldwijd wilt vergelijken.

### Eurostat

Eurostat is het statistisch bureau van de Europese Unie. Het verzamelt en publiceert cijfers over de lidstaten van de EU en biedt een grote hoeveelheid vergelijkbare gegevens aan over onder meer economie, arbeid, onderwijs en bevolking.

Voor onze eerste verkenning kiezen we voor **Eurostat**. Die bron is interessant omdat:

* de data officieel en betrouwbaar is;
* je vaak meerdere jaren naast elkaar kan bekijken;
* de structuur geschikt is om te analyseren in Excel of Power BI;
* je als gebruiker zelf filters kan instellen voor je download.

## 4. Hoe gaan we te werk?

Grote datasets analyseer je niet met pen en papier, en ook niet met SQL alleen. In dit deel van de cursus gebruiken we daarvoor **Power BI**, een programma van Microsoft waarmee je data importeert, bewerkt en visualiseert in interactieve rapporten.

We bouwen dit deel stap voor stap op:

1. In het **volgende hoofdstuk** leer je Power BI kennen. Dat doen we nog niet met big data, maar met vertrouwde data: de webshop uit de SQL-lessen. Zo kan je je volledig concentreren op het programma zelf.
2. Daarna volgen drie **onderzoeksopdrachten** met echte Eurostat-data: eerst volledig begeleid, daarna steeds zelfstandiger.
3. Na de eerste onderzoeksopdracht staan we stil bij **misleidende grafieken**: je leert hoe een visualisatie een verkeerd beeld kan geven, en waarom dat soms bewust gebeurt.
4. We sluiten af met een **overhoring** waarin je zelfstandig een Eurostat-dataset analyseert.
