# 2. Aan de slag met Power BI

In het vorige hoofdstuk zag je waar je grote datasets kan vinden. Maar hoe haal je daar inzichten uit? Daarvoor gebruiken we **Power BI Desktop**: een gratis programma van Microsoft waarmee je data importeert, bewerkt en omzet in interactieve rapporten. Het is een van de meest gebruikte **Business Intelligence**-programma's (BI) in bedrijven.

In dit hoofdstuk leer je Power BI kennen. Dat doen we bewust **niet** met big data, maar met data die je door en door kent: de webshop uit de SQL-lessen. Zo weet je perfect wat elke rij en elke kolom betekent, en kan je je volledig concentreren op het programma zelf. Vanaf het volgende hoofdstuk passen we dezelfde technieken toe op echte Eurostat-data.

Het hoofdstuk bestaat uit twee delen. Eerst bouwen we **klassikaal** een eerste rapport, stap voor stap. Daarna werk je **zelfstandig** drie challenges af: je krijgt telkens een afbeelding van een afgewerkt dashboard, en jij bouwt het exact na.

:::{admonition} Leerdoelen
:class: tip

Na dit hoofdstuk kan je:
- een Excel-bestand importeren in Power BI en de gegevensweergave controleren
- kaarten, lijndiagrammen, staafdiagrammen en cirkeldiagrammen maken
- visualisaties sorteren, filteren met een Top N-filter en duidelijke titels geven
- een berekende kolom maken met een eenvoudige DAX-formule
- een slicer toevoegen zodat je rapport interactief wordt
- een afgewerkt dashboard zelfstandig nabouwen op basis van een voorbeeld
:::

## 1. Power BI Desktop installeren

Power BI Desktop is gratis. Installeer het via de **Microsoft Store**: zoek op "Power BI Desktop" en klik op *Installeren*. Start het programma daarna één keer op om te controleren dat alles werkt. Je hoeft je **niet aan te melden** met een account om rapporten te kunnen maken: het aanmeldvenster mag je gewoon wegklikken.


## 2. Van database naar dataset

Power BI werkt het liefst met een **platte tabel**: één rij per gebeurtenis, met alle bijbehorende informatie in kolommen ernaast. Onze webshop-database is net **niet** plat: de gegevens zitten verspreid over vier tabellen (`customers`, `orders`, `products` en `order_lines`). Dat is ideaal om bestellingen correct op te slaan, maar onhandig om te analyseren.

Gelukkig weet jij intussen hoe je die tabellen weer aan elkaar plakt. Wij hebben de database alvast voor jou omgezet naar een Excel-bestand, met precies het soort query dat jij in de SQL-lessen leerde schrijven:

```sql
SELECT o.order_id, o.order_date,
       c.first_name || ' ' || c.last_name AS klant, c.city,
       p.category, p.name,
       ol.quantity, ol.unit_price,
       ol.quantity * ol.unit_price AS regelbedrag
FROM order_lines ol
JOIN orders o    ON o.order_id = ol.order_id
JOIN customers c ON c.customer_id = o.customer_id
JOIN products p  ON p.product_id = ol.product_id
WHERE o.status != 'cancelled';
```

👉 Download het resultaat: [webshop_verkopen.xlsx](/_static/excel/webshop_verkopen.xlsx)

Het bestand bevat twee tabbladen:

* **Verkopen** — 6.355 rijen: elke rij is één orderregel, met het bestelnummer, de datum (plus aparte kolommen *Jaar* en *Maand*), de klant, de stad, de categorie, het product, het aantal, de eenheidsprijs en het regelbedrag (aantal × eenheidsprijs).
* **Voorraad** — de productcatalogus: elk product met zijn categorie, eenheidsprijs en huidige voorraad.

:::{admonition} Data wordt onderweg al opgekuist
:class: note

De export bevat geen geannuleerde bestellingen (zie de `WHERE`-regel hierboven), en het foutieve stadje "gent" uit de database werd rechtgezet naar "Gent". Onthoud dit principe: de dataset die je analyseert is bijna nooit de ruwe data zelf — er gingen al keuzes en opkuiswerk aan vooraf. Dit proces heet **ETL** (zie het vorige hoofdstuk): de gegevens werden uit de webshop-database gehaald (*Extract*), opgekuist en samengevoegd tot één platte tabel (*Transform*), en in een Excel-bestand geladen dat jij straks in Power BI importeert (*Load*). Bij de Eurostat-data in de volgende hoofdstukken doe je dat opkuiswerk zelf.
:::


## 3. Importeren in Power BI

Voer de volgende stappen uit:

1. Open **Power BI Desktop** en start met een leeg rapport.
2. Klik in het lint op **Gegevens ophalen** en kies **Excel-werkmap**.
3. Zoek het bestand `webshop_verkopen.xlsx` dat je zonet hebt gedownload en klik op **Openen**.
4. In het **navigatievenster** zie je welke werkbladen Power BI in het bestand herkent. Vink **Verkopen** én **Voorraad** aan.
5. Klik op **Laden**.

Power BI importeert nu de gegevens in je rapport.

### Controleer of de gegevens correct zijn ingelezen

Klik links op de **gegevensweergave** (het tabel-icoontje). Controleer voor beide tabellen:

* Staan de kolomnamen er duidelijk in?
* Wordt de kolom *Datum* herkend als datum?
* Worden *Aantal*, *Eenheidsprijs* en *Regelbedrag* als getallen gelezen?
* Zie je onderaan hoeveel rijen er zijn? (*Verkopen* heeft er 6.355.)

Wanneer een kolom verkeerd werd ingelezen, kan dat later voor problemen zorgen. Sla deze controle dus nooit over.


## 4. Samen: je eerste rapport

We bouwen nu klassikaal een eerste rapport met vier visualisaties. Werk mee op je eigen laptop.

### 4.1 Een kaart met de totale omzet

Een **kaart** (card) toont één groot getal. Handig voor het belangrijkste cijfer van je rapport.

1. Klik in het deelvenster **Visualisaties** op de kaart.
2. Sleep het veld `Regelbedrag` uit de tabel *Verkopen* naar de visual.
3. Power BI telt automatisch alle regelbedragen op.

> **Controle:** je kaart toont afgerond **1,13 miljoen**. De exacte som is **€ 1.128.345,39**.

### 4.2 De omzet doorheen de tijd

1. Voeg een **lijndiagram** toe.
2. Sleep `Datum` naar de **X-as** en `Regelbedrag` naar de **Y-as**.
3. Power BI maakt van een datumkolom automatisch een **datumhiërarchie**: jaar → kwartaal → maand → dag. Standaard zie je de omzet per jaar.
4. Gebruik de **drilldown-knoppen** bovenaan de visual (de gevorkte pijltjes) om af te dalen naar kwartalen en maanden. Probeer ook eens tot op de dag: dan zie je vooral ruis. Voor deze data is **per maand** het meest leesbare niveau.

Wat valt op? De omzet piekt telkens in **november en december** (eindejaarsaankopen), en de webshop **groeit** elk jaar.

### 4.3 De populairste producten

1. Voeg een **staafdiagram** (liggende balken) toe.
2. Sleep `Product` naar de **Y-as** en `Regelbedrag` naar de **X-as**.
3. Je ziet nu álle producten — dat zijn er te veel. We beperken tot de top 10: open het deelvenster **Filters**, kies bij *Product* als filtertype **Top N**, vul **10** in en sleep `Regelbedrag` naar het vak **Op waarde**. Klik op **Filter toepassen**.
4. Controleer dat de balken aflopend gesorteerd staan (drie puntjes rechtsboven de visual → **Sorteren op**).

> **Controle:** bovenaan staat **Pro Laptop 16** met **€ 81.562,05**.

### 4.4 Zelf doen: de topklanten

Maak nu zelf een staafdiagram met de **top 10 klanten volgens omzet**. Dezelfde stappen als hierboven, maar dan met het veld `Klant`.

> **Controle:** je beste klant is **Axel Aerts** met **€ 26.017,42**.

### 4.5 Titels en opmaak

Een rapport moet ook zonder mondelinge uitleg leesbaar zijn:

1. Selecteer een visual en klik op het **penseel-icoontje** (visual opmaken).
2. Geef elke visual een duidelijke Nederlandse titel, bijvoorbeeld *"Omzet per maand"* in plaats van *"Som van Regelbedrag per Datum"*.
3. Versleep de visuals en maak ze groter of kleiner tot je een overzichtelijk geheel hebt.

Sla je bestand op als `webshop_rapport.pbix`. Klaar — je hebt je eerste dashboard gebouwd!


## 5. Challenges: bouw het dashboard na

Nu is het aan jou. Hieronder staan drie challenges, telkens een beetje moeilijker. Je krijgt per challenge een **afbeelding van een afgewerkt dashboard**, en jij bouwt het **exact na** met de dataset die je al geïmporteerd hebt. De cijfers in de afbeeldingen komen rechtstreeks uit jouw dataset: als jouw getallen afwijken, is er iets mis met je visual.

Werk verder in hetzelfde Power BI-bestand en maak per challenge een **nieuwe pagina** (met de tabs onderaan, net zoals in Excel). Vink de checklist af: klaar is klaar.

### Challenge 2 — Voorraadrapport

*Moeilijkheid: ⭐⭐ — berekende kolom en opmaak. Tabblad: Voorraad.*

De zaakvoerder wil weten hoeveel geld er "op de plank" ligt: de **voorraadwaarde** van elk product. Die kolom bestaat niet in de dataset — je maakt ze zelf met een **berekende kolom**:

1. Klik in de gegevensweergave op de tabel *Voorraad*.
2. Kies **Modellering → Nieuwe kolom** en typ deze DAX-formule:

   ```DAX
   Voorraadwaarde = 'Voorraad'[Eenheidsprijs] * 'Voorraad'[Voorraad]
   ```

3. Geef de nieuwe kolom een **valutanotatie**: selecteer ze en klik bij *Modellering* op het valutasymbool.

Bouw daarna dit dashboard na:

![Voorbeelddashboard challenge 2: voorraadrapport](../../figures/powerbi_challenge2.svg)

Checklist:

- [ ] Een berekende kolom `Voorraadwaarde` met valutanotatie.
- [ ] Een **tekstvak** met de rapporttitel "Voorraadrapport webshop".
- [ ] Een **kaart** met de totale voorraadwaarde, in euro. *Controle: € 186.320,50.*
- [ ] Een **staafdiagram** met de voorraadwaarde per categorie, met gegevenslabels, en in het **oranje** in plaats van het standaardblauw. *Controle: Computers staat bovenaan met € 75.497,00.*
- [ ] Een **tabel** met per product de voorraad en de voorraadwaarde, **aflopend gesorteerd** op voorraadwaarde. *Controle: bovenaan staat Pro Laptop 16 met € 17.091,00, en de totaalrij onderaan toont € 186.320,50.*

:::{admonition} Hints
:class: dropdown

* Een tekstvak voeg je toe via **Invoegen → Tekstvak**.
* De kleur van de balken verander je via het penseel-icoontje → *Balken* → *Kleuren*.
* Sorteren in een tabel: klik gewoon op de kolomkop *Voorraadwaarde*.
:::


### Challenge 3 — Interactief verkoopdashboard

*Moeilijkheid: ⭐⭐⭐ — slicers en meerdere visuals op één pagina. Tabblad: Verkopen.*

Tot nu toe toonden je dashboards altijd álle data. Met een **slicer** laat je de lezer zelf filteren: klik op een jaar of een categorie, en alle visuals op de pagina passen zich aan. Zo wordt je rapport pas echt interactief.

Bouw dit dashboard na. Op de afbeelding is in de slicer **Jaar** het jaar **2024** aangeklikt, dus alle visuals tonen enkel 2024:

![Voorbeelddashboard challenge 3: interactief verkoopdashboard](../../figures/powerbi_challenge3.svg)

Checklist:

- [ ] Een **slicer** met het veld `Jaar`, en een tweede slicer met het veld `Categorie`.
- [ ] Een **kaart** met de omzet.
- [ ] Een **kaart** met het aantal **bestellingen**. Let op: dat is niet het aantal rijen — meerdere rijen kunnen bij dezelfde bestelling horen.
- [ ] Een **lijndiagram** met de omzet per maand (gebruik de kolom `Maand`).
- [ ] Een **staafdiagram** met de top 10 producten volgens verkochte stuks (Top N-filter).
- [ ] Selecteer 2024 in de jaar-slicer en controleer: omzet **€ 393.843,54**, **989** bestellingen, december is de piekmaand met **€ 98.334,52**, en bovenaan de top 10 staat **Cable Kit USB-C** met **315** stuks.
- [ ] Klik ook eens op een categorie in de tweede slicer en kijk hoe alle visuals mee veranderen.

:::{admonition} Hints
:class: dropdown

* Het aantal bestellingen tel je met het veld `Bestelnummer`: sleep het naar de kaart, klik op het pijltje bij het veld en kies **Aantal (uniek)**. Hetzelfde bestelnummer wordt dan maar één keer geteld.
* Ziet je slicer eruit als een schuifbalk? Klik op het pijltje rechtsboven in de slicer en kies **Lijst**.
:::
