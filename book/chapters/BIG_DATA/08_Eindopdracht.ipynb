# 8. Eindopdracht: kies je eigen dataset

In dit deel kreeg je bij elke opdracht een dataset van mij, met instructies erbij. In een echt onderzoek werkt het net omgekeerd: je vertrekt van een vraag of een thema, en je gaat zelf op zoek naar data waarmee je die vraag kan beantwoorden. Dat is precies de vaardigheid die deze eindopdracht toetst: **een dataset kiezen en inschatten of ze bruikbaar is voor jouw onderzoek**.

Helemaal in het wilde weg zoeken doen we niet — het internet staat vol datasets die er interessant uitzien, maar onbruikbaar blijken zodra je ze downloadt. Daarom werk je met een **keuzemenu**: acht datasets waarvan ik zeker weet dat ze werken. Jij kiest er één, formuleert er zelf je onderzoeksvragen bij, en toetst met een checklist of je keuze standhoudt.

## Wat ga je doen?

* Je kiest **één dataset** uit het keuzemenu (stap 1). Wie meer uitdaging wil, mag ook daarbuiten zoeken — lees dan zeker het einde van stap 1.
* Je formuleert er **vier eigen onderzoeksvragen** bij (stap 2). Deze keer krijg je er géén van mij.
* Je toetst met de **checklist** of je dataset die vragen echt aankan (stap 3).
* Je haalt de data op, maakt ze klaar en bouwt je PowerBI-rapport (stappen 4 en 5): per vraag een tab met visualisatie en interpretatie, plus een verantwoordingstab.

## AI gebruiken

Dezelfde regels als bij de overhoring: AI mag je helpen bij het **interpreteren** van je grafieken, niet bij het bouwen van je visualisaties. Je blijft zelf verantwoordelijk voor elke tekst in je rapport. Eén belangrijke toevoeging: je **onderzoeksvragen formuleer je zelf**. Dat is net wat deze opdracht toetst.

## Hoe wordt dit beoordeeld?

De rubric vind je in MS Teams. Alle zes de onderzoekscompetenties uit het kader hierboven tellen nu mee. Nieuw ten opzichte van de vorige opdrachten:

* De **kwaliteit van je vier onderzoeksvragen** weegt zwaarder door — ze zijn nu immers allemaal zelfgekozen.
* Ook je **verantwoording** (stap 3: waarom is deze dataset bruikbaar voor jouw vragen?) wordt beoordeeld.

Verder blijft alles gelden wat je ondertussen kent: datavoorbereiding telt mee, presentatie telt mee, en per vraag tellen visualisatie én interpretatie.

## Stap 1: kies een dataset uit het keuzemenu

Hieronder vind je acht datasets die zeker bruikbaar zijn: ze zijn downloadbaar, niet te groot, en ze bevatten genoeg dimensies om er interessante vragen over te stellen. Bij elke dataset hoort een korte **fiche**: het thema, de dimensies waarop je kan filteren en vergelijken, en de valkuilen waar je op moet letten.

Lees méér dan één fiche voor je kiest. Een goede keuze is een dataset waarvan het thema je interesseert én waarvan je meteen voelt dat er vragen in zitten. Twijfel je tussen twee? Kies dan degene met de eenvoudigste structuur.

### Fiche 1 — Minimumlonen in de EU

👉 [Eurostat — earn_mw_cur](https://ec.europa.eu/eurostat/databrowser/view/earn_mw_cur/default/table)

**Thema:** het wettelijke maandelijkse minimumloon in de EU-landen — een lange tijdreeks, twee keer per jaar gemeten. Goed voor vragen over loonverschillen tussen landen en hun evolutie.

**Dimensies:** land, tijd (per halfjaar), munteenheid (euro, koopkrachtstandaard of nationale munt).

**Valkuilen:**

* Vijf EU-landen (Denemarken, Italië, Oostenrijk, Finland en Zweden) hebben geen wettelijk minimumloon en ontbreken dus in de data.
* Kies bewust tussen euro en koopkrachtstandaard (PPS). In euro lijken de Oost-Europese lonen extreem laag; in PPS — gecorrigeerd voor de prijzen in elk land — wordt het verschil een stuk kleiner. Dat verschil kan op zich al een onderzoeksvraag zijn.
* De data is halfjaarlijks (S1 en S2). Vergelijk je jaren, kies dan telkens hetzelfde halfjaar.
* Er zitten ook een paar niet-EU-landen in (zoals de Verenigde Staten en Turkije).

### Fiche 2 — Huizenprijzen in de EU

👉 [Eurostat — prc_hpi_a](https://ec.europa.eu/eurostat/databrowser/view/prc_hpi_a/default/table)

**Thema:** de huizenprijsindex: hoe snel worden woningen duurder in elk EU-land?

**Dimensies:** land, jaar, type aankoop (totaal, nieuwbouw, bestaande woningen), eenheid (index of jaarlijkse verandering).

**Valkuilen:**

* Dit is een **index** (2015 = 100), geen prijs in euro. Je kan dus zeggen dat huizen in Hongarije sneller duurder werden dan in België — niet dat ze er duurder zíjn.
* Kies bewust tussen de index en het jaarlijkse veranderingspercentage, en meng de twee nooit in één grafiek.

### Fiche 3 — Levenstevredenheid in de EU

👉 [Eurostat — ilc_pw01](https://ec.europa.eu/eurostat/databrowser/view/ilc_pw01/default/table)

**Thema:** hoe tevreden zijn Europeanen met hun leven, op een schaal van 0 tot 10?

**Dimensies:** land, jaar (2013, 2018 en jaarlijks vanaf 2021), geslacht, leeftijdsgroep, opleidingsniveau.

**Valkuilen:**

* Dit is enquêtedata en dus subjectief. Ook cultuurverschillen spelen mee in hoe mensen scoren — wees voorzichtig met straffe conclusies.
* Veel dimensies. Zet alles wat je niet onderzoekt op "Total", anders wordt je download onoverzichtelijk groot.
* Vóór 2021 zijn er maar twee meetjaren (2013 en 2018) — geen doorlopende tijdreeks.

### Fiche 4 — Hernieuwbare energie

👉 [Eurostat — nrg_ind_ren](https://ec.europa.eu/eurostat/databrowser/view/nrg_ind_ren/default/table)

**Thema:** het aandeel hernieuwbare energie in het energieverbruik van elk EU-land, van 2004 tot nu.

**Dimensies:** land, jaar, indicator (totaal, elektriciteit, verwarming en koeling, transport).

**Valkuilen:**

* De vier indicatoren zijn aparte metingen. Kies er één of vergelijk ze, maar tel ze nooit op.
* Grote verschillen tussen landen hebben vaak een geografische verklaring (waterkracht in Zweden en Oostenrijk, bijvoorbeeld). Hou daar rekening mee in je interpretatie.

### Fiche 5 — NEET-jongeren

👉 [Eurostat — edat_lfse_20](https://ec.europa.eu/eurostat/databrowser/view/edat_lfse_20/default/table)

**Thema:** het percentage jongeren dat niet werkt én geen opleiding volgt ("Not in Employment, Education or Training").

**Dimensies:** land, jaar, geslacht, leeftijdsgroep, plus extra detaildimensies (werkstatus, soort opleiding).

**Valkuilen:**

* Kies één leeftijdsgroep (bijvoorbeeld 15–29 jaar) en hou die aan in je hele rapport.
* Laat de detaildimensies op "Total" staan, tenzij je vraag er expliciet om vraagt.
* NEET is niet hetzelfde als jeugdwerkloosheid: wie voltijds studeert, is geen NEET, ook zonder job. Lees de definitie goed voor je interpreteert.

### Fiche 6 — Online winkelen in de EU

👉 [Eurostat — isoc_ec_ib20](https://ec.europa.eu/eurostat/databrowser/view/isoc_ec_ib20/default/table)

**Thema:** wie koopt er online? Het percentage mensen dat online aankopen doet, per land en per bevolkingsgroep.

**Dimensies:** land, jaar (vanaf 2020), bevolkingsgroep (leeftijd, geslacht, opleidingsniveau, ...), indicator.

**Valkuilen:**

* De tijdreeks start pas in 2020 — voor evolutievragen heb je dus maar een handvol jaren.
* Er zijn veel indicatoren en bevolkingsgroepen. Kies er vooraf gericht een beperkt aantal, anders verdrink je in de download.

### Fiche 7 — Vastgoedprijzen in België

👉 [Statbel — Vastgoedprijzen](https://statbel.fgov.be/nl/themas/bouwen-wonen/vastgoedprijzen)

**Thema:** de mediaanprijs van huizen en appartementen in België, tot op gemeenteniveau. De enige dataset in het menu waarmee je *binnen* België kan vergelijken.

**Dimensies:** gemeente (of arrondissement/provincie), jaar, woningtype.

**Valkuilen:**

* Dit is geen Eurostat: zoek op de pagina zelf het downloadbare Excel-bestand en verken de structuur voor je begint.
* Gemeenten met weinig verkopen krijgen geen cijfer (privacy) — verwacht dus lege cellen.
* Het gaat om de **mediaanprijs**, niet de gemiddelde prijs. Denk na over wat dat verschil betekent.
* Met honderden gemeenten wordt een grafiek snel onleesbaar. Filter op een regio, of werk per arrondissement of provincie.

### Fiche 8 — Levensverwachting wereldwijd

👉 [World Bank — Life expectancy at birth](https://data.worldbank.org/indicator/SP.DYN.LE00.IN)

**Thema:** de levensverwachting bij geboorte voor bijna elk land ter wereld, sinds 1960. De enige dataset in het menu waarmee je wereldwijd kan vergelijken.

**Dimensies:** land, jaar.

**Valkuilen:**

* Download via de knop **CSV** of **EXCEL** op de pagina. Het bestand heeft metadatarijen bovenaan en de jaren als kolommen — je zal dus moeten unpivotten in Power Query, zoals in hoofdstuk 6.
* Tussen de landen staan ook regio's en groepen ("World", "European Union", "Sub-Saharan Africa", ...). Filter die weg voor je een top 10 maakt.
* Voor de recentste jaren ontbreken soms nog cijfers.

### Liever een dataset buiten het menu?

Dat mag — op drie voorwaarden:

1. Je dataset doorstaat de **volledige checklist** uit stap 3.
2. Je komt ze **vooraf aan mij tonen**, samen met je ingevulde checklist. Pas na mijn akkoord begin je te bouwen.
3. De opdracht blijft verder identiek: vier eigen vragen, dezelfde tabs, dezelfde rubric.

Goede plekken om te zoeken zijn de portalen uit hoofdstuk 1: Eurostat, Statbel, Provincies in Cijfers, data.europa.eu, de World Bank. Eén waarschuwing: zoeken kost méér tijd dan je denkt. Vind je binnen het eerste lesuur niets dat de checklist doorstaat, val dan terug op het menu.

## Stap 2: formuleer vier onderzoeksvragen

Gekozen? Formuleer dan **vier onderzoeksvragen** bij je dataset. Schrijf ze voluit op — je zet ze straks letterlijk op je tabs.

Wat een goede onderzoeksvraag is, weet je ondertussen:

* **beantwoordbaar** met de data die je hebt — controleer of jouw dataset alle nodige dimensies bevat;
* **niet triviaal** — geen vraag waarop je in twee seconden het antwoord kan aflezen;
* **scherp geformuleerd** — niet "hoe zit het met minimumlonen in Europa", wel iets concreets waarop een grafiek een duidelijk antwoord geeft;
* de vier vragen **overlappen niet** — vier keer dezelfde vraag met een ander land levert geen punten op.

Zorg bovendien voor **variatie**. Een goed rapport bevat bijvoorbeeld:

* minstens één vraag over een **rangschikking of vergelijking** op één moment (welke landen scoren het hoogst? hoe positioneert België zich?);
* minstens één vraag over een **evolutie doorheen de tijd**;
* minstens één vraag die **groepen vergelijkt** (leeftijd, geslacht, woningtype, regio, ... — afhankelijk van je dataset).

## Stap 3: is je dataset bruikbaar? De checklist

Nu je weet wat je wil onderzoeken, toets je je keuze. In hoofdstuk 1 leerde je de **5 V's** van big data. Die gebruiken we hier als checklist: per V beantwoord je schriftelijk een paar vragen. Je antwoorden komen straks op de **verantwoordingstab** van je rapport.

| V | Vragen die je beantwoordt |
|---|---|
| **Volume** | Bevat de dataset genoeg data voor mijn vragen (genoeg landen, jaren, groepen)? En niet té veel: kan ik vóór het downloaden filteren, zodat het bestand hanteerbaar blijft? |
| **Velocity** | Hoe recent is de data, en hoe vaak wordt ze bijgewerkt? Volstaat dat voor mijn vragen? |
| **Variety** | In welk formaat krijg ik de data? Bevat ze álle dimensies die mijn vier vragen nodig hebben? Hoeveel opkuiswerk verwacht ik? |
| **Veracity** | Wie publiceert deze data, en hoe betrouwbaar is die bron? Zijn er ontbrekende waarden of waarschuwingen (zoals "experimental statistics")? |
| **Value** | Welk inzicht kan deze data opleveren? Beantwoordt ze mijn vragen écht, of alleen ongeveer? |

Botst een van je vragen op een "nee" — de dimensie ontbreekt, de jaren volstaan niet, de bron is twijfelachtig? Dan heb je twee opties: **pas je vraag aan**, of kies een andere dataset. Dat is geen mislukking; dat is precies hoe echt onderzoek werkt. Beter nu bijsturen dan na twee uur bouwen.

**Controlemoment:** kom je vier vragen en je ingevulde checklist even aan mij tonen vóór je begint te bouwen.

## Stap 4: haal je data op en maak ze klaar

Vanaf hier ken je het werk — deze keer zonder stappenplan. Denk aan wat je in de vorige opdrachten leerde:

* **Filter vóór je downloadt.** Selecteer alleen de landen, jaren en dimensies die je vragen nodig hebben. Alles wat je nu wegfiltert, hoef je straks niet op te kuisen.
* **Download als spreadsheet (.xlsx)** waar dat kan.
* **Kuis op in Excel**: metadatarijen weg, lege flag-kolommen weg, duidelijke kolomnamen en tabbladnamen, cellen met `:` leegmaken.
* **Importeer in PowerBI** en controleer de datatypes in de table view.
* Staan je jaren als kolommen naast elkaar? Dan moet je **unpivotten** in Power Query — kijk terug naar de tussenstap in hoofdstuk 6.

## Stap 5: bouw je rapport

Je rapport bevat **vijf tabs**.

**Tab 1 — Verantwoording.** Hierop staan:

* welke dataset je koos (naam + bron);
* je checklist uit stap 3, in eigen woorden — een paar zinnen per V volstaan.

**Tabs 2 t.e.m. 5 — je vier onderzoeksvragen.** Per tab:

* de vraag letterlijk bovenaan;
* één of meerdere visualisaties die de vraag beantwoorden;
* je interpretatie als tekstvak: wat zie je in de data, en wat betekent dat?

Kies per vraag bewust je grafiektype — de tabel bij competentie C5 op de pagina [Onderzoekscompetenties](01b_Onderzoekscompetenties.ipynb) helpt je kiezen — en denk terug aan hoofdstuk 4: een eerlijke grafiek misleidt niet.
