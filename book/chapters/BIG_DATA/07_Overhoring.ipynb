## Wat ga je doen?

Je werkt met een Eurostat-dataset over **dubbele armoede** in Europese huishoudens. Je maakt een PowerBI-rapport waarin je **vier onderzoeksvragen** beantwoordt. Drie vragen krijg je van mij. Eén vraag formuleer je zelf.

Per vraag werk je in een aparte tab. Bij elke tab hoort:

* één of meerdere visualisaties die de vraag helpen beantwoorden;
* een korte interpretatie waarin je uitlegt wat je in de data ziet.

## AI gebruiken

**Nieuw vanaf deze opdracht**: voor het schrijven van je **interpretaties** mag je een AI-assistent gebruiken (Claude, ChatGPT, Copilot, ...). Voorwaarden:

* AI mag je helpen bij het **interpreteren** van wat je in je grafieken ziet — niet bij het bouwen van de visualisaties zelf. Geef haar je cijfers mee, anders beschrijft ze een grafiek die ze zich inbeeldt.
* Je blijft verantwoordelijk voor de tekst. Controleer elk cijfer en elke bewering aan je eigen grafiek en dataset — niet bij een tweede AI, want die maakt dezelfde soort fouten. Hoe je dat aanpakt, leerde je op de pagina [Kritisch werken met AI](01c_Kritisch_werken_met_AI.ipynb).
* Kopieer nooit blindelings. Pas de tekst aan zodat hij past bij **jouw** grafieken en bij wat jij ziet.

## Stap 1: data ophalen op Eurostat

We werken met de dataset **Persons at two-fold risk of poverty by household type**. Je vindt die hier:

👉 [Eurostat — icw_pov_11](https://ec.europa.eu/eurostat/databrowser/view/icw_pov_11__custom_21680596/default/table)

### Hoe orden je je data?

* **Row**: Geopolitical entity (reporting)
* **Column**: Household composition

### Welke waarden kies je?

* **Geopolitical entity**: alle EU-landen, **behalve Italië, Zweden en het Verenigd Koninkrijk** (die hebben voor de meeste jaren geen data) en **behalve de groepsaggregaten** zoals "European Union", "Euro area" en de varianten daarvan (die overlappen met de individuele landen en zouden je grafieken vervuilen).
* **Time**: **alle beschikbare jaren** (2010, 2015 en 2020).
* **Household composition**: alle types (Total, One adult, One adult with dependent children, Two adults, Two adults with dependent children, Three or more adults, Three or more adults with dependent children).
* **Economic well-being indicator**: Income poverty and consumption poverty (staat al goed).
* **Unit of measure**: Percentage (staat al goed).

### Downloaden

Kies bij **Download**:

* **File format**: Spreadsheet (.xlsx)
* **Data scope**: All selected dimensions
* Vink **niet** "Consolidate all data sheets in a single one" aan. We willen 3 aparte tabbladen — één per jaar.

### Aandachtspunten

Onze dataset meet het percentage personen dat **dubbele armoede** loopt. Dat is een striktere armoededefinitie dan je gewend bent:

* **Inkomensarmoede** = je inkomen ligt onder 60% van het mediaaninkomen van je land.
* **Consumptiearmoede** = je consumptieuitgaven liggen onder 60% van de mediaanconsumptie van je land.
* **Dubbele armoede** = je voldoet aan **beide** criteria tegelijk — je verdient weinig én je consumeert weinig. Dit toont de mensen die *structureel* arm zijn, niet wie tijdelijk een laag inkomen heeft (bv. studenten die nog spaargeld hebben) of wie zuinig leeft uit overtuiging.

De cijfers zijn dus altijd **lager** dan klassieke armoedecijfers (AROPE), maar identificeren de meest kwetsbare groep scherper.

Verder: deze dataset is gemarkeerd als **experimental statistics**. Eurostat geeft zelf aan dat de meetmethode nog niet de robuustheid heeft van klassieke indicatoren. Wees voorzichtig met je conclusies.

## Stap 2: data klaarmaken in Excel en importeren in PowerBI

Je weet ondertussen wat te doen:

## De onderzoeksvragen

Werk elke vraag uit op een **aparte tab** in PowerBI. Geef die tab een duidelijke naam. Per tab toon je:

* één of meerdere visualisaties die je antwoord ondersteunen;
* je interpretatie als tekstvak op de tab (AI mag je daarbij helpen — zie de inleiding).

Lees elke vraag goed voor je begint. Bedenk welk type grafiek het beste past, welke tabel je gebruikt en of je extra filters nodig hebt.

### Vraag 1 — Hoogste dubbele armoede in 2020

**Welke EU-landen kennen in 2020 de hoogste totale dubbele armoede? Hoe positioneert België zich in dat klassement?**

Denk na over:

* Welke tabel heb je nodig?
* Hoe maak je België visueel herkenbaar?
* Zijn er landen die je positioneel verrassen (rijk land met hoge armoede, of omgekeerd)?

### Vraag 2 — Evolutie in België en buurlanden

**Hoe evolueert de algemene dubbele armoede in België en zijn buurlanden (Duitsland, Frankrijk, Luxemburg, Nederland) tussen 2010, 2015 en 2020?**

Denk na over:

* Hoe toon je drie tijdsmomenten en vijf landen samen op een leesbare manier?
* Volgen alle buurlanden dezelfde trend, of gaat België een eigen weg?
* Welke factoren zou je kunnen aanhalen om een sterke stijging of daling te verklaren?

### Vraag 3 — Beschermt een tweede ouder?

**Beschermt een tweede ouder tegen armoede bij gezinnen met kinderen? Vergelijk eenoudergezinnen, tweeoudergezinnen en uitgebreide gezinnen in de EU in 2020.**

Denk na over:

* Welke drie kolommen uit je 2020-tabel gebruik je?
* Is het effect even groot in elk land?
* Welk verschil zie je tussen West- en Oost-Europa?

### Vraag 4 — Eén zelfgekozen vraag

Formuleer zelf één onderzoeksvraag die je met deze dataset kan beantwoorden. Werk ze op dezelfde manier uit: visualisatie + interpretatie op een eigen tab.

**Schrijf de vraag duidelijk bovenaan de tab.**

Een goede onderzoeksvraag:

* is **beantwoordbaar** met de data die je hebt — kijk dus na of jouw selectie alle nodige dimensies bevat;
* is **niet triviaal** — geen vraag waarop je in twee seconden het antwoord kan aflezen;
* is **scherp geformuleerd** — niet "hoe zit het met armoede in Europa", wel iets concreets waarop een grafiek een duidelijk antwoord geeft;
* voegt iets toe aan wat je in de eerste drie vragen al onderzocht hebt.

Met drie jaartallen en zeven huishoudtypes zit er veel meer in de data dan deze opdracht behandelt. Denk bijvoorbeeld aan: evolutie per huishoudtype, kinderarmoede zonder ouders erbij (one adult vs one adult with children), of een landengroep die jou interesseert.

De **kwaliteit van je vraag** wordt apart beoordeeld. Een zwakke vraag, ook al is de uitwerking netjes, levert minder punten op.
