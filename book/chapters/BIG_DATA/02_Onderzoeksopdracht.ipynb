# Onderzoeksopdracht: werkzaamheidsgraad in de Europese Unie

Vorige week kreeg je nog stap voor stap te horen wat je moest doen. Deze keer werk je een stuk zelfstandiger. Je kiest zelf welke data je downloadt, hoe je die ordent, en welke grafieken je maakt. Je moet dus **zelf nadenken** over keuzes die vorige week nog voor jou werden gemaakt.

## Wat ga je doen?

Je werkt met een dataset van Eurostat over de **werkzaamheidsgraad** in de Europese Unie. Je maakt een PowerBI-rapport waarin je **zes onderzoeksvragen** beantwoordt. Vier vragen krijg je van mij. Twee vragen formuleer je zelf.

Per vraag werk je in een aparte tab. Bij elke tab hoort:

* één of meerdere visualisaties die de vraag helpen beantwoorden;
* een korte interpretatie waarin je in eigen woorden uitlegt wat je in de data ziet.

## Hoe wordt dit beoordeeld?

Op deze opdracht staan punten. De rubric vind je in MS Teams. Lees die zeker **vóór** je begint, want daarin staat wat ik verwacht. Een paar zaken om alvast te onthouden:

* Je datavoorbereiding telt mee. Slordig opgekuiste data kost punten.
* De presentatie van je rapport telt mee. Eén vraag per tab, duidelijke titels, verzorgde layout.
* Voor elke vraag tellen visualisatie én interpretatie. Een mooie grafiek zonder uitleg is niet voldoende.
* Voor de twee zelfgekozen vragen telt ook de **kwaliteit van de vraag zelf**. Een vraag waarop je in twee seconden "ja" of "nee" kan antwoorden, is niet relevant genoeg.
* Vragen die je niet uitwerkt, leveren geen punten op. Wie traag werkt, mag thuis verder werken.

## Tijd

Je krijgt twee lesuren. Wie niet klaar is, werkt thuis verder en levert in via MS Teams.

## Stap 1: data ophalen op Eurostat

We werken met de dataset **Employment rates by educational attainment level and citizenship**. Je vindt die hier:

👉 [Eurostat](https://ec.europa.eu/eurostat/databrowser/view/lfsa_ergaedn/default/table)

Vorige week kreeg je precies te horen welke filters je moest aanvinken. Deze week krijg je wel duidelijke aanwijzingen welke data je nodig hebt, maar je moet zelf nadenken over hoe je die ordent en welke waarden je binnen elke dimensie kiest. Bekijk dus eerst de zes onderzoeksvragen onderaan deze opdracht voor je begint te selecteren.

### Hoe orden je je data?

* Je hebt twee belangrijke dimensies nodig: **Geopolitical entity** en **ISCED** (onderwijsniveau). Welke daarvan plaats je in de **rij**, en welke in de **kolom**?

### Welke waarden kies je?

* **ISCED**: welke opleidingsniveaus heb je nodig? Kijk naar de onderzoeksvragen — niet elk niveau is relevant, en sommige niveaus zijn een totaal van andere.
* **Age class**: dit hoort bij de dimensies. Welke leeftijdsgroepen heb je nodig? Ook hier: kijk naar de vragen.
* **Country of citizenship**: kies **Total**.
* **Time**: kies **2025**.

### Aandachtspunten

Onze dataset bevat **werkzaamheidsgraad**, niet werkloosheid zoals vorige week. Dat is een belangrijk verschil:

* **Werkloosheidsgraad** = aandeel van de mensen die werk zoeken, maar geen werk hebben.
* **Werkzaamheidsgraad** = aandeel van de mensen in een bevolkingsgroep die effectief werken.

Studenten, gepensioneerden en mensen die niet actief werk zoeken tellen bij de werkzaamheidsgraad mee in de **noemer**, maar bij de werkloosheidsgraad niet. Daardoor kunnen die cijfers er soms onverwacht uitzien (bijvoorbeeld lage werkzaamheidsgraad bij 15–24-jarigen, omdat veel jongeren nog studeren). Hou daar rekening mee bij je interpretatie.

## Stap 2: data klaarmaken in Excel

Het bestand dat Eurostat je geeft, is **niet meteen klaar** voor PowerBI. Je moet het eerst opkuisen. Vorige week deden we dat samen. Deze keer moet je zelf bepalen wat er moet gebeuren.

Open je gedownloade Excel en stel jezelf deze vragen:

* Staan er **rijen of kolommen bovenaan** die niet bij de eigenlijke data horen? Wat doe je daarmee?
* Zijn de **kolomnamen** duidelijk en bruikbaar? Zou PowerBI ermee overweg kunnen?
* Zitten er **lege kolommen** tussen de gegevens? Waarom staan die er, en wat doe je ermee?
* Zijn er **cellen met `:` of andere niet-numerieke tekens** in plaats van een getal? Wat betekenen die volgens jou? Hoe los je dat op?
* Klopt het **type** van elke kolom (tekst, getal, percentage, …)?
* Werk je met **één tabblad** of meerdere? Welke namen geef je aan je tabbladen?

Werk netjes. Sla je bestand op met een duidelijke naam.

> **Tip:** Wie slordig werk inlevert in Excel, krijgt vaak ook fouten in PowerBI. Investeer dus tijd in deze stap.

## Stap 3: importeer in PowerBI

Open PowerBI Desktop en importeer je Excel-bestand. Controleer in de **table view**:

* Zijn alle kolommen correct ingelezen?
* Klopt het datatype van elke kolom?
* Zijn er kolommen die je niet nodig hebt? Verwijder die gerust.
* Zijn de tabelnamen duidelijk?

Pas dit aan voor je begint te visualiseren. Achteraf is dat veel lastiger.

## Tussenstap: berekende kolommen en metingen

Voor sommige vragen heb je informatie nodig die niet rechtstreeks in je data staat. Bijvoorbeeld: het verschil tussen werkzaamheidsgraad bij hoog- en laaggeschoolden. Dat moet je dan zelf laten berekenen door PowerBI.

Vorige week werkten we hier al even mee. Een korte herhaling:

### Berekende kolom

Een **berekende kolom** voegt een nieuwe kolom toe aan een tabel. Voor elke rij wordt het resultaat berekend. Dat is handig wanneer je per rij iets wil vergelijken — bijvoorbeeld per land het verschil tussen twee onderwijsniveaus.

Je maakt er één via **Modelling → New column**. Voorbeeld:

```DAX
Diplomakloof = 'Mijntabel'[Hoger Onderwijs] - 'Mijntabel'[Laaggeschoold]
```

Hierin zijn `Hoger Onderwijs` en `Laaggeschoold` kolommen die jij in je tabel hebt staan. De namen hangen dus af van wat jij gedownload hebt en hoe je je kolommen noemt. `Mijntabel` is de naam van je tabel.

### Meting

Een **meting** berekent één getal op basis van de gegevens die zichtbaar zijn in je grafiek of filter. Dat is handig wanneer je een totaal, gemiddelde of verschil wil tonen dat afhangt van wat de lezer geselecteerd heeft.

Je maakt er één via **Modelling → New measure**. Voorbeeld:

```DAX
Gemiddelde diplomakloof = AVERAGE('Mijntabel'[Diplomakloof])
```

### Wanneer wat?

* Heb je voor elke rij apart een waarde nodig? → **berekende kolom**
* Heb je één samenvattend getal nodig dat verandert met je filter? → **meting**

Voor deze opdracht heb je waarschijnlijk vooral berekende kolommen nodig. Maar denk er bij elke vraag even over na.

## De onderzoeksvragen

Werk elke vraag uit op een **aparte tab** in PowerBI. Geef die tab een duidelijke naam. Per tab toon je:

* één of meerdere visualisaties die je antwoord ondersteunen;
* je interpretatie als tekstvak op de tab.

Lees elke vraag goed voor je begint. Vraag je af welk type grafiek het beste past, en of je extra berekeningen nodig hebt. Soms moet je experimenteren voor je iets bruikbaar hebt — dat hoort erbij.

### Vraag 1 — Werkzaamheidsgraad bij laaggeschoolden

**Welke EU-landen hebben de hoogste en laagste werkzaamheidsgraad bij laaggeschoolden (ED0-2) in de leeftijdsgroep 25–49 jaar? Hoe positioneert België zich in dat klassement?**

Denk na over:

* Welke visualisatie laat een ranking goed zien?
* Hoe maak je België visueel herkenbaar?
* Wat valt op aan de uitschieters?

### Vraag 2 — De diplomakloof

**Hoe groot is de "diplomakloof" per land? Dat is het verschil in werkzaamheidsgraad tussen hooggeschoolden (ED5-8) en laaggeschoolden (ED0-2). In welke landen is die kloof het grootst, en in welke het kleinst?**

Voor deze vraag heb je een **berekende kolom** nodig. Bekijk de tussenstap hierboven nog eens als je niet meer weet hoe dat werkt.

Denk na over:

* Voor welke leeftijdsgroep maak je deze vergelijking? Of toon je meerdere?
* Wat zegt een grote diplomakloof eigenlijk over een land?

### Vraag 3 — Algemeen versus beroepsgericht secundair

**Loont een beroepsgerichte opleiding op secundair niveau meer dan een algemene opleiding? Vergelijk de werkzaamheidsgraad bij ED34_44 (algemeen secundair) met ED35_45 (beroepsgericht secundair) over de EU-landen heen.**

Voor deze vraag heb je waarschijnlijk **meerdere visualisaties** nodig. Denk na over hoe je de twee groepen naast elkaar zet.

Denk na over:

* Wat is het meest opvallende verschil tussen beide?
* Geldt het patroon voor alle landen, of zijn er uitzonderingen?
* Speelt leeftijd hierin een rol?

### Vraag 4 — Leeftijd en opleidingsniveau

**Speelt leeftijd anders mee bij verschillende opleidingsniveaus? Vergelijk de werkzaamheidsgraad tussen de drie leeftijdsgroepen (15–24, 25–49, 50–64) voor laag-, midden- en hooggeschoolden.**

Ook hier heb je waarschijnlijk **meerdere visualisaties** nodig — of één goed gekozen visual die de drie opleidingsniveaus en de drie leeftijdsgroepen tegelijk toont.

Denk na over:

* Bij welke leeftijdsgroep is het effect van opleiding het grootst?
* Hoe verklaar je de cijfers bij 15–24-jarigen? (Denk terug aan het verschil tussen werkzaamheidsgraad en werkloosheidsgraad.)

### Vraag 5 en 6 — Twee zelfgekozen vragen

Formuleer zelf twee onderzoeksvragen die je met deze dataset kan beantwoorden. Werk ze op dezelfde manier uit als de vorige vragen: visualisatie + interpretatie op een eigen tab.

**Schrijf de vraag duidelijk bovenaan de tab.**

Een goede onderzoeksvraag:

* is **beantwoordbaar** met de data die je hebt — kijk dus na of jouw selectie alle nodige dimensies bevat;
* is **niet triviaal** — geen vraag waarop je in twee seconden het antwoord kan aflezen;
* is **scherp geformuleerd** — niet "hoe zit het met opleiding in Europa", wel iets concreets waarop een grafiek een duidelijk antwoord geeft;
* voegt iets toe aan wat je in de eerste vier vragen al onderzocht hebt.

De **kwaliteit van je vraag** wordt apart beoordeeld. Een zwakke vraag, ook al is de uitwerking netjes, levert minder punten op.
