# Onderzoekscompetenties: van vraag naar conclusie

In dit deel leer je meer dan een programma bedienen. Elke opdracht doorloopt dezelfde cyclus: je vertrekt van een **onderzoeksvraag**, je zoekt en beoordeelt **data**, je kiest er de juiste **grafiek** bij en je trekt een **conclusie**. Die vaardigheden noemen we **onderzoekscompetenties**. Je hebt ze niet alleen in dit vak nodig: ook de onderzoeksopdracht in de lessen economie steunt erop, net als elk verslag waarin je met cijfers argumenteert.

Deze pagina beschrijft de zes competenties, toont waar je ze in dit deel oefent en geeft de rubric waarmee je werk beoordeeld wordt. Kom er bij elke opdracht op terug: de rubric vertelt je vooraf waar je op moet letten.

:::{admonition} Leerdoelen
:class: tip

Na dit deel kan je:
- een scherpe, beantwoordbare onderzoeksvraag formuleren bij een thema
- een dataset zoeken die bij je vraag past en de selectie beperken tot wat je nodig hebt
- een dataset beoordelen op bron, actualiteit en definities, met de 5 V's als leidraad
- ruwe data klaarmaken voor analyse
- de grafiekvorm kiezen die bij je vraag past en die eerlijk weergeven
- uit een grafiek een genuanceerde conclusie trekken die je vraag beantwoordt
:::

## 1. De onderzoekscyclus

Onderzoek met data verloopt in stappen die telkens terugkomen:

1. **Vraag** — wat wil je precies weten?
2. **Data zoeken** — welke dataset bevat het antwoord?
3. **Data beoordelen** — kan je die dataset vertrouwen, en meet ze wat jij bedoelt?
4. **Data klaarmaken** — ruwe data opkuisen tot een tabel waarmee je kan werken.
5. **Visualiseren** — de grafiek kiezen die je vraag het duidelijkst beantwoordt.
6. **Interpreteren en concluderen** — wat zie je, en wat betekent dat voor je vraag?

Een rechte lijn is dat zelden. Een dataset die tegenvalt, stuurt je vraag bij; een grafiek die niets toont, stuurt je grafiekkeuze bij. Terugkeren naar een vorige stap is geen mislukking — het is precies hoe onderzoek werkt. Wat wél vaststaat: elke stap is een competentie die je apart kan oefenen en beoordelen. Hieronder krijgen ze elk een code, van C1 tot C6, zodat we er in de hoofdstukken kort naar kunnen verwijzen.

## 2. De zes competenties

### C1 — Onderzoeksvraag formuleren

Een onderzoek begint niet bij een dataset, maar bij een vraag. Een goede onderzoeksvraag:

* is **beantwoordbaar** met data die bestaat en die je kan krijgen;
* is **niet triviaal** — geen vraag waarop je in twee seconden "ja" of "nee" kan antwoorden;
* is **scherp geformuleerd** — niet "hoe zit het met werkloosheid in Europa", wel "in welke EU-landen is de jeugdwerkloosheid sinds 2015 het sterkst gedaald?";
* past bij het **thema** en voegt iets toe aan wat je al weet.

De meeste vragen horen bij één van vier types. Herken je het type, dan weet je meteen welke data en welke grafiek je nodig hebt (zie C5):

| Type vraag | Voorbeeld |
|---|---|
| **Rangschikking of vergelijking** op één moment | Welke landen scoren het hoogst? Hoe positioneert België zich? |
| **Evolutie** doorheen de tijd | Stijgt of daalt het cijfer sinds 2010? Zijn er breukpunten? |
| **Groepen vergelijken** | Verschilt het cijfer tussen mannen en vrouwen, leeftijdsgroepen, opleidingsniveaus? |
| **Verband** tussen twee grootheden | Hangt een hoger opleidingsniveau samen met een lagere werkloosheid? |

### C2 — Data zoeken

Bij een vraag hoort een dataset met de juiste **dimensies**: de landen, jaren en groepen waarover je vraag gaat. Zoek dus niet naar "iets over werkloosheid", maar naar een dataset met werkloosheid **per opleidingsniveau, per jaar, voor België**. De portalen uit hoofdstuk 1 (Eurostat, Statbel, Provincies in Cijfers, de World Bank, ...) laten je meestal filteren vóór je downloadt. Doe dat: alles wat je nu wegfiltert, hoef je straks niet op te kuisen, en een bestand met precies de juiste dimensies is het bewijs dat je je vraag goed begrepen hebt.

### C3 — Data beoordelen

Niet elke dataset is bruikbaar, ook al lijkt ze te passen. Drie vragen stel je altijd, nog vóór je iets downloadt:

* **Bron** — wie publiceert deze cijfers, en met welk doel? Een statistiekbureau zoals Eurostat of Statbel is iets anders dan een bedrijf, een belangengroep of een post op sociale media. Kan je de cijfers vergelijken met een tweede bron?
* **Actualiteit** — wanneer werd de data laatst bijgewerkt, en dekt ze de periode waarover je vraag gaat? Cijfers van 2015 zeggen weinig over "vandaag".
* **Definities** — wat meet het cijfer precies? Werkloosheidsgraad is niet hetzelfde als werkzaamheidsgraad, directe belastingen zijn niet de hele belastingdruk, een mediaan is geen gemiddelde en een index is geen prijs. Wie de definitie niet leest, trekt de verkeerde conclusie uit een correcte grafiek.

Kijk daarnaast naar **ontbrekende waarden** (lege cellen, een `:` in Eurostat) en naar **waarschuwingen** van de bron zelf, zoals "experimental statistics". De **5 V's** uit hoofdstuk 1 bundelen dit alles in één checklist; bij de eindopdracht vul je die schriftelijk in.

### C4 — Data klaarmaken

Ruwe data is bijna nooit meteen bruikbaar: metadatarijen bovenaan, lege kolommen, jaren als kolomkoppen, getallen die als tekst gelezen worden. Klaarmaken betekent: overbodige rijen en kolommen weg, duidelijke namen voor tabbladen en kolommen, het juiste datatype per kolom en ontbrekende waarden correct verwerkt. Dit is de **Transform**-stap van ETL. Ze lijkt saai, maar elke fout die je hier laat zitten, zit straks in álle grafieken.

### C5 — Visualiseren

Een grafiek is een antwoord op een vraag. Kies het grafiektype dus op basis van het type vraag uit C1:

| Wat wil je tonen? | Geschikte grafiek | Let op |
|---|---|---|
| Een **rangschikking** of vergelijking op één moment | Staafdiagram (liggend bij veel categorieën), aflopend gesorteerd | y-as bij nul; maak België herkenbaar met een aparte kleur |
| Een **evolutie** doorheen de tijd | Lijndiagram; kolommen bij een handvol tijdstippen | toon de volledige relevante periode; één y-as |
| **Aandelen** van één geheel | Cirkeldiagram of 100% gestapeld kolomdiagram | de delen tellen op tot 100%; hou het bij een zestal schijven |
| **Groepen** naast elkaar, eventueel doorheen de tijd | Geclusterd kolomdiagram of meerdere lijnen in één grafiek | niet meer dan een handvol reeksen |
| Eén **kerncijfer** | Kaart (card) | zet er altijd context naast |
| Een **verband** tussen twee grootheden | Spreidingsdiagram | een verband is nog geen oorzaak |

Eerlijk tonen betekent dat je grafiek de checklist uit hoofdstuk 4 doorstaat: geen afgekapte as zonder reden, geen cherry-picking van de periode, geen kleuren of titels die meer beweren dan de cijfers. Kan je uitleggen **waarom** je voor dit grafiektype koos? Dan zit je goed.

### C6 — Interpreteren en concluderen

Een grafiek zonder uitleg is geen antwoord. Een goede interpretatie bestaat uit drie lagen:

1. **Wat zie je?** Het patroon (stijgt, daalt, blijft stabiel), de volgorde, de uitschieters. Gebruik cijfers waar ze het verhaal scherper maken, maar som niet elke waarde op.
2. **Wat betekent dat voor je vraag?** Beantwoord de vraag letterlijk, in volledige zinnen.
3. **Hoe zeker ben je?** Benoem de beperkingen van de data (C3), alternatieve verklaringen, en het verschil tussen een verband en een oorzaak.

> In deze tabel is de werkloosheid in bijna alle jaren het hoogst bij mensen zonder diploma secundair, en het laagst bij mensen met een diploma hoger onderwijs. Dat wijst op een duidelijk verband tussen opleidingsniveau en werkloosheid. Of dat verband ook geldt voor 65-plussers, kan deze dataset niet zeggen: zij zitten er niet in.

Zo'n conclusie is kort, beantwoordt de vraag en geeft haar grenzen aan.

## 3. Waar oefen je wat?

De competenties komen niet allemaal tegelijk aan bod. Je begint begeleid en werkt steeds zelfstandiger, tot je bij de eindopdracht de volledige cyclus zelf doorloopt. Elk hoofdstuk begint met een kader **"Onderzoekscompetenties in dit hoofdstuk"** dat aangeeft wat je daar oefent.

| Hoofdstuk | C1 Vraag | C2 Zoeken | C3 Beoordelen | C4 Klaarmaken | C5 Visualiseren | C6 Concluderen |
|---|---|---|---|---|---|---|
| 2. Aan de slag met Power BI | — | — | — | begeleid | begeleid | — |
| 3. Onderzoeksopdracht: werkloosheid | — | begeleid | begeleid | begeleid | begeleid | begeleid |
| 4. Liegen met grafieken | — | — | zelfstandig | — | zelfstandig | zelfstandig |
| 5. Onderzoeksopdracht: werkzaamheidsgraad | zelfstandig | deels zelfstandig | begeleid | zelfstandig | zelfstandig | zelfstandig |
| 6. Onderzoeksopdracht: belastingdruk | zelfstandig | begeleid | begeleid | deels zelfstandig | zelfstandig | zelfstandig |
| 7. Overhoring | zelfstandig | begeleid | deels zelfstandig | zelfstandig | zelfstandig | zelfstandig |
| 8. Eindopdracht | zelfstandig | zelfstandig | zelfstandig | zelfstandig | zelfstandig | zelfstandig |

*begeleid* = de stappen worden je voorgezegd; *deels zelfstandig* = je krijgt aanwijzingen, maar maakt de keuzes zelf; *zelfstandig* = je doet het zonder stappenplan; — = komt in dat hoofdstuk niet aan bod.

## 4. De rubric

Met deze rubric wordt je werk in de onderzoeksopdrachten, de overhoring en de eindopdracht beoordeeld. Per competentie zie je wat "beheerst" betekent — en wat er nog aan ontbreekt bij de lagere niveaus. Welke competenties in een opdracht meetellen, staat telkens in het kader bovenaan het hoofdstuk.

| Competentie | Nog niet | Op weg | Beheerst |
|---|---|---|---|
| **C1** Onderzoeksvraag formuleren | De vraag is vaag of triviaal, of ze is niet te beantwoorden met de beschikbare data. | De vraag is beantwoordbaar, maar te breed, te eenvoudig of een herhaling van een vorige vraag. | De vraag is scherp, beantwoordbaar en niet triviaal; ze past bij de data en voegt iets toe. |
| **C2** Data zoeken | De dataset past niet bij de vraag: dimensies die de vraag nodig heeft, ontbreken. | De dataset past, maar de selectie is te ruim of te eng: overbodige data, of net te weinig. | De dataset bevat precies de dimensies die de vraag nodig heeft; de selectie gebeurde al vóór de download. |
| **C3** Data beoordelen | Bron, actualiteit en definities zijn niet nagekeken; het cijfer wordt verkeerd begrepen. | Bron en actualiteit zijn benoemd, maar de definities zijn maar half gelezen of de beperkingen blijven onvermeld. | Bron, actualiteit en definities zijn nagekeken en benoemd; de beperkingen van de data staan in de verantwoording of de interpretatie. |
| **C4** Data klaarmaken | De data is ruw gebleven: metadatarijen, lege kolommen, verkeerde datatypes. | De data is opgekuist, met slordigheden: onduidelijke namen, getallen die als tekst gelezen worden. | De data is proper: duidelijke tabbladen en kolomnamen, juiste datatypes, ontbrekende waarden correct verwerkt. |
| **C5** Visualiseren | Het grafiektype past niet bij de vraag, of de grafiek misleidt (afgekapte as, cherry-picking, verkeerd type, ...). | Het grafiektype past, maar titel, assen of legende zijn onvolledig of onhandig. | Het grafiektype past bij de vraag en is eerlijk; titel en labels zijn duidelijk; de keuze kan verantwoord worden. |
| **C6** Interpreteren en concluderen | Er is geen interpretatie, of de tekst beschrijft de grafiek niet correct. | De tekst beschrijft wat te zien is, maar beantwoordt de vraag niet expliciet of zonder nuance. | De conclusie beantwoordt de vraag in volledige zinnen, benoemt patroon én uitschieters, en geeft nuance: beperkingen, verband versus oorzaak. |

:::{admonition} Dezelfde rubric in de lessen economie
:class: note

Deze rubric is niet aan Power BI gebonden. Ze geldt evengoed voor de onderzoeksopdracht in de lessen economie — ook als je daar met Excel werkt of je grafiek in een verslag opneemt. Dezelfde zes competenties, dezelfde criteria: wat je hier oefent, telt daar mee.
:::

### Checklist voor je indient

- [ ] **C1** Elke onderzoeksvraag staat letterlijk in mijn rapport en is scherp, beantwoordbaar en niet triviaal.
- [ ] **C2** Mijn dataset bevat precies de dimensies die mijn vragen nodig hebben — niet meer, niet minder.
- [ ] **C3** Ik heb bron, actualiteit en definities nagekeken en benoem de beperkingen van de data.
- [ ] **C4** Mijn data is proper: duidelijke namen, juiste datatypes, geen metadatarijen of lege kolommen.
- [ ] **C5** Elk grafiektype past bij de vraag en doorstaat de checklist van hoofdstuk 4.
- [ ] **C6** Elke grafiek heeft een interpretatie die de vraag beantwoordt, met nuance.
