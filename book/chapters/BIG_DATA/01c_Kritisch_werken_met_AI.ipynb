# Kritisch werken met AI: hulpmiddel, geen bron

Grote kans dat je tijdens een onderzoeksopdracht een AI-assistent inschakelt: om een begrip uit te leggen, om mee te denken over een onderzoeksvraag, om een tekst vlotter te maken. Dat mag, en het kan je onderzoek beter maken. Maar een AI-assistent heeft één eigenschap die je nooit mag vergeten: ze klinkt altijd even zeker, ook wanneer ze cijfers, bronnen of verklaringen **verzint**. Wie dat niet doorheeft, zet zonder het te merken foute cijfers in een rapport — met een bron erbij die niet bestaat.

Deze les leert je twee dingen: **goede vragen stellen**, zodat je bruikbare antwoorden krijgt, en **antwoorden controleren**, zodat je alleen overneemt wat klopt. Daarna bekijken we per stap van de onderzoekscyclus waar een AI-assistent je helpt, en waar je ze beter niet inzet.

:::{admonition} Welke AI-assistent?
:class: note

Deze les gaat niet over één bepaald programma. Alles wat hier staat, geldt voor elke AI-assistent (chatbot) die je gebruikt, op school of thuis. De oefeningen kan je bovendien maken **zonder account**: de gesprekken die je moet beoordelen, staan hier al uitgeschreven.
:::

:::{admonition} Leerdoelen
:class: tip

Na deze les kan je:
- een AI-assistent een vraag stellen met voldoende context, en doorvragen tot het antwoord bruikbaar is
- de beweringen in een AI-antwoord één voor één controleren aan de hand van de echte databron, en benoemen wat niet te controleren valt
- uitleggen waarom een AI-assistent bronnen en cijfers kan verzinnen, en waarom een tweede AI geen controle is
- per stap van de onderzoekscyclus zeggen waarvoor je AI wél en niet gebruikt
- in je rapport eerlijk vermelden hoe je AI gebruikte
:::

:::{admonition} Onderzoekscompetenties in dit hoofdstuk
:class: note

- **C1 Onderzoeksvraag formuleren** — *deels zelfstandig*: je laat een AI-assistent meebrainstormen over vragen bij een dataset en toetst haar voorstellen zelf aan de criteria voor een goede onderzoeksvraag.
- **C3 Data beoordelen** — *deels zelfstandig*: wat een AI-assistent beweert, behandel je als een bron die je controleert — bewering per bewering, naast de echte data. Je krijgt het gesprek en de bron; het oordeel vel je zelf.
- **C6 Interpreteren en concluderen** — *deels zelfstandig*: je mag een AI-assistent je formulering laten verbeteren, maar de analyse en de conclusie blijven van jou.

Wat de codes betekenen en hoe ze beoordeeld worden, lees je op de pagina [Onderzoekscompetenties](01b_Onderzoekscompetenties.ipynb).
:::

## 1. Wat een AI-assistent wel en niet is

Een AI-assistent is een **taalmodel**: een programma dat getraind is op enorme hoeveelheden tekst en dat telkens het meest waarschijnlijke volgende woord voorspelt. Daardoor schrijft ze vlot, in elke stijl, over elk onderwerp. Maar ze **weet** niets zoals een databank iets weet. Ze zoekt geen cijfer op in een tabel; ze produceert een tekst die eruitziet als een antwoord. Meestal klopt die tekst. Soms niet — en dat zie je er niet aan.

Drie gevolgen die je moet kennen:

* **Ze verzint.** Vooral cijfers, bronnen, citaten, rapporttitels en links: precies de dingen waar het in een onderzoek op aankomt. Dat heet **hallucineren**. Een verzonnen bron ziet er even overtuigend uit als een echte, met jaartal en publicatienummer erbij.
* **Ze kent jouw data niet.** Tenzij je je tabel meegeeft, heeft ze je Eurostat-download nooit gezien. Vraag je "wat zie je in mijn grafiek?" zonder de cijfers te geven, dan beschrijft ze een grafiek die ze zich inbeeldt.
* **Ze twijfelt niet.** Een statistiekbureau zet een `:` bij een ontbrekende waarde en waarschuwt bij *experimental statistics*. Een AI-assistent antwoordt met dezelfde zekerheid als ze gelijk heeft en als ze fout zit. Haar toon zegt niets over haar betrouwbaarheid.

Vergelijk het met een welbespraakte klasgenoot die alles gelezen heeft, maar nooit iets nakijkt. Prima om mee te brainstormen. Niet iemand van wie je cijfers overneemt zonder ze zelf te controleren.

In de woorden van competentie C3: een AI-assistent is **geen bron**. Een bron is wie de cijfers *meet en publiceert* — Eurostat, Statbel, de World Bank. Een AI-assistent *genereert* tekst over die cijfers. Alles wat ze zegt, toets je dus aan de bron, net zoals je dat met een krantenartikel of een post op sociale media zou doen.

## 2. Goede vragen stellen

De kwaliteit van een antwoord hangt sterk af van de kwaliteit van je vraag. Drie gewoontes maken het verschil.

### Geef context

Een AI-assistent weet niet wie je bent, wat je onderzoekt of welke data je hebt. Zeg het haar. Een bruikbare vraag bevat:

* **wie je bent en wat je doet** — een leerling die een onderzoeksopdracht maakt over ...;
* **je onderzoeksvraag** — letterlijk;
* **je data** — welke dataset, welke dimensies (landen, jaren, groepen), welke definitie en eenheid, en liefst de cijfers zelf (plak je tabel erbij);
* **wat je terug wilt** — een uitleg van een begrip, drie voorstellen van vragen, een verbeterde formulering, ...

Vergelijk deze twee vragen over de dataset uit [hoofdstuk 3](03_Onderzoeksopdracht.ipynb):

> **Zwak:** Leg de werkloosheid in België uit.

> **Beter:** Ik ben een leerling van het vijfde jaar en maak een onderzoeksopdracht met een Eurostat-dataset over de werkloosheidsgraad in België, per opleidingsniveau, voor 20- tot 64-jarigen, 2015–2025. Mijn onderzoeksvraag is: "Welk verband zie je tussen opleidingsniveau en werkloosheid?" Hieronder plak ik mijn tabel. Beschrijf in drie zinnen het patroon dat je in deze cijfers ziet, zonder verklaringen te geven die niet in de cijfers zitten.

Op de zwakke vraag krijg je een algemeen stukje tekst dat op elk land past. Op de betere vraag krijg je een antwoord over *jouw* cijfers — dat je vervolgens nog altijd controleert.

### Vraag door

Het eerste antwoord is een ruwe versie, geen eindresultaat. Goede vervolgvragen:

* "Waarop baseer je dat?" — en kijk of het antwoord naar iets concreets verwijst dat je kan nakijken.
* "Welke aanname maak je hier?"
* "Geef een tegenargument." — een AI-assistent schrijft even vlot het tegendeel; dat helpt je om nuance te vinden voor je conclusie (C6).
* "Wat kan je met deze cijfers *niet* besluiten?"
* "Wat zou een statisticus hierop aanmerken?"

### Itereer

Bevalt het antwoord niet? Verbeter je vraag in plaats van hetzelfde nog eens te vragen. Geef ontbrekende context, beperk de vraag, vraag een ander formaat (een lijst, een tabel, drie zinnen). Een gesprek met een AI-assistent verloopt zoals onderzoek zelf: in rondes.

### Oefening 1 — Verbeter de vraag

Herschrijf elk van deze vragen zodat een AI-assistent er iets bruikbaars mee kan doen. Voeg context toe: wie je bent, welke data je hebt, wat je precies terug wilt. Je hebt er geen AI-assistent voor nodig — schrijf de betere vraag gewoon uit.

1. "Leg deze grafiek uit."
2. "Welk grafiektype moet ik gebruiken?"
3. "Is Eurostat betrouwbaar?"

:::{admonition} Oplossing oefening 1
:class: dropdown

Voorbeelden van betere vragen — jouw formulering mag anders zijn, als de context er maar in zit:

1. "Ik maak een lijndiagram in Power BI met de werkloosheidsgraad in België (20–64 jaar) per opleidingsniveau, 2015–2025. Dit zijn de cijfers: [tabel]. Beschrijf in drie zinnen wat er te zien is: de volgorde van de lijnen, de trend en de uitschieters. Geen verklaringen." Een AI-assistent kan je grafiek niet zien; je geeft haar de cijfers.
2. "Mijn onderzoeksvraag is: 'In welke EU-landen steeg het minimumloon sinds 2015 het sterkst?' Ik heb per land het minimumloon in euro voor 2015 en 2025. Welk grafiektype past bij deze vraag, en waarom? Geef twee opties met voor- en nadelen." Daarna leg je het antwoord naast de tabel bij competentie C5 en de checklist van [hoofdstuk 4](04_Liegen_met_grafieken.ipynb).
3. Deze vraag stel je beter niet aan een AI-assistent, maar aan de bron zelf: wie publiceert de cijfers, hoe worden ze gemeten, wanneer werden ze bijgewerkt? Wat je wél kan vragen: "Wat betekent 'experimental statistics' bij een Eurostat-dataset, en waar vind ik de methodologie van de dataset lfsa_urgaed?" — en dan lees je die methodologie zelf.
:::

## 3. Antwoorden controleren

Een antwoord van een AI-assistent bestaat uit **beweringen**. Elke bewering controleer je apart, en je deelt ze in drie soorten in:

| Soort bewering | Hoe controleer je ze? |
|---|---|
| **Een cijfer** — "in 2024 lag de werkloosheid rond 11,7 %" | Leg het naast de bron: je eigen tabel, of de databank zelf. Klopt het cijfer, het jaar én de definitie? |
| **Een verwijzing** — een rapport, een studie, een link | Zoek de titel letterlijk op. Bestaat de publicatie, en staat erin wat de AI beweert? Vind je ze niet, dan bestaat ze waarschijnlijk niet. |
| **Een verklaring** — "dat komt door de digitalisering" | Data toont een verband, geen oorzaak. Een verklaring is een hypothese die je alleen overneemt als de bron ze zelf geeft — en dan met nuance, zoals je bij C6 leerde. |

Een bewering die je met geen enkele bron kan nakijken, neem je **niet** over. Dat is geen strengheid, dat is precies wat de rubric bij C3 en C6 van je verwacht: beperkingen benoemen en alleen besluiten wat de data draagt.

:::{admonition} Een tweede AI is geen controle
:class: warning

Je zou denken: als twee AI-assistenten hetzelfde zeggen, zal het wel kloppen. Helaas: ze zijn op grotendeels dezelfde teksten getraind en maken dezelfde soort fouten. Twee keer hetzelfde verzonnen cijfer is nog altijd verzonnen. De enige controle is de **bron**.
:::

### Oefening 2 — De controleoefening

Een leerling vroeg een AI-assistent om de dataset uit [hoofdstuk 3](03_Onderzoeksopdracht.ipynb) te duiden, zonder de cijfers mee te geven. Hieronder staat het gesprek uitgeschreven, en daaronder de echte cijfers uit Eurostat. Leg ze naast elkaar en beoordeel **elke genummerde bewering**: klopt / klopt niet / niet te controleren met deze bron. Schrijf er telkens bij wat je zou doen voor je de bewering in een rapport overneemt.

**De vraag van de leerling:**

> Ik doe voor school een onderzoek naar de werkloosheid in België per opleidingsniveau (Eurostat, 20- tot 64-jarigen, 2015 tot 2025). Wat kan je me daarover vertellen?

**Het antwoord van de AI-assistent** (uitgeschreven; de nummers zijn toegevoegd):

> Graag! De Belgische werkloosheidscijfers per opleidingsniveau tonen een duidelijk patroon.
>
> (1) Hoe lager het opleidingsniveau, hoe hoger de werkloosheid. In 2024 lag de werkloosheidsgraad bij laaggeschoolden rond 11,7 %, tegenover ongeveer 3 % bij mensen met een diploma hoger onderwijs.
>
> (2) Sinds 2015 is de werkloosheid bij laaggeschoolden bijna gehalveerd.
>
> (3) De coronacrisis van 2020 deed de werkloosheid bij alle opleidingsniveaus sterk stijgen.
>
> (4) Volgens het Eurostat-rapport *Education and Unemployment in Belgium 2024* (publicatie KS-2024-117) is de hoge werkloosheid bij laaggeschoolden vooral te wijten aan de digitalisering van laaggeschoolde jobs.
>
> (5) Bij mensen met een diploma hoger onderwijs schommelde de werkloosheid de hele periode tussen 3 en 5 %.
>
> (6) In 2025 daalde de werkloosheid verder, tot het laagste niveau in tien jaar.
>
> (7) Belgische laaggeschoolden zijn ruim dubbel zo vaak werkloos als het EU-gemiddelde.
>
> Deze cijfers zijn gebaseerd op de meest recente Eurostat-gegevens.

**De echte bron.** Werkloosheidsgraad in België, 20- tot 64-jarigen, mannen en vrouwen samen, in procent, per opleidingsniveau. Uittreksel uit de Eurostat-dataset [lfsa_urgaed — Unemployment rates by educational attainment level](https://ec.europa.eu/eurostat/databrowser/view/lfsa_urgaed/default/table), geraadpleegd op 28 augustus 2026 (laatste update door Eurostat: 30 juni 2026). Een *b* betekent dat Eurostat dat jaar een breuk in de tijdreeks meldt.

| Jaar | Geen diploma secundair (ISCED 0–2) | Diploma secundair (ISCED 3–4) | Diploma hoger onderwijs (ISCED 5–8) |
|---|---|---|---|
| 2015 | 16,6 | 8,5 | 4,6 |
| 2016 | 15,7 | 7,9 | 4,2 |
| 2017 | 14,4 *b* | 7,0 *b* | 4,3 *b* |
| 2018 | 12,9 | 5,9 | 3,5 |
| 2019 | 11,8 | 5,2 | 3,2 |
| 2020 | 11,6 | 5,7 | 3,5 |
| 2021 | 13,8 *b* | 6,7 *b* | 3,6 *b* |
| 2022 | 12,5 | 6,1 | 3,1 |
| 2023 | 12,3 | 6,1 | 3,0 |
| 2024 | 11,7 | 6,4 | 3,1 |
| 2025 | 13,8 | 6,5 | 3,7 |

:::{admonition} Oplossing oefening 2
:class: dropdown

| # | Oordeel | Waarom |
|---|---|---|
| 1 | **Klopt** | 11,7 % en 3,1 % in 2024; het patroon (lager diploma, hogere werkloosheid) geldt elk jaar. Overnemen mag — met de exacte cijfers uit je tabel. |
| 2 | **Klopt niet** | Van 16,6 % (2015) naar 11,7 % (2024) is een daling van bijna 30 %, geen halvering. En in 2025 staat het cijfer weer op 13,8 %. "Bijna gehalveerd" is een overdrijving die goed klinkt. |
| 3 | **Klopt niet** | In 2020 daalde het cijfer bij laaggeschoolden zelfs licht (11,8 → 11,6); bij de andere groepen steeg het met hooguit een half procentpunt. "Alle groepen" en "sterk" kloppen geen van beide. De echte stijging zit in 2021 — met een breuk in de tijdreeks, dus ook daar is voorzichtigheid geboden. |
| 4 | **Niet te controleren — verzonnen bron** | Zoek de titel en het publicatienummer letterlijk op: je vindt ze niet. Zo'n verwijzing ziet er echt uit, maar bestaat niet. Bovendien is "digitalisering" een verklaring die niet uit deze cijfers volgt: de tabel toont een verband tussen diploma en werkloosheid, geen oorzaak. Nooit overnemen. |
| 5 | **Klopt** | Tussen 2015 en 2025 schommelt de kolom hoger onderwijs tussen 3,0 en 4,6 %. |
| 6 | **Klopt niet** | In 2025 stéég de werkloosheid bij alle drie de groepen; bij laaggeschoolden zelfs terug tot het niveau van 2021. De AI-assistent beschrijft een trend die ze verwacht, niet een die in de data zit. |
| 7 | **Niet te controleren met deze bron** | Het EU-gemiddelde staat niet in dit uittreksel. Misschien klopt het, misschien niet: haal eerst de EU-cijfers uit dezelfde dataset op voor je dit overneemt. |
| slot | **Niet te controleren** | De assistent heeft geen toegang tot de Eurostat-databank en kreeg de cijfers niet van de leerling. "Gebaseerd op de meest recente gegevens" is een geruststellende zin zonder inhoud. |

Van acht beweringen kloppen er dus twee volledig. Wie het antwoord had gekopieerd, had drie foute cijferbeweringen, een verzonnen bron en een niet-onderbouwde verklaring in zijn rapport gezet.

Wat de leerling beter had gedaan: de tabel **meegeven** in de vraag (zie *Goede vragen stellen* hierboven) en het antwoord beperken tot wat de cijfers tonen. Dan is de kans op verzinsels veel kleiner — en de controle blijft nodig.
:::

## 4. Waar gebruik je AI wél voor, en waar niet?

De vuistregel: een AI-assistent is goed in **taal** en slecht in **feiten**. Alles wat met formuleren, uitleggen en brainstormen te maken heeft, mag ze doen. Alles wat een cijfer, een bron of jouw eigen oordeel vraagt, doe je zelf. Per stap van de onderzoekscyclus:

| Competentie | Wél | Niet |
|---|---|---|
| **C1** Onderzoeksvraag formuleren | Brainstormen: "geef tien mogelijke onderzoeksvragen bij deze dataset met deze dimensies"; een vage vraag scherper laten formuleren. | De vraag laten kiezen. Jij toetst elk voorstel: is het beantwoordbaar met *jouw* data, niet triviaal, scherp? |
| **C2** Data zoeken | Uitleggen wat een dimensie of code betekent (wat is ISCED 3–4? wat is een quintiel?); suggesties vragen voor zoektermen of portalen. | Cijfers opvragen. "Geef me de werkloosheid per land in 2024" levert een tabel op die eruitziet als data, maar geen data is. De dataset komt van het portaal. |
| **C3** Data beoordelen | Een definitie of methodologie in eenvoudige woorden laten uitleggen — en die uitleg daarna nalezen bij de bron. | De AI laten beslissen of een bron betrouwbaar is. Bron, actualiteit en definities kijk je zelf na. |
| **C4** Data klaarmaken | Uitleggen wat een Excel- of Power Query-stap doet; een foutmelding of een DAX-formule laten verklaren. | Rekenwerk. Sommen, percentages en gemiddelden bereken je in Excel of Power BI, niet in een chatbot: die rekent onbetrouwbaar en toont haar werk niet. |
| **C5** Visualiseren | Advies vragen over een grafiektype bij een vraagtype — en dat toetsen aan de tabel bij C5 en de checklist van hoofdstuk 4. | Een grafiek laten "beschrijven" die ze niet gezien heeft. Geef de cijfers, niet alleen de titel. |
| **C6** Interpreteren en concluderen | Je eigen conclusie vlotter laten formuleren; tegenargumenten en nuances vragen; controleren of je tekst de vraag écht beantwoordt. | Je analyse en conclusie laten schrijven. De interpretatie vertrekt van wat *jij* in *jouw* grafiek ziet — dat kan een AI-assistent niet voor je zien. |

Twee regels gelden altijd:

1. **Elk cijfer en elke bron die uit een AI-antwoord komt, controleer je aan de bron.** Kan dat niet, dan gebruik je het niet.
2. **Je vermeldt je AI-gebruik.** Schrijf in je rapport kort waarvoor je een AI-assistent inzette — in de eindopdracht op je verantwoordingstab. Bijvoorbeeld: *"Ik liet een AI-assistent de interpretaties bij vraag 2 en 4 herformuleren. De cijfers en conclusies zijn van mij en gecontroleerd in Power BI."* Dat is geen bekentenis, dat is onderzoekshygiëne — precies zoals je ook je dataset en je bron vermeldt.

Wat er in een concrete opdracht mag, staat telkens in de opdracht zelf. In de [overhoring](07_Overhoring.ipynb) en de [eindopdracht](08_Eindopdracht.ipynb) mag je een AI-assistent gebruiken bij je interpretaties; je onderzoeksvragen en je visualisaties maak je zelf.

## 5. Meer oefeningen

### Oefening 3 — Beoordeel de brainstorm

Een leerling bereidt de [eindopdracht](08_Eindopdracht.ipynb) voor met fiche 1, de minimumlonen in de EU. De dataset heeft drie dimensies: **land**, **tijd** (per halfjaar) en **munteenheid** (euro of koopkrachtstandaard). De leerling vroeg een AI-assistent om vijf onderzoeksvragen. Beoordeel elk voorstel met de criteria van C1: is de vraag **beantwoordbaar** met deze dataset, **niet triviaal** en **scherp**? Welke vragen hou je, welke stuur je bij, welke gooi je weg — en waarom?

> 1. Hoe zit het met de minimumlonen in Europa?
> 2. Welk EU-land heeft het hoogste minimumloon in 2025?
> 3. In welke EU-landen steeg het minimumloon sinds 2015 het sterkst, en verandert die rangschikking als je in koopkrachtstandaard in plaats van in euro vergelijkt?
> 4. Leidt een hoger minimumloon tot meer werkloosheid?
> 5. Hoe evolueerde het minimumloon in Denemarken sinds 2010?

:::{admonition} Oplossing oefening 3
:class: dropdown

1. **Weg.** Niet scherp: er is geen grafiek die deze vraag beantwoordt. Bruikbaar als *thema*, niet als vraag.
2. **Bijsturen.** Beantwoordbaar, maar triviaal: één blik op een gesorteerde tabel volstaat. Scherper: "Hoe positioneert België zich in 2025 tegenover de buurlanden, in euro en in koopkrachtstandaard?"
3. **Houden.** Beantwoordbaar (land, tijd en munteenheid zitten in de dataset), niet triviaal en scherp. Het verschil tussen euro en koopkracht is precies de valkuil die de fiche noemt — hier wordt ze een onderzoeksvraag.
4. **Weg — of een tweede dataset zoeken.** Interessant, maar niet beantwoordbaar met deze data: er zit geen werkloosheid in. En "leidt tot" vraagt om een oorzaak, terwijl data hoogstens een verband toont. De AI-assistent kent je dataset niet; jij wel.
5. **Weg.** Denemarken heeft geen wettelijk minimumloon en staat dus niet in de dataset — dat staat letterlijk in de fiche. De AI-assistent stelde een vraag voor over data die niet bestaat. Kijk altijd eerst in je dataset voor je een voorstel overneemt.
:::

### Oefening 4 — Met je eigen data (als je een AI-assistent kan gebruiken)

Deze oefening doe je in de klas, of thuis als je toegang hebt tot een AI-assistent. Heb je die niet, dan volstaan oefeningen 1 tot 3.

1. Neem je Power BI-rapport uit [hoofdstuk 3](03_Onderzoeksopdracht.ipynb) en kies één grafiek waarvoor je al een conclusie schreef.
2. Stel de AI-assistent een vraag **met context** (zie *Goede vragen stellen*) en **plak je cijfers erbij**. Vraag een beschrijving van het patroon, zonder verklaringen.
3. Controleer het antwoord bewering per bewering, zoals in oefening 2. Noteer wat klopt, wat niet klopt en wat je niet kan nakijken.
4. Vraag door: "Wat kan je met deze cijfers niet besluiten?" Vergelijk het antwoord met de beperkingen die jij zelf al kende (C3).
5. Verbeter nu je eigen conclusie — niet door die van de AI over te nemen, maar door wat je leerde te gebruiken: een scherpere formulering, een nuance die je vergat.
6. Schrijf één zin die je op je verantwoordingstab zou zetten over dit AI-gebruik.

:::{admonition} Voor de leraar: link met het leerplan
:class: dropdown

Het leerplan Informaticawetenschappen (D/2024/13.758/202) noemt AI-assistenten niet bij naam. Deze les verankert het gebruik ervan in de bestaande doelen van het onderdeel *Bevragen, analyseren en presenteren van een grote hoeveelheid data - big data*:

* De wenk bij het leerplandoel over het samenstellen van een dataset: "Laat de leerlingen kritisch reflecteren over de betrouwbaarheid van de dataset door onder meer controle van de bron en waar mogelijk vergelijking met datasets van andere bronnen." Oefening 2 past precies die werkwijze toe op een AI-antwoord: controle van de bron, bewering per bewering.
* De wenk bij het leerplandoel over visualisaties: "Toon hen dat met foute visualisaties verkeerde analyses worden gemaakt en dat dit soms bewust wordt gedaan voor desinformatie." Waar [hoofdstuk 4](04_Liegen_met_grafieken.ipynb) dat doet voor grafieken, doet deze les het voor gegenereerde tekst: correct klinkende cijfers en verzonnen bronnen.
* Het handelingswerkwoord *reflecteren over* uit het glossarium: "Kritisch nadenken over en argumenten afwegen zoals in een dialoog, een gedachtewisseling, een paper". Het doorvragen in punt 2 en de oefeningen 3 en 4 zijn zo'n dialoog.

De les is bewust toolonafhankelijk en werkt zonder account, zodat ze ook bruikbaar is in de lessen economie en de leerlingen zich nergens hoeven te registreren. Het uittreksel uit Eurostat in oefening 2 draagt een raadpleegdatum: laat de leerlingen gerust de actuele cijfers vergelijken — een herziening door Eurostat is op zich een mooie les over actualiteit (C3).
:::
