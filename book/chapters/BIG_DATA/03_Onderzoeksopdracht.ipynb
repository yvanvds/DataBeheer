# 3. Onderzoeksopdracht: werkloosheid in België

Je kent Power BI nu goed genoeg om er echte data mee te onderzoeken. In deze les werken we voor het eerst met een **Eurostat-dataset**: de werkloosheidscijfers in België, opgesplitst naar geslacht, leeftijd en opleidingsniveau. We doorlopen alle stappen samen: de data selecteren en downloaden op Eurostat, voorbereiden in Excel, importeren in Power BI en analyseren met gerichte onderzoeksvragen. In de volgende opdrachten moet je die stappen steeds meer zelfstandig zetten, dus werk aandachtig mee.

## Werken met Eurostat

Eurostat bevat erg veel data. Dat is tegelijk een voordeel en een uitdaging. Het voordeel is dat je veel mogelijke onderzoeksvragen kan formuleren. De uitdaging is dat je eerst moet leren hoe je de juiste selectie maakt. Als je zomaar alles downloadt, krijg je ofwel een onoverzichtelijk bestand, ofwel een tabel die eigenlijk niet past bij je vraag.

Daarom is het belangrijk om niet alleen "een bestand te downloaden", maar om eerst na te denken over de **organisatie van de data**. Voor je de export maakt, moet je dus goed kijken:

* welke dimensies zichtbaar zijn in de tabel;
* welke dimensies als filter zijn ingesteld;
* welke combinaties je in je uiteindelijke Excel wilt terugzien.

Wanneer je in Eurostat werkt, kan je dus niet zomaar veronderstellen dat de standaardweergave ook de beste is. Vaak bevat een dataset meerdere dimensies, maar zie je er maar twee onmiddellijk in de tabel. De andere zitten verborgen in filters. Dat betekent dat je eerst moet beslissen welke informatie je zichtbaar wilt houden.

In deze les doorlopen we stap voor stap hoe je aan de juiste data raakt.

### Stap 1: de dataset vinden

In deze les werken we met de dataset **"Unemployment by sex, age and educational attainment - annual data"**. Je kan die eenvoudig zoeken in de beschikbare databases.

### Stap 2: orden je data

* Verplaats **Time** naar *Row*.
* Verplaats **Education** naar *Column*.
* Selecteer bij **Geopolitical entity** enkel *Belgium*.
* Kies bij **Sex** *Male* en *Female*, maar niet *Total*.

### Stap 3: controleer het geografisch niveau

Zorg ervoor dat je werkt met **België** en niet met alle landen samen. Als er meerdere landen geselecteerd zijn, verklein je de selectie zodat enkel België overblijft. Dit is belangrijk, want als je alle landen selecteert is je dataset te groot en kan je die niet downloaden.

### Stap 4: download het bestand

Kies voor **download als Excel**. Sla het bestand op een logische plaats op, bijvoorbeeld in een map voor deze les of voor dit hoofdstuk. **Bij data scope kies je voor 'all selected dimensions'.**

### Stap 5: controleer het resultaat

Open het Excelbestand en kijk of je gegevens bevat voor:

* meerdere jaren;
* meerdere opleidingsniveaus;
* mannen en vrouwen;
* de gekozen leeftijdsafbakeningen.

Als je merkt dat een van die dimensies ontbreekt, dan heb je de data te sterk gefilterd en moet je terug naar Eurostat om de selectie aan te passen.

### Stap 6: verwijder overbodige gegevens

Je kan je bestand gebruiken zoals het er nu uitziet. Maar je maakt het jezelf gemakkelijker als je je gegevens zo duidelijk mogelijk maakt.

* Geef elk tabblad een naam die duidelijk aangeeft over welke data het gaat.
* Verwijder rijen die geen data bevatten.
* Verwijder kolommen die je niet nodig hebt.
* Geef elke kolom een duidelijke naam.


## De dataset inladen in Power BI

Nu je de dataset uit Eurostat hebt gedownload en voorbereid hebt in Excel, gaan we ze inladen in **Power BI**. Daar zullen we de gegevens stap voor stap onderzoeken. In deze eerste verkennende les is het niet de bedoeling dat je zomaar enkele mooie grafieken maakt. Je moet leren hoe je met een dataset werkt om er gerichte antwoorden uit af te leiden.

Hoe je een Excel-werkmap importeert in Power BI leerde je in het vorige hoofdstuk. Doe dat nu met je Eurostat-bestand: selecteer in het navigatievenster **alle werkbladen** en klik op **Laden**.

Controleer daarna in de **gegevensweergave**:

* staan de kolomnamen duidelijk vermeld?
* worden jaren herkend als jaartallen of als tekst?
* worden getallen als getallen gelezen?
* zie je alle gegevens die je wilde?

Wanneer een kolom verkeerd werd ingelezen, kan dat later voor problemen zorgen. Het is dus belangrijk om deze controle niet over te slaan.


## Onderzoeksvragen

Je dataset bestaat uit **zes tabellen**. Elke tabel bevat:

* een kolom **jaar**
* een kolom **Geen diploma secundair**
* een kolom **Diploma secundair**
* een kolom **Diploma hoger onderwijs**

De tabellen zijn opgesplitst volgens:

1. female 15 tot 74
2. female 20 tot 64
3. female 25 tot 74
4. male 15 tot 74
5. male 20 tot 64
6. male 25 tot 74

Dat betekent dat je in Power BI niet meteen kan werken met slicers voor geslacht of leeftijd. Je moet dus goed nadenken **welke tabellen je met elkaar vergelijkt**.

In de eerste opdrachten mag je daarom bewust nog werken met afzonderlijke tabellen en afzonderlijke visualisaties. Later zullen we leren hoe je zulke data beter kan combineren.

### Onderzoeksvraag 1: Welk verband zie je tussen opleidingsniveau en werkloosheid?

Voor deze eerste vraag werken we zeer concreet. We nemen één tabel als uitgangspunt, zodat je eerst het patroon in de data leert herkennen.

#### Stap 1: kies één tabel

Gebruik de tabel:

**female 20 tot 64**

Die tabel is geschikt omdat:

* ze een duidelijke werkende leeftijdsgroep bevat;
* ze meerdere jaren bevat;
* ze drie opleidingsniveaus naast elkaar toont.

#### Stap 2: maak drie metingen

In deze tabel staan de drie opleidingsniveaus als aparte kolommen. Daardoor kan je ze niet meteen als één legenda gebruiken. Voor deze eerste vraag lossen we dat eenvoudig op: je maakt drie afzonderlijke lijnen in één visualisatie.

Maak een **lijngrafiek** en gebruik:

* **X-as**: `jaar`

Voeg daarna drie waarden toe aan de grafiek:

* `Geen diploma secundair`
* `Diploma secundair`
* `Diploma hoger onderwijs`

Vergeet niet de namen van de rijen en van de grafiek aan te passen.

#### Stap 3: lees de grafiek

Kijk nu naar de algemene volgorde van de drie lijnen.

Beantwoord deze vragen voor jezelf:

* Welke lijn ligt meestal het hoogst?
* Welke lijn ligt meestal het laagst?
* Blijven de drie lijnen meestal in dezelfde volgorde?

#### Stap 4: schrijf je conclusie

Formuleer een besluit in volledige zinnen.

Een goed antwoord is bijvoorbeeld:

> In deze tabel is de werkloosheid in bijna alle jaren het hoogst bij mensen zonder diploma secundair. Bij mensen met een diploma hoger onderwijs ligt de werkloosheid het laagst. Dat wijst op een duidelijk verband tussen opleidingsniveau en werkloosheid.

Je hoeft de exacte percentages niet voor elk jaar op te sommen. Het gaat om het **patroon**.


### Onderzoeksvraag 2: Blijft dat verband stabiel doorheen de tijd?

Gebruik opnieuw de grafiek uit vraag 1. Maar zet er nu een 100% gestapeld kolomdiagram naast met dezelfde data. Daardoor zie je niet alleen de absolute cijfers, maar ook of de percentages evolueren.

Onderzoek nu niet alleen **of** er een verschil is tussen de drie opleidingsniveaus, maar ook of dat verschil **doorheen de jaren ongeveer gelijk blijft**. Dat kan je absoluut bekijken, maar ook percentueel.

Let op deze zaken:

* Blijven de drie lijnen in dezelfde volgorde?
* Wordt de afstand tussen de lijnen groter of kleiner?
* Zijn er jaren waarin de lijnen dichter naar elkaar toe gaan?

Schrijf daarna een korte conclusie.

---

### Onderzoeksvraag 3: Is dat patroon hetzelfde bij mannen en vrouwen?

Nu ga je twee tabellen vergelijken:

* **female 20 tot 64**
* **male 20 tot 64**

Omdat de gegevens in twee aparte tabellen zitten, werk je hier het best met **twee visualisaties naast elkaar**.

#### Stap 1: maak twee grafieken

Experimenteer met meerdere voorstellingen en kies zelf wat het meest geschikt lijkt.

#### Stap 2: vergelijk beide grafieken

Kijk nu niet alleen binnen één grafiek, maar ook tussen beide grafieken.

Stel jezelf deze vragen:

* Is de volgorde van de drie opleidingsniveaus dezelfde?
* Is het verschil tussen laag en hoog opgeleid bij mannen even groot als bij vrouwen?
* Zijn er jaren waarin het patroon opvallend anders lijkt?

#### Stap 3: formuleer een genuanceerde conclusie

Je antwoord hoeft niet te betekenen dat alles identiek is. Je mag ook aangeven dat het basispatroon hetzelfde is, maar dat er kleine verschillen zijn.


### Onderzoeksvraag 4: Verandert je conclusie wanneer je 15–74 vergelijkt met 20–64?

In de Eurostat-data wordt werkloosheid vaak berekend voor de groep 15–74 jaar. Dat betekent niet dat alle 65-plussers automatisch als werkloos meetellen. Alleen personen zonder werk die actief werk zoeken en beschikbaar zijn om te starten, tellen mee als werkloos. Gepensioneerden die niet meer naar werk zoeken, vallen daar dus buiten. De afbakening 20–64 wordt gebruikt wanneer men meer specifiek naar de klassieke werkende leeftijd wil kijken. Beide keuzes zijn verdedigbaar, maar ze leiden niet noodzakelijk tot exact dezelfde cijfers.

#### Stap 1: Maak je grafieken

Voor deze vergelijking werk je met 3 tabellen. Je kiest best één geslacht, en je vergelijkt in elke tabel een opleidingsniveau voor de twee leeftijdsgroepen.

#### Stap 2: vergelijk de patronen

Let op:

* Blijft de volgorde van de opleidingsniveaus hetzelfde?
* Liggen de percentages in beide leeftijdsafbakeningen even hoog?
* Lijkt het verschil tussen de opleidingsniveaus even groot?

#### Stap 3: schrijf een conclusie

Je hoeft hier niet te bewijzen dat alles helemaal verandert. Het is al voldoende als je kan uitleggen of de algemene conclusie behouden blijft en of de gekozen bevolking een invloed heeft op de exacte waarden.



### Onderzoeksvraag 5: Welke beperkingen heeft deze dataset voor het verklaren van werkloosheid?

Deze vraag is anders dan de vorige. Je moet nu niet vooral grafieken maken, maar kritisch nadenken over wat deze dataset **wel en niet** kan tonen.

Je mag hiervoor gerust terugkijken naar de structuur van de dataset.



## Extra: nieuwe kolommen of metingen maken

Voor de meeste bovenstaande vragen heb je nog geen extra metingen nodig. Je kan al veel afleiden uit de bestaande kolommen en uit vergelijkingen tussen tabellen.

Toch kan het nuttig zijn om één extra meting te maken, zodat je het verschil tussen laag- en hoogopgeleiden explicieter kan tonen.

### Meting: verschil tussen laag en hoog opgeleid

Maak in één tabel een nieuwe meting aan met deze expressie:

```DAX
Verschil laag-hoog = 
MAX('female 20 tot 64'[Geen diploma secundair]) - MAX('female 20 tot 64'[Diploma hoger onderwijs])
```

Met deze meting bereken je het verschil tussen de werkloosheid van de laagste en hoogste opleidingsgroep.

Belangrijk:

* vervang `female 20 tot 64` door de naam van de tabel waarmee jij werkt
* gebruik deze meting alleen binnen die ene tabel

Deze meting kan nuttig zijn als je expliciet wilt nagaan:

* hoe groot het verschil is tussen laag en hoog opgeleid
* of dat verschil doorheen de tijd verandert

### Berekende kolom: verschil laag-hoog per rij

Je kan ook een nieuwe kolom maken in een tabel. Gebruik dan deze expressie:

```DAX
Verschil laag-hoog = 
'female 20 tot 64'[Geen diploma secundair] - 'female 20 tot 64'[Diploma hoger onderwijs]
```

Deze berekende kolom maakt voor elk jaar het verschil zichtbaar tussen beide opleidingsniveaus.

Dat is eenvoudiger te begrijpen dan een meting, omdat elke rij gewoon een nieuw getal krijgt.
