# 3. Onderzoeksopdracht: belastingdruk in de Europese Unie

In de vorige opdracht onderzocht je de werkzaamheidsgraad. Deze keer kijk je naar **belastingen**: hoeveel procent van het inkomen dragen huishoudens af in de verschillende EU-landen, en verschilt dat naargelang hoe rijk of arm je bent? Je werkt opnieuw in PowerBI en je beantwoordt zelf een aantal onderzoeksvragen.

## Wat ga je doen?

Je werkt met een dataset van Eurostat over de **directe belastingen die huishoudens betalen als percentage van hun bruto-inkomen**. Je maakt een PowerBI-rapport waarin je **zes onderzoeksvragen** beantwoordt. Vier vragen krijg je van mij. Twee vragen formuleer je zelf.

Per vraag werk je in een aparte tab. Bij elke tab hoort:

* één of meerdere visualisaties die de vraag helpen beantwoorden;
* een korte interpretatie waarin je in eigen woorden uitlegt wat je in de data ziet.

## Hoe wordt dit beoordeeld?

Op deze opdracht staan punten. De rubric vind je in MS Teams. Lees die zeker **vóór** je begint, want daarin staat wat ik verwacht. Een paar zaken om alvast te onthouden:

* Je datavoorbereiding telt mee. Slordig opgekuiste data kost punten.
* De presentatie van je rapport telt mee. Eén vraag per tab, duidelijke titels, verzorgde layout.
* Voor elke vraag tellen visualisatie én interpretatie. Een mooie grafiek zonder uitleg is niet voldoende.
* Voor de twee zelfgekozen vragen telt ook de **kwaliteit van de vraag zelf**. Een vraag waarop je in twee seconden "ja" of "nee" kan antwoorden, is niet relevant genoeg.
* Vragen die je niet uitwerkt, leveren geen punten op. Wie traag werkt, mag thuis verder werken.

## Tijd

Je krijgt twee lesuren. Wie niet klaar is, werkt thuis verder en levert in via MS Teams.

## Stap 1: data ophalen op Eurostat

We werken met de dataset **Distribution of direct taxes paid by households as a percentage of their gross income by income quintile**. Je vindt die hier:

👉 [Eurostat — icw_tax_06](https://ec.europa.eu/eurostat/databrowser/view/icw_tax_06/default/table)

Omdat de dataset complex is en kleine selectiefouten je hele rapport overhoop halen, krijg je deze keer **exacte instructies** voor wat je moet aanvinken. Volg ze nauwkeurig.

### Hoe orden je je data?

* **Row (rij)**: Geopolitical entity (reporting)
* **Column (kolom)**: Time

### Welke waarden kies je?

* **Geopolitical entity**: alle EU-landen, **behalve Kroatië, Italië en het Verenigd Koninkrijk**. Die landen hebben voor sommige jaren geen data, en zouden je grafieken vervuilen.
* **Time**: enkel de even jaren tussen 2008 en 2020 — dus **2008, 2010, 2012, 2014, 2016, 2018 en 2020**. Dat geeft een mooie spreiding zonder dat je rapport overvol raakt.
* **Income quintile**: alle zes — **Total, First quintile, Second quintile, Third quintile, Fourth quintile, Fifth quintile**.
* **Quantile**: enkel **Median**. (De andere waarden — P25 en P75 — laten zien hoe sterk de belastingdruk binnen één inkomensgroep varieert. Daar werken we deze keer niet mee.)
* **Unit of measure**: Percentage (staat al goed).

### Downloaden

Klik op **Download** rechts boven de tabel. Kies:

* **File format**: Spreadsheet (.xlsx)
* **Data scope**: All selected dimensions
* Vink **niet** "Consolidate all data sheets in a single one" aan. We willen 6 aparte tabbladen — één per inkomensquintiel.

Bewaar het bestand op een plek die je terugvindt.

### Aandachtspunten

Onze dataset bevat enkel de **directe belastingen** — niet de volledige belastingdruk. Dat is een belangrijk verschil:

* **Directe belastingen** = inkomstenbelasting, gemeentebelasting, bedrijfsvoorheffing — wat rechtstreeks van je loonbrief afgaat.
* **Indirecte belastingen** = BTW, accijnzen — die je betaalt als je iets koopt.
* **Sociale zekerheidsbijdragen** = werknemers- én werkgeversbijdragen voor pensioen, ziekteverzekering, werkloosheid.

België staat internationaal vaak hoog in **totale** belastingdruk, vooral door de zware werkgeversbijdragen. In onze cijfers (alleen directe belastingen) zal België dus mogelijk minder hoog scoren dan je zou verwachten. Hou daar rekening mee bij je interpretatie.

Verder: de cijfers tonen telkens de **mediaan** belastingdruk binnen een groep. Dat is het typische huishouden in die groep — niet het gemiddelde, niet het hoogste of laagste.

## Stap 2: data klaarmaken in Excel

Het bestand dat Eurostat je geeft, is **niet meteen klaar** voor PowerBI. Je moet het eerst opkuisen. Vorige les deden we dat samen. Deze keer moet je zelf bepalen wat er moet gebeuren — maar de structuur is gelukkig voor alle zes de tabbladen identiek.

Open je gedownloade Excel en stel jezelf deze vragen:

* Bovenaan elk tabblad staan een aantal **metadatarijen** (Dataset, Last updated, Time frequency, ...). Wat doe je daarmee?
* Tussen elke jaarkolom staat een **lege of vrijwel lege kolom** met af en toe een `i`-flag. Waarom staan die er, en wat doe je ermee?
* Zijn de **kolomnamen** duidelijk en bruikbaar? Zou PowerBI ermee overweg kunnen?
* Klopt het **type** van elke kolom (tekst, getal, ...)?
* Zijn er **cellen met `:`** of een ander niet-numeriek teken in plaats van een getal? Wat betekenen die? Hoe los je dat op?
* Hebben de **tabbladen duidelijke namen**? `Sheet 1` t.e.m. `Sheet 6` is niet erg behulpzaam. Geef ze logische namen, bijvoorbeeld op basis van de inkomensquintiel.

Werk netjes. Sla je bestand op met een duidelijke naam.

> **Tip:** Doe het opkuisen één keer goed op één tabblad, en pas dan dezelfde stappen toe op de andere vijf. Wie slordig werk inlevert in Excel, krijgt vaak ook fouten in PowerBI.

## Stap 3: importeer in PowerBI

Open PowerBI Desktop en importeer je Excel-bestand. **Selecteer alle zes de tabbladen** bij het importeren — je hebt ze allemaal nodig.

Controleer in de **table view**:

* Zijn alle zes de tabellen correct ingelezen?
* Klopt het datatype van elke kolom?
* Zijn er kolommen die je niet nodig hebt? Verwijder die gerust.
* Zijn de tabelnamen duidelijk?

Pas dit aan voor je begint te visualiseren. Achteraf is dat veel lastiger.

## Tussenstap: gegevens transformeren met Power Query

Als je je tabellen nu zou gebruiken in een grafiek, loop je vast. Probeer maar eens een lijngrafiek te maken die de evolutie van de belastingdruk in België tussen 2008 en 2020 toont. Je krijgt zeven losse stippen in plaats van één lijn. **Waarom?**

Omdat de jaren in je tabel als **kolommen** staan (2008, 2010, 2012, ...) in plaats van als rijen. PowerBI ziet die als zeven aparte cijfers per land, niet als één tijdreeks. Voor een grafiek hebben we de jaren als één kolom "Jaar" nodig, met de bijhorende waarde in een aparte kolom "Belastingdruk".

Dat ombouwen heet **draaien ongedaan maken** of **unpivot**. Je doet het in de **Power Query Editor**.

### Stappenplan

Open de Power Query Editor via **Start → Gegevens transformeren**. Voer voor **elke** van je zes tabellen de volgende stappen uit:

1. Klik links in de querylijst op de tabel die je wil bewerken (bijvoorbeeld `Total`).
2. Klik met de **rechtermuisknop** op de kolomkop **Land** (Afhankelijk van hoe je je eerste kolom genoemd hebt).
3. Kies **"Draaien ongedaan maken voor andere kolommen"** (Engels: *Unpivot Other Columns*). Je ziet nu drie kolommen: `Land`, `Kenmerk` en `Waarde`.
4. Dubbelklik op de kolomkop `Kenmerk` en hernoem hem naar **Jaar**.
5. Dubbelklik op de kolomkop `Waarde` en hernoem hem naar **Belastingdruk**.
6. Klik op het symbooltje links naast `Jaar` (`ABC` of `123`) en kies **Geheel getal** als gegevenstype.
7. Klik op het symbooltje links naast `Belastingdruk` en kies **Decimaal getal** als gegevenstype.

Herhaal stap 1 t.e.m. 7 voor de andere vijf tabellen. Klik dan boven in het lint op **Sluiten en toepassen**.

Na deze stap heeft elke tabel drie kolommen: `Land`, `Jaar`, `Belastingdruk`. Daarmee kan PowerBI vlot lijngrafieken, staafgrafieken en andere visualisaties maken.

> **Tip:** Maak je een fout? In het paneel rechts zie je onder *Toegepaste stappen* een lijst met al je acties. Klik op het kruisje naast een stap om die ongedaan te maken.

## De onderzoeksvragen

Werk elke vraag uit op een **aparte tab** in PowerBI. Geef die tab een duidelijke naam. Per tab toon je:

* één of meerdere visualisaties die je antwoord ondersteunen;
* je interpretatie als tekstvak op de tab.

Lees elke vraag goed voor je begint. Bedenk welk type grafiek het beste past, welke tabel je gebruikt en of je extra filters nodig hebt. Soms moet je experimenteren voor je iets bruikbaar hebt — dat hoort erbij.

### Vraag 1 — Hoogste belastingdruk in de EU

**Welke EU-landen kennen de hoogste mediaan belastingdruk op het totale inkomen? Hoe positioneert België zich in dat klassement?**

Denk na over:

* Welk(e) jaar(en) gebruik je? Maakt de keuze veel verschil?
* Hoe maak je België visueel herkenbaar in je grafiek?
* Wat valt op aan de uitschieters bovenaan en onderaan?

### Vraag 2 — Belastingdruk in België door de jaren

**Hoe evolueerde de mediaan belastingdruk in België tussen 2008 en 2020?**

Denk na over:

* Zie je een duidelijke trend (stijging, daling, stabiliteit)?
* Zijn er opvallende sprongen die je kan koppelen aan beleidsbeslissingen of economische gebeurtenissen?
* Welk type grafiek toont een evolutie het duidelijkst?

### Vraag 3 — Belasting op de laagste inkomens in 2020

**Welke EU-landen belasten de armste 20% (Q1) het zwaarst in 2020? Hoe positioneert België zich?**

Denk na over:

* Wat zegt het cijfer voor Q1 over het sociaal vangnet en de fiscale vrijstellingen in een land?
* Komt het beeld overeen met wat je verwachtte?
* Welke tabel uit je dataset gebruik je voor deze vraag?

### Vraag 4 — Progressiviteit in 2020

**Hoe progressief belasten EU-landen hun burgers in 2020? Vergelijk de top 10 voor de armste (Q1) en de rijkste (Q5) inkomensgroep.**

Een belastingsysteem is **progressief** wanneer mensen met een hoger inkomen procentueel meer betalen dan mensen met een lager inkomen. Hoe groter het verschil tussen Q5 en Q1, hoe progressiever het systeem.

Denk na over:

* Welke landen staan in beide top 10s? Welke maar in één van de twee?
* Wat zegt dat over de progressiviteit van hun belastingsysteem?
* Hoe positioneert België zich in beide grafieken?

### Vraag 5 en 6 — Twee zelfgekozen vragen

Formuleer zelf twee onderzoeksvragen die je met deze dataset kan beantwoorden. Werk ze op dezelfde manier uit als de vorige vragen: visualisatie + interpretatie op een eigen tab.

**Schrijf de vraag duidelijk bovenaan de tab.**

Een goede onderzoeksvraag:

* is **beantwoordbaar** met de data die je hebt — kijk dus na of jouw selectie alle nodige dimensies bevat;
* is **niet triviaal** — geen vraag waarop je in twee seconden het antwoord kan aflezen;
* is **scherp geformuleerd** — niet "hoe zit het met belastingen in Europa", wel iets concreets waarop een grafiek een duidelijk antwoord geeft;
* voegt iets toe aan wat je in de eerste vier vragen al onderzocht hebt.

Je hebt zes tabellen tot je beschikking (Total + Q1 t.e.m. Q5), elk met zeven jaar aan data over 25 landen. Daar zit veel meer in dan wat de eerste vier vragen behandelen — denk bijvoorbeeld aan de middenklasse (Q3), aan de evolutie van progressiviteit door de jaren, of aan een vergelijking tussen specifieke landengroepen.

De **kwaliteit van je vraag** wordt apart beoordeeld. Een zwakke vraag, ook al is de uitwerking netjes, levert minder punten op.
