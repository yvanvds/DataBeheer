Zo zie je bijvoorbeeld dat de tabel `Customer` geen adressen bevat. Dat is eigenlijk beter dan de sterk vereenvoudigde database die we in de vorige hoofdstukken gebruiken. Denk even terug aan de laatste keer dat je online iets bestelde. Je kan toch elke keer een ander adres opgeven? Misschien laat je de ene keer iets thuis afleveren, en een andere keer bij familie.

Wanneer het adres in je `customer` tabel zit, dan hebben we slechts plaats voor één adres. Dat is niet handig. Daarom heeft deze database een afzonderlijke tabel `Address`.

Maar zo ontstaat er een nieuw probleem. Hoe weten we nu welke adressen er bij welke gebruiker horen? Daarvoor dient de tabel `CustomerAddress`. Je ziet het al een beetje aan de naam. Het enige wat die tabel doet, is gebruikers linken aan adressen. 

`CustomerAddress` bevat enkel twee sleutelvelden: `CustomerID` en `AddressID`. Zolang de combinatie van die twee uniek is, kunnen we gebruikers aan adressen blijven koppelen.
