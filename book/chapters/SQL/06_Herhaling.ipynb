# 6. Herhaling, maar dan anders

:::{admonition} Leerdoelen
:class: tip

In deze les leer je je kennis toepassen op een compleet nieuwe database.

Na deze les kan je:
- de structuur van een onbekende database analyseren aan de hand van een schema
- uitleggen waarom een btw-tarief in een aparte opzoektabel staat
- uitleggen waarom een factuurlijn zijn eigen eenheidsprijs bewaart
- SELECT, WHERE en ORDER BY toepassen op een nieuwe database
- tabellen combineren met JOIN en LEFT JOIN
- groeperen en filteren met GROUP BY en HAVING
:::

Tot hiertoe gebruikte je twee databases: de webshop om nieuwe leerstof te leren, en AdventureWorks om ze toe te passen. Maar SQL beperkt zich niet tot verkoop. In dit hoofdstuk kruip je in de rol van boekhouder bij **Kantoorplus bv**, een fictieve groothandel in kantoormateriaal uit Gent. Kantoorplus levert papier, meubilair, koffie en opleidingen aan Vlaamse bedrijven, scholen en verenigingen — en stuurt daar facturen voor.

De structuur van de database ziet er zo uit:

![Schema van de facturatie-database](../../figures/facturatie_schema.svg)

Enkele zaken vallen op:

- **`btw_tarieven` is een opzoektabel.** In België betaal je niet op alles evenveel btw: kantoormateriaal valt onder het standaardtarief van 21%, voeding en dranken onder het verlaagd tarief van 6%, en opleidingen zijn vrijgesteld (0%). In plaats van het tarief bij elk product te herhalen, verwijst elk product met `btw_code` naar één rij in `btw_tarieven`. Wijzigt de overheid een tarief, dan moet Kantoorplus maar één getal aanpassen.
- **`factuurlijnen` bewaart zijn eigen `eenheidsprijs`.** Vreemd, want die staat toch al in `producten`? Toch is dit bewust: een factuur is een officieel document dat achteraf niet meer mag veranderen. Als Kantoorplus volgend jaar de prijs van kopieerpapier verhoogt, moeten oude facturen hun oorspronkelijke prijs behouden. Bovendien krijgen vaste klanten soms korting.
- **`betaald_op` mag NULL zijn.** Een factuur zonder betaaldatum is nog niet betaald. Elke factuur moet betaald worden vóór de `vervaldatum` (30 dagen na de factuurdatum) — en zoals elke boekhouder weet, lukt dat niet bij iedereen.

## Oefenreeks 1: basics


```sql
--- 1. Toon de bedrijfsnaam en de gemeente van alle klanten, alfabetisch gesorteerd op bedrijfsnaam.

--- 2. Toon de naam en de prijs van alle producten die meer dan 100 euro kosten. Sorteer van duur naar goedkoop.

--- 3. Toon de vijf recentste facturen.

--- 4. In welke gemeenten heeft Kantoorplus klanten? Toon elke gemeente maar één keer, alfabetisch gesorteerd.

--- 5. Toon per factuurlijn het aantal, de eenheidsprijs en een berekende kolom lijntotaal (aantal x eenheidsprijs).
```

## Oefenreeks 2: realistische toepassingen

Een boekhouder let vooral op wat ontbreekt of te laat komt. Onthoud: `betaald_op` is NULL zolang een factuur niet betaald is.


```sql
--- 1. Welke facturen zijn nog niet betaald? Toon het factuurnummer, de factuurdatum en de vervaldatum.

--- 2. Welke facturen werden te laat betaald? Vergelijk de betaaldatum met de vervaldatum.

--- 3. Vandaag is het 20 augustus 2025. Welke onbetaalde facturen zijn al vervallen? Dit zijn de klanten die een aanmaning moeten krijgen.

--- 4. Scholen en vzw's hebben geen btw-nummer. Toon de klanten zonder btw-nummer.

--- 5. Toon alle klanten waarvan de bedrijfsnaam 'vzw' bevat.

--- 6. Toon alle facturen van juni 2025.

--- 7. Toon de producten met het verlaagd tarief of de vrijstelling (btw_code 'VERL' of 'VRIJ').
```

## Oefenreeks 3: JOIN


```sql
--- 1. Toon elk factuurnummer met de factuurdatum en de bedrijfsnaam van de klant.

--- 2. Toon per factuurlijn de productnaam, het aantal en de eenheidsprijs.

--- 3. Toon elke productnaam met zijn prijs en zijn btw-tarief (het percentage, niet de code).

--- 4. Toon per factuurlijn het factuurnummer, de bedrijfsnaam van de klant, de productnaam en het aantal. Je hebt vier tabellen nodig!
```

## Oefenreeks 4: LEFT JOIN


```sql
--- 1. Toon alle klanten, samen met hun factuurnummers (indien aanwezig).

--- 2. Welke klanten hebben nog geen enkele factuur gekregen? Dat zijn wellicht nieuwe klanten.

--- 3. Welke producten werden nog nooit gefactureerd? De aankoopdienst wil weten of ze uit het aanbod moeten.
```

## Oefenreeks 5: eenvoudige aggregaties


```sql
--- 1. Tel hoeveel facturen elke klant al kreeg. Toon de bedrijfsnaam en sorteer van veel naar weinig.

--- 2. Bereken per factuur het totaalbedrag (aantal x eenheidsprijs, opgeteld over alle lijnen).

--- 3. Bereken de omzet per productcategorie. Sorteer van hoog naar laag.

--- 4. Tel hoeveel klanten er in elke gemeente zijn.
```

## Oefenreeks 6: geavanceerde aggregaties


```sql
--- 1. Toon enkel de klanten met minstens 15 facturen. Dit zijn de beste klanten van Kantoorplus.

--- 2. Toon enkel de gemeenten met meer dan 2 klanten.

--- 3. Toon enkel de facturen met een totaalbedrag boven 1500 euro. Sorteer van hoog naar laag.

--- 4. Van welke producten werden er in totaal meer dan 200 stuks gefactureerd?

--- 5. Uitdaging: bereken hoeveel btw Kantoorplus moet doorstorten per btw-tarief. De btw op een factuurlijn is: aantal x eenheidsprijs x tarief / 100.
```
