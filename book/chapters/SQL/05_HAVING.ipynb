# 5. Groepen Filteren

:::{admonition} Leerdoelen
:class: tip

In deze les leer je hoe je groepen kan filteren.

Na deze les kan je:
- HAVING gebruiken
- het verschil tussen HAVING en WHERE uitleggen
:::

HAVING gebruik je om groepen te filteren, net zoals WHERE individuele rijen filtert.

| Gebruik 	| Voor                         	| Voorbeeld                        	|
|---------	|------------------------------	|----------------------------------	|
| WHERE   	| filter rijen vóór groepering 	| WHERE order_date >= '2025-01-01' 	|
| HAVING  	| filter groepen na berekening 	| HAVING SUM(total) > 1000         	|

Hieronder zie je een voorbeeld. Test de query.

Je kan `WHERE` en `HAVING` ook samen gebruiken. Probeer deze instructie op de correcte plaats toe te voegen:

```SQL
WHERE status = 'delivered'
```

````sql
SELECT customer_id,
       COUNT(*) AS order_count
FROM orders
GROUP BY customer_id
HAVING COUNT(*) >= 20          -- toon enkel klanten met 20 of meer bestellingen
ORDER BY order_count DESC;
````
