# Toepassing 5: HAVING in AdventureWorks

De samenvattingen uit de vorige toepassing leverden soms lange lijsten op. Het management wil enkel de groepen zien die er echt toe doen: de landen met veel klanten, de categorieën met een groot aanbod. Daarvoor filter je groepen met `HAVING`.

Elke oefening hieronder bouwt verder op een `GROUP BY`-query uit Toepassing 4. Schrijf eerst de groepering, en voeg daarna de filter toe.

## Oefeningen


```sql
--- 1. Toon enkel de landen waar minstens 100 klanten wonen.

--- 2. Toon enkel de kleuren waarvoor de gemiddelde catalogusprijs hoger is dan 500 euro.

--- 3. Toon enkel de categorieën met 10 of meer producten. Toon de categorienaam en het aantal.

--- 4. Toon enkel de bestellingen waarvan het totaal van de orderregels (LineTotal) meer dan 5000 euro bedraagt.

--- 5. Toon enkel de categorieën waarvan het duurste product minstens 2000 euro kost.
```
