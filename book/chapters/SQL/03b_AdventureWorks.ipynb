# Toepassing 3: JOIN in AdventureWorks

Met JOIN kunnen we nu eindelijk de hele AdventureWorks-database gebruiken. Kijk nog eens naar het schema:

![Schema van de AdventureWorks-database](../../figures/adventure_works.png)

Je ziet bijvoorbeeld dat de tabel `Customer` geen adressen bevat. Dat is eigenlijk beter dan onze sterk vereenvoudigde webshop-database. Denk even terug aan de laatste keer dat je online iets bestelde. Je kan toch elke keer een ander adres opgeven? Misschien laat je de ene keer iets thuis afleveren, en een andere keer bij familie.

Wanneer het adres in de `Customer`-tabel zit, dan is er slechts plaats voor één adres. Dat is niet handig. Daarom heeft deze database een afzonderlijke tabel `Address`.

Maar zo ontstaat er een nieuw probleem. Hoe weten we nu welke adressen bij welke klant horen? Daarvoor dient de tabel `CustomerAddress`. Je ziet het al een beetje aan de naam. Het enige wat die tabel doet, is klanten linken aan adressen.

`CustomerAddress` bevat vooral twee sleutelvelden: `CustomerID` en `AddressID`. Zolang de combinatie van die twee uniek is, kunnen we klanten aan adressen blijven koppelen. Zo'n tabel noemen we een **koppeltabel**. Om een klant met zijn adres te tonen, heb je dus twee JOINs nodig: van `Customer` naar `CustomerAddress`, en van `CustomerAddress` naar `Address`.

## Oefeningen


```sql
--- 1. Toon het bestellingsnummer (SalesOrderID), het totaalbedrag (TotalDue) en de voor- en achternaam van de klant die de bestelling plaatste.

--- 2. Toon de productnaam, kleur en prijs, samen met de naam van de categorie waartoe het product behoort.

--- 3. Toon per orderregel het SalesOrderID, de productnaam, het aantal (OrderQty) en de eenheidsprijs (UnitPrice).

--- 4. Toon de voornaam, achternaam, stad en land van elke klant met een adres. Je hebt drie tabellen nodig: denk aan de koppeltabel!
```

## Oefeningen: LEFT JOIN


```sql
--- 1. Toon alle klanten, samen met hun SalesOrderID (indien aanwezig). Wat valt je op aan het resultaat?

--- 2. Toon alle categorieën en de namen van hun producten, ook als een categorie geen producten bevat.

--- 3. Toon alle adressen (AddressLine1, City), samen met de achternaam van de klant (indien gekoppeld).

--- 4. Uitdaging: toon elke productmodelnaam (ProductModel) met zijn Engelse beschrijving. De koppeltabel ProductModelProductDescription verbindt modellen met beschrijvingen; de Engelse versie heeft Culture = 'en'.
```
