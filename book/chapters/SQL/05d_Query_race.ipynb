# Query-race: AdventureWorks

Vandaag geen gewone oefenles. De vragen gaan één voor één open, iedereen schrijft queries — en wie het eerst het juiste antwoord vindt, pakt het punt. Welkom bij de eerste **query-race** van het jaar.

Elke vraag heeft precies één antwoord: **één getal of één naam**. Gokken heeft geen zin, want je antwoord telt alleen als je query het bewijst. Nieuwe leerstof is er niet: alles wat je nodig hebt komt uit de voorbije hoofdstukken, van SELECT tot HAVING, op de AdventureWorks-database die je intussen goed kent.

:::{admonition} Spelregels
:class: warning
- De leerkracht opent de vragen één voor één. Antwoorden op een vraag die nog niet open is, tellen niet — vooruitwerken heeft dus geen zin.
- Denk je het antwoord te hebben? Steek je hand op en toon je scherm: query én resultaat. Het eerste juiste antwoord scoort het punt.
- Fout antwoord? Dan zoekt de rest verder — voor die ene vraag ben jij uitgeteld.
- De vragen in ronde 3 zijn moeilijker en tellen voor **twee punten**.
- Alle punten gaan in het **jaarklassement**. Er valt niets te winnen behalve eeuwige roem — maar daar is het ons om te doen.
:::

## Ronde 1: opwarmen

Eén tabel per vraag, geen JOIN nodig.


```sql
--- 1. Hoeveel producten staan er in de tabel Product?

--- 2. Wat is de catalogusprijs (ListPrice) van het duurste product?

--- 3. Hoeveel producten hebben geen kleur?

--- 4. In hoeveel verschillende kleuren bestaan de producten? Producten zonder kleur tellen niet mee.
```

## Ronde 2: groeperen

Vanaf hier draait alles om GROUP BY — en soms heb je ook een JOIN nodig.


```sql
--- 5. In welk land liggen de meeste adressen uit de tabel Address?

--- 6. Welke kleur heeft de hoogste gemiddelde catalogusprijs?

--- 7. Welke productcategorie telt de meeste producten?

--- 8. Wat is de gemiddelde catalogusprijs van alle Mountain Bikes, afgerond op twee cijfers na de komma?
```

## Ronde 3: de finale

Twee punten per vraag. Hier beslis je de race.


```sql
--- 9. Hoeveel stuks (OrderQty) werden er in totaal besteld, over alle orderregels heen?

--- 10. Van welk product werden er in totaal de meeste stuks besteld?

--- 11. Welk bedrijf (CompanyName) plaatste de bestelling met het hoogste totaalbedrag (TotalDue)?

--- 12. Hoeveel productcategorieën tellen 10 of meer producten?
```

Klaar? Tel de punten samen en zet ze op het bord. De revanche volgt later dit jaar — op een database die je nu nog niet kent.
