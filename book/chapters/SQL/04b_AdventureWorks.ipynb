# Toepassing 4: groeperen in AdventureWorks

Je zit weer op de stoel van de data-analist bij Adventure Works Cycles, en deze keer zijn de vragen een maat groter. De afdelingen willen geen lange lijsten meer, maar **samenvattingen** waarmee ze een beslissing kunnen nemen: waar zitten onze klanten, welke categorie breiden we uit? Dat zijn typische vragen voor `GROUP BY` en aggregatiefuncties.

Voor sommige oefeningen moet je eerst tabellen combineren met JOIN, en daarna pas groeperen. Zo werk je in een echte data-analyse ook.

## Oefeningen


```sql
--- 1. De directie overweegt een distributiecentrum buiten de VS, maar alleen in een land met genoeg klanten. Tel hoeveel klanten er in elk land wonen. Je hebt hiervoor de koppeltabel CustomerAddress en de tabel Address nodig.

--- 2. De marketingafdeling vermoedt dat sommige kleuren vooral in de dure gamma's voorkomen. Bereken de gemiddelde catalogusprijs (ListPrice) van de producten per kleur, dan weet marketing welke kleuren in de premiumcampagne thuishoren.

--- 3. De aankoopdienst evalueert het assortiment: categorieën met amper producten worden misschien geschrapt. Tel hoeveel producten er in elke categorie zitten. Toon de categorienaam, niet enkel het ID.

--- 4. De boekhouding wil controleren of de gefactureerde bedragen kloppen. Bereken per bestelling (SalesOrderID) het totale bedrag van alle orderregels (LineTotal).

--- 5. De winkelverkopers willen bij een adviesgesprek meteen het prijsbereik van elke categorie kunnen noemen. Toon per categorie de laagste en de hoogste catalogusprijs.
```
