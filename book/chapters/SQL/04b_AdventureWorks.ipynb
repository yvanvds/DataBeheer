# Toepassing 4: groeperen in AdventureWorks

Adventure Works Cycles heeft honderden klanten en producten. Niemand wil die allemaal één voor één bekijken: het management wil **samenvattingen**. Hoeveel klanten hebben we per land? Wat is de gemiddelde prijs per categorie? Dat zijn typische vragen voor `GROUP BY` en aggregatiefuncties.

Voor sommige oefeningen moet je eerst tabellen combineren met JOIN, en daarna pas groeperen. Zo werk je in een echte data-analyse ook.

## Oefeningen


```sql
--- 1. Tel hoeveel klanten er in elk land wonen. Je hebt hiervoor de koppeltabel CustomerAddress en de tabel Address nodig.

--- 2. Bereken de gemiddelde catalogusprijs (ListPrice) van de producten per kleur.

--- 3. Tel hoeveel producten er in elke categorie zitten. Toon de categorienaam, niet enkel het ID.

--- 4. Bereken per bestelling (SalesOrderID) het totale bedrag van alle orderregels (LineTotal).

--- 5. Toon per categorie de laagste en de hoogste catalogusprijs.
```
