# Toepassing 1: eerste queries op AdventureWorks

Wat je in de vorige les leerde, kan je toepassen op elke database. Daarom oefenen we vanaf nu telkens ook op een tweede database: die van **Adventure Works Cycles**, een fictieve Amerikaanse fietsenfabrikant. Het bedrijf verkoopt fietsen, onderdelen en kledij aan klanten in de VS, Canada en het Verenigd Koninkrijk.

Deze database is een stuk groter dan onze webshop. Hieronder zie je de structuur:

![Schema van de AdventureWorks-database](../../figures/adventure_works.png)

Er zijn twee belangrijke verschillen met de webshop:

- De tabel- en kolomnamen zijn **Engels**, zoals in de meeste echte bedrijven. Klanten vind je in `Customer`, producten in `Product`.
- Er zijn **meer tabellen**, die op meer manieren met elkaar verbonden zijn. Voorlopig heb je alleen `Customer` en `Product` nodig; de andere tabellen komen in de volgende toepassingen aan bod.

:::{Tip}
Weet je niet meer welke kolommen een tabel heeft? Klik op de **Schema**-knop naast een oefening. Zo zie je alle tabellen met hun kolommen, en kan je zelfs een voorbeeld van de data bekijken.
:::

## Oefeningen


```sql
--- 1. Toon de naam (Name) en de catalogusprijs (ListPrice) van alle producten.

--- 2. Toon de voornaam, achternaam en bedrijfsnaam (CompanyName) van alle klanten. Sorteer op achternaam en daarna op voornaam.

--- 3. Toon alle producten die meer dan 1000 euro kosten. Sorteer van duur naar goedkoop.

--- 4. Toon de naam, kleur en prijs van alle rode producten (Color = 'Red'). Sorteer ze dalend op prijs.

--- 5. Toon de vijf duurste producten.

--- 6. Toon alle unieke kleuren die voorkomen in de tabel Product.

--- 7. Elk product heeft een kostprijs (StandardCost) en een verkoopprijs (ListPrice). Toon per product de winstmarge als een berekende kolom met de naam winstmarge. Toon de tien producten met de grootste marge.
```
