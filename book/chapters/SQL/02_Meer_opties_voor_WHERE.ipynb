Met **WHERE** kunnen we gegevens filteren. Dat is essentieel om een goed zicht te krijgen op de data die we hebben. In het vorige hoofdstuk zag je al enkele eenvoudige queries met **WHERE**. Nu leer je hoe je nog beter kan filteren.

## 1. Speciale waarden: NULL (onbekend/ontbreekt)
Als een veld ontbreekt, gebruiken databases een speciale waarde: NULL of null.
NULL is geen lege tekst en ook niet 0. Je vergelijkt het niet met = maar met IS NULL of IS NOT NULL.

Kijk even naar de data van de customers tabel. (via schema -> Data -> table customers -> preview)

Welke waarde is soms null? Voeg dan de query uit. Zie je hoe je nu alle gebruikers gebruikers ziet zonder e-mail adres?

Je kan dit ook omkeren. Wijzig de laatste regel in `WHERE email IS NOT NULL`. Wat zie je nu?

````python
SELECT first_name, last_name, email
FROM customers
WHERE email IS NULL;
````

## 2. Patronen zoeken met LIKE
Met `LIKE` kan je patronen zoeken. Vaak wil je niet enkel filteren op een exacte waarde, maar op een deel van die waarde. Bij like heb je daarom jokertekens. Daarmee geef je aan "hier mag eender wat staan".

Er zijn twee jokertekens:
- \% = een willekeurige tekenreeks, of zelfs niets
- _ = precies één willekeurig teken.

Hieronder zie je een voorbeeld van beide opties. Voorspel de uitvoer voordat je de queries uitvoert. Wissel de comments om het tweede voorbeeld uit te voeren.

````python
-- Productnamen die 'Pro' bevatten
SELECT name
FROM products
WHERE name LIKE '%Pro%';

-- Postcodes die starten met '20'
-- SELECT first_name, last_name, postal_code
-- FROM customers
-- WHERE postal_code LIKE '20__';  -- '20' + exact twee tekens
````

:::{Tip}
Soms ben je niet zeker of een item met hoofdletters geschreven is of niet. Wel je zeker zijn dat je alle waarden krijgt die je zoekt, ongeacht hoofdletters, dan je beide kanten omzetten naar kleine letters. Dat doe je zo: `WHERE LOWER(name) LIKE LOWER('%Pro%');`
:::

## 3. IN, NOT IN
Via `WHERE` kan je eenvoudig een categorie selecteren. Maar wat als je wil filteren op meerdere categorieën? Dan gebruik je `IN`, of `NOT IN` als je bepaalde categorieën niet wil zien.

Bekijk het volgende voorbeeld en draai de selectie ook om via `NOT`.

````python
-- Producten in bepaalde categorieën
SELECT name, category, unit_price
FROM products
WHERE category IN ('Gaming', 'Audio', 'Displays');
````

## 4. BETWEEN
Als je wil weten welke data er tussen een bepaald bereik ligt, dan kan `AND` gebruiken:

```SQL
-- Producten in bepaalde categorieën
SELECT name, category, unit_price
FROM products
WHERE unit_price > 10 AND unit_price < 100;
```

Maar je hebt ook een andere optie via `BETWEEN`. Het resultaat is hetzelfde, maar het is minder typwerk en het leest eenvoudiger.

````python
-- Producten in bepaalde categorieën
SELECT name, category, unit_price
FROM products
WHERE unit_price BETWEEN 10 AND 100;
````

## Datum filteren

In een database slaan we datums altijd op een vaste manier op: YYYY-MM-DD. Dat betekent dat het jaartal altijd uit 4 cijfers bestaat. (Wat we in het jaar 10000 gaan doen is nog niet bekend.)

De maand zal steeds twee cijfers lang zijn. Januari is dus niet `1`, maar `01`. Hetzelfde geldt voor de dag.

Hierdoor kunnen we eenvoudig filteren op een deel van de datum. Je kan een datum ook vergelijken. Vervang in de query de `WHERE` regel door:

``` SQL
WHERE order_date >= '2025-04-01';
```

Daarmee toon je alle bestellingen vanaf 1 april.

````python
-- Alle bestellingen in juli 2025
SELECT order_id, order_date, status
FROM orders
WHERE order_date LIKE '2025-07-%'
ORDER BY order_date;
````

:::{Tip}
Je hebt nu de meest voorkomende `SELECT` queries gezien. Als je je voorbereidt op een test, denk er dan aan dat de volgorde steeds dezelfde is:

`SELECT - FROM - WHERE - ORDER BY - LIMIT`
:::
