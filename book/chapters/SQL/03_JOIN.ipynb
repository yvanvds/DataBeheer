:::{admonition} Leerdoelen

In deze les leer je hoe je meerdere tabellen kan combineren in één query.

Na deze les kan je:
- JOIN, LEFT JOIN en INNER JOIN  gebruiken
- gegevens uit twee of meer tabellen combineren

## 1. Waarom JOIN?
Tot nu toe werkten we met één tabel tegelijk.
Maar in een relationele databank zijn tabellen met elkaar verbonden via sleutels.

De afspraak is dat we de naam van een sleutelveld eindigen met `_id`. Het eerste veld van een tabel is is de primary key (PK). Die hoort voor elke rij uniek te zijn. Dat zie je bijvoorbeeld in de tabel `customers`. 

Maar vaak wil je ook verwijzen naar andere tabellen. Een order wordt geplaatst door een customer. We gaan de gegevens van de customer niet opnieuw in order plaatsen, want dat zou dubbele informatie zijn. Als we bijvoorbeeld het adres van een customer willen wijzigen, dan willen we dat op één plaats kunnen doen, niet bij elke order van die customer.

Daarom verwijzen we in de `orders` tabel naar de customer via zijn `id`. Dat is in deze database `customer_id`. Dit noemen we een foreign key (FK). De key hoort dus niet bij de tabel, maar verwijst naar een andere tabel.

Als we nu de naam van de klant én zijn bestellingen willen tonen, dan hebben we gegevens uit twee tabellen nodig. Dat kan als we een JOIN gebruiken.

## 2. INNER JOIN

Een inner join toont enkel records waar in **beide** tabellen een match bestaat.

| customers   	| orders      	|
|-------------	|-------------	|
| customer_id 	| order_id    	|
| first_name  	| customer_id 	|
| last_name   	| order_date  	|

Het veldnamen kunnen in meer dan één tabel voorkomen. Daarom willen bij `JOIN` altijd aanduiden welke tabel we bedoelen. We schrijven dus niet `first_name` maar `customers.first_name`. Zo verwijzen we steeds naar het correcte veld.

In het voorbeeld hieronder combineren we klanten en hun bestellingen. Omdat het een `INNER JOIN` is, zie je enkel klanten mét een bestelling. Klanten zonder bestelling worden niet getoond.

````python
SELECT customers.first_name, customers.last_name, orders.order_id, orders.order_date
FROM customers
INNER JOIN orders
    ON customers.customer_id = orders.customer_id;
````

:::{Tip}
Telkens de naam van de tabel uitschrijven is wel veel werk. Je kan het ook korter doen door de tabelnamen in te korten. Hieronder zie je een voorbeeld. In de rest van dit hoofdstuk gebruiken we deze schrijfwijze, maar het is niet fout om alles voluit te schrijven.
:::

``` SQL
SELECT c.first_name, c.last_name, o.order_id, o.order_date
FROM customers c
INNER JOIN orders o
    ON c.customer_id = o.customer_id;
```

### Meerdere JOIN's

Je hoeft het ook niet te laten bij twee tabellen. Je kan ook meer tabellen combineren, zolang er maar een verwante ID bestaat. Bekijk even dit voorbeeld. Uit welke tabellen komen deze gegevens? En hoe combineren we die?

````python
SELECT o.order_id, c.first_name, p.name, ol.quantity
FROM orders o
INNER JOIN customers c
    ON o.customer_id = c.customer_id
INNER JOIN order_lines ol
    ON o.order_id = ol.order_id
INNER JOIN products p
    ON ol.product_id = p.product_id;
````

## 3. LEFT JOIN

Een `INNER JOIN` toont enkel rijen waarvoor er een match bestaat in beide tabellen. Dat is niet altijd wat je wil. Daarom is er de `LEFT JOIN`. Die toont alle records uit de linkertabel. Dat is de tabel die bij `FROM` staat.

Wat zie je hier wanneer een klant geen bestellingen heeft geplaatst? Probeer de query ook eens aan te passen. Hoe sorteer je op familienaam?

````python
SELECT c.first_name, c.last_name, o.order_id
FROM customers c
LEFT JOIN orders o
    ON c.customer_id = o.customer_id;
````
