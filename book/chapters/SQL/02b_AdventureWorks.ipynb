# Toepassing 2: filteren in AdventureWorks

Tijd om je nieuwe filtertechnieken los te laten op de database van Adventure Works Cycles. In een echte database zoals deze zijn er altijd onvolledige gegevens: producten zonder kleur, klanten zonder aanspreektitel. Precies daarvoor heb je `IS NULL`, `LIKE`, `IN` en `BETWEEN` nodig.

:::{Tip}
Gebruik de **Schema**-knop om de kolomnamen op te zoeken. De datums in deze database bevatten ook een tijdstip (bv. `2007-07-01 00:00:00.000`), maar omdat een datum als tekst begint met `YYYY-MM-DD` kan je er gewoon mee vergelijken en filteren zoals je dat geleerd hebt.
:::

## Oefeningen


```sql
--- 1. Sommige producten hebben geen kleur ingevuld. Toon de naam en de kleur van deze producten.

--- 2. Toon de voornaam, achternaam en aanspreektitel (Title) van de klanten zonder aanspreektitel.

--- 3. Toon alle producten waarvan de naam 'Mountain' bevat. Sorteer op ProductNumber.

--- 4. Toon alle klanten waarvan de achternaam begint met 'S'. Sorteer op achternaam en voornaam.

--- 5. Toon de stad en het land van alle adressen in Canada of het Verenigd Koninkrijk ('Canada' of 'United Kingdom').

--- 6. Toon alle producten met een prijs tussen 100 en 500 euro. Gebruik BETWEEN en sorteer oplopend op prijs.

--- 7. Elk product heeft een datum waarop de verkoop startte (SellStartDate). Toon de producten die pas vanaf 2007 verkocht worden.

--- 8. Sommige producten zijn uit het gamma genomen: hun SellEndDate is ingevuld. Toon de naam en de einddatum van deze producten.

--- 9. De boekhouding wil de bestellingen met een hoge verzendkost controleren. Toon de bestellingen (SalesOrderID, Freight, TotalDue) waarvan de verzendkost (Freight) boven 50 euro ligt. Toon de hoogste verzendkost eerst.
```
