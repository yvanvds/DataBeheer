# Toepassing 2: filteren in AdventureWorks

In deze toepassing werk je als **data-analist** bij Adventure Works Cycles. De collega's van de andere afdelingen kennen geen SQL — wie iets uit de database nodig heeft, komt bij jou. Elke oefening hieronder is een vraag van een afdeling, en jouw queryresultaat is het antwoord waarmee zij een beslissing nemen.

Je nieuwe filtertechnieken komen daarbij goed van pas. In een echte database zoals deze zijn er altijd onvolledige gegevens: producten zonder kleur, klanten zonder aanspreektitel. Precies daarvoor heb je `IS NULL`, `LIKE`, `IN` en `BETWEEN` nodig.

:::{Tip}
Gebruik de **Schema**-knop om de kolomnamen op te zoeken. De datums in deze database bevatten ook een tijdstip (bv. `2007-07-01 00:00:00.000`), maar omdat een datum als tekst begint met `YYYY-MM-DD` kan je er gewoon mee vergelijken en filteren zoals je dat geleerd hebt.
:::

## Oefeningen


```sql
--- 1. De webshopbeheerder krijgt klachten over productpagina's zonder kleurvermelding. Toon de naam en de kleur van de producten waarvan de kleur niet is ingevuld, dan kan hij ze aanvullen.

--- 2. De marketingafdeling bereidt een mailing voor en wil elke klant correct aanspreken. Toon de voornaam, achternaam en aanspreektitel (Title) van de klanten zonder aanspreektitel: die moeten eerst opgezocht worden.

--- 3. De verkoopafdeling stelt een folder samen met het volledige Mountain-gamma. Toon alle producten waarvan de naam 'Mountain' bevat. Sorteer op ProductNumber, dan staat de folder meteen in de juiste volgorde.

--- 4. De klantendienst verdeelt het klantenbestand onder de medewerkers. Jouw collega volgt voortaan alle klanten op waarvan de achternaam begint met 'S'. Maak haar lijst en sorteer op achternaam en voornaam.

--- 5. De logistieke dienst onderzoekt of een eigen transportronde buiten de VS de moeite loont. Toon de stad en het land van alle adressen in Canada of het Verenigd Koninkrijk ('Canada' of 'United Kingdom').

--- 6. De marketingafdeling plant een kortingsactie in het middensegment. Toon alle producten met een prijs tussen 100 en 500 euro. Gebruik BETWEEN en sorteer oplopend op prijs.

--- 7. Het management wil weten hoe recent het aanbod nog vernieuwd is. Elk product heeft een datum waarop de verkoop startte (SellStartDate): toon de producten die pas vanaf 2007 verkocht worden.

--- 8. De aankoopdienst ruimt de catalogus op. Producten die uit het gamma zijn genomen, hebben een ingevulde SellEndDate. Toon de naam en de einddatum van deze producten, dan kunnen ze van de website gehaald worden.

--- 9. De boekhouding wil de bestellingen met een hoge verzendkost controleren. Toon de bestellingen (SalesOrderID, Freight, TotalDue) waarvan de verzendkost (Freight) boven 50 euro ligt. Toon de hoogste verzendkost eerst.
```
