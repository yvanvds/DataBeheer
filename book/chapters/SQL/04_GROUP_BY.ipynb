# 4. Groeperen

:::{admonition} Leerdoelen
:class: tip

In deze les leer je hoe je gegevens uit meerdere tabellen kan samenvatten.

Na deze les kan je:
- GROUP BY gebruiken
- Aggregatiefuncties gebruiken
:::

In de vorige les leerden we JOIN gebruiken om gegevens uit meerdere tabellen te combineren.

Voorbeeld: toon elke klant met zijn bestellingen.

```SQL
SELECT customers.first_name,
       customers.last_name,
       orders.order_id,
       orders.order_date
FROM customers
JOIN orders ON customers.customer_id = orders.customer_id
ORDER BY last_name, order_date;
```
We kunnen nu dus gegevens uit meerdere tabellen samenvoegen. Maar wat als we samenvattingen willen, zoals:
- Hoeveel bestellingen plaatste elke klant?
- Wat is de totale omzet per klant?
- Wat is de gemiddelde prijs per categorie?

In SQL gebruiken we daarvoor **aggregatiefuncties** samen met `GROUP BY`.

## 1. Aggregatiefuncties

Aggregatiefuncties berekenen één waarde op basis van meerdere rijen.

| Functie 	| Beschrijving          	| Voorbeeld                   	|
|---------	|-----------------------	|-----------------------------	|
| COUNT() 	| telt het aantal rijen 	| COUNT(*) of COUNT(order_id) 	|
| SUM()   	| som van de waarden    	| SUM(total)                  	|
| AVG()   	| gemiddelde            	| AVG(unit_price)             	|
| MIN()   	| kleinste waarde       	| MIN(unit_price)             	|
| MAX()   	| grootste waarde       	| MAX(unit_price)             	|

Een eenvoudig voorbeeld zie je hieronder.

````sql
SELECT COUNT(*) AS aantal_orders
FROM orders;
````

## 2. GROUP BY

Met GROUP BY kan je gegevens per groep samenvatten. Bijvoorbeeld het aantal bestellingen per klant.

Het voorbeeld hieronder toont hoe het niet moet. Je krijgt slechts één customer ID te zien, en het aantal bestellingen is het totaal van alle bestellingen. Dat komt omdat we met COUNT(*) alles samentellen. Dat gaat niet samen met het tonen van alle customer ID's.

Omdat dit op te lossen voeg je een derde regel toe: `GROUP BY customer_id;`

Kijk hoe dat het resultaat wijzigt.

````sql
SELECT customer_id, COUNT(*) AS aantal_bestellingen
FROM orders;
````

Met `GROUP BY` vormt elke klant een groep. SQL berekent dus het aantal rijen per groep. Sommige klanten bestelden maar één keer, andere tientallen keren. Dat verschil zie je nu meteen in het resultaat.
